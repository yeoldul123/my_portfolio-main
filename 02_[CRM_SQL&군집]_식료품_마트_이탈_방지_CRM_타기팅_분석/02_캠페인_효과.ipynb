# 02. 지난 캠페인 효과

지속 감소 가구를 되돌릴 수단을 정하기 전에, 이 마트가 실제로 했던 캠페인이 효과가 있었는지 봄. 여기서 얻은 착안점을 03 노트북(해결 방안)에서 01 노트북의 원인과 함께 씀.

> 이 노트북은 교육 과정 강사님의 라이브러리(hossam)를 바탕으로 직접 만든 분석 모듈(`helpers` 패키지의 `my_*`)을 사용함.
> 코드 작성·디버깅과 문장 교정에 생성형 AI(Claude Code)를 보조 도구로 사용함.

## 1. 왜 캠페인을 보나

01 노트북의 결론은 '지속 감소 가구는 장보기의 일부를 다른 곳으로 옮겼다'. 그래서 해결 방안의 방향은 다른 곳에서 사는 몫을 이 마트로 되돌리는 것(03 노트북 2절). 이 마트가 가구를 골라 혜택을 준 기록이 남아 있는 것은 캠페인(쿠폰 묶음 발송)뿐이라, 이 기록으로 세 가지를 봄.

1. 캠페인은 누구에게 나갔나
2. 대전제(같은 가구가 계속 오고 꾸준히 사서 구매액이 유지되거나 늘어나는 것) 관점에서 효과가 있었나, 있었다면 얼마나
3. 지속 감소 가구처럼 줄어드는 중인 가구에도 통했나

## 2. 캠페인 자료 (00 노트북 1.7절)

- 캠페인 30개(A형 5, B형 19, C형 6). 첫 캠페인은 224일째에 시작했고, 길이는 32∼161일
- 받은 가구는 1,584가구, 가구와 캠페인의 쌍으로는 7,208쌍. 한 가구가 평균 4.6개를 받음
- 쿠폰을 나눠 준 방식
  - A형: 받은 가구마다 그 가구의 이전 구매를 바탕으로 고른 16장. 어떤 16장인지는 데이터에 없음
  - B·C형: 받은 가구 모두 그 캠페인의 쿠폰 전부
- 받은 캠페인에서 쿠폰을 한 장이라도 쓴 비율: A형 16.0%, B형 7.9%, C형 7.7%
- 어떤 가구에 보냈는지 선정 기준은 안내서에 없음. 다만 첫 캠페인 전 구매액이 가장 적던 10% 가구는 11%만, 가장 많던 20% 가구는 97%가 받았음. 원래 많이 사던 가구에 몰려 나감

## 3. 분석 설계

### 3.1 효과를 무엇으로 재나

쿠폰을 쓴 비율이 아니라, 받은 가구가 대전제에 맞게 바뀌었는지로 잼. 대전제는 '같은 가구들이 계속 오고 꾸준히 사서 가구들의 구매액이 유지되거나 늘어나는 것'(00 노트북 3.5절). 늘어야 하는 것은 구매액이고, 방문은 '계속 오는 것', 곧 줄지 않으면 됨. 더 자주 오는 것까지 요구하지는 않음.

그래서 캠페인 효과는 '구매액을 늘렸고, 방문 수는 줄이지 않았나'로 봄. 쿠폰만 쓰고 원래대로 돌아갔다면 효과로 보지 않음.

- 구매액은 '방문 수 × 장바구니당 구매액'이라, 두 값으로 나눠 캠페인이 어느 쪽을 움직였는지 봄. 예를 들어 4주당 구매액이 10% 늘었다면, 더 자주 와서인지 올 때 더 담아서인지를 가름
  - 판정: 4주당 구매액이 늘었나, 4주당 방문 수(장바구니 수)가 줄지 않았나. 더 자주 와서 늘든 올 때 더 담아서 늘든 구매액이 늘고 계속 오면 대전제에 맞음. 4주당 구매액은 그 기간에 한 번도 안 온 가구도 0으로 들어가, 받은 가구 모두를 잴 수 있음
  - 주 지표는 4주당 방문 수. 지속 감소 가구가 줄인 구매액의 90%가 방문 횟수에서 왔기 때문(00 노트북 9.5절). 결과는 방문 수를 먼저 보여 주고, 우리 대상에도 통할지(3.3)는 방문 수로 판정함
  - 장바구니당 구매액은 판정에 쓰지 않고, 구매액 효과가 방문과 장바구니 가운데 어디서 왔는지 나눠 보는 데 씀. 그 기간에 한 번이라도 온 가구만 잴 수 있어(끝난 뒤 12주에 한 번도 안 온 가구는 빠짐), 받은 가구 모두를 잴 수 없기 때문
  - 8절에서 효과를 금액으로 바꿀 때도 4주당 구매액을 씀
- 세 기간으로 나눔. 12주는 01 노트북에서 '낮아진 상태가 석 달은 이어져야 뚝 떨어졌다'고 본 것과 같은 길이

| 기간 | 범위 | 보는 것 |
|---|---|---|
| 받기 전 | 캠페인 시작일 바로 앞 12주 | 출발점 |
| 캠페인 중 | 시작일부터 끝나는 날까지 | 쿠폰이 있는 동안 늘었나 |
| 끝난 뒤 | 끝난 다음 날부터 12주 | 쿠폰이 없어진 뒤에도 남았나 (대전제 관점의 효과) |

판정의 중심은 '받기 전 12주 → 끝난 뒤 12주'의 변화. 캠페인 중 기간은 쿠폰이 있는 동안만 늘었는지(반짝 효과) 가를 때 씀.

예를 들어 받은 가구 A가 받기 전 4주에 4번 오다가, 캠페인 중 6번, 끝난 뒤 4번 왔다면 쿠폰이 있는 동안만 반짝 늘어난 것. 끝난 뒤에도 5번이면 1번 늘어난 만큼이 남은 것.

- 받기 전 기간의 길이는 두 가지 사이에서 고름. 너무 짧으면 출발점이 흔들림(4주만 보면 2주에 한 번 오는 가구는 두 번뿐이라, 한 번만 빠져도 값이 절반이 됨). 너무 길면 앞선 다른 캠페인의 영향이 출발점에 섞임(한 가구가 평균 4.6개를 받음). 끝난 뒤와 같은 12주로 맞춘 것은 편의이고, 4주당 값으로 바꿔 견주므로 길이가 달라도 비교는 성립함
- 끝난 뒤 12주를 볼 수 있는 캠페인만 씀. 데이터는 711일째에서 끝나, 627일째 뒤에 끝난 캠페인은 끝난 뒤를 다 볼 수 없음. 남는 캠페인은 21개(A형 4, B형 13, C형 4)
- 나머지 9개(A형 18번, B형 19·21·22·23·24·25번, C형 15·20번)는 판정에 쓰지 않고, 캠페인 중 효과만 7절에 참고로 따로 보여 줌. 쿠폰이 있는 동안 늘었는지는 볼 수 있지만, 대전제의 '꾸준히'는 확인할 수 없음. 24번은 데이터가 끝난 뒤(719일째)에 끝나 캠페인 중 기간도 다 볼 수 없어 이마저 뺌
- 받기 전 12주는 21개 모두 분석 기간(18주, 118일째부터) 안에 들어감. 가장 일찍 시작한 26번도 받기 전 12주가 140일째부터. 그 전은 가구들이 아직 관측에 들어오던 기간이라 00 노트북에서 뺐음

### 3.2 무엇과 견주나

받은 가구가 늘었다는 것만으로는 효과라고 할 수 없음. 같은 때에 마트 전체가 늘었을 수도 있고, 원래 많이 사던 가구라 다르게 움직였을 수도 있기 때문. 그래서 받은 가구의 변화에서, 받기 전 구매가 비슷했던 받지 않은 가구의 변화를 뺌(이중차분).

이중차분은 '차이'를 두 번 빼서 캠페인 효과만 남기는 방법. 예를 들어 받은 가구가 받기 전 4주에 4번 오다가 끝난 뒤 5번 왔다면 1번 늘었음. 하지만 같은 때에 계절이나 마트 전체의 변화로 모든 가구가 조금씩 더 왔을 수 있어, 이 1번이 모두 캠페인 덕분은 아님. 그래서 받기 전 구매가 비슷한 받지 않은 가구도 같은 때에 얼마나 바뀌었는지 봄.

| | 받기 전 | 끝난 뒤 | 변화 (첫 번째 빼기) |
|---|---|---|---|
| 받은 가구 | 4번 | 5번 | +1번 |
| 받지 않은 가구 | 4번 | 4.5번 | +0.5번 |
| 캠페인 효과 (두 번째 빼기) | | | +1 − 0.5 = +0.5번 |

- 첫 번째 빼기(끝난 뒤 − 받기 전): 묶음마다 그 기간에 얼마나 바뀌었나
- 두 번째 빼기(받은 가구의 변화 − 받지 않은 가구의 변화): 받지 않은 가구도 0.5번 늘었으므로, 그만큼은 캠페인과 상관없이 모두에게 생긴 변화로 보고 뺌. 남은 0.5번이 캠페인 효과
- 끝난 뒤 값만 견주지 않는 이유: 캠페인은 원래 많이 사던 가구에 갔으므로, 받은 가구는 원래부터 더 자주 왔을 수 있음. 묶음마다 '변화'를 견주면 원래 수준의 차이는 빠짐

- 비교 가구: 그 캠페인을 받지 않았고, 받기 전 12주부터 끝난 뒤 12주까지 다른 캠페인도 받지 않은 가구. 다른 캠페인을 받고 있던 가구와 견주면 효과가 작게 나옴
- 받기 전 구매가 비슷한 가구끼리 견줌. 캠페인마다 받기 전 12주 구매액으로 모든 가구를 다섯 단계로 나누고, 같은 단계 안에서만 받은 가구와 비교 가구를 견준 뒤, 받은 가구 수로 가중 평균함
  - 가장 많이 사는 단계는 거의 모두 받아 비교 가구가 거의 없을 수 있음. 비교 가구가 너무 적은 단계(아래 6절에서 결과 보기 전에 정함)는 빼고, 결과가 어느 단계 가구에 대한 것인지 밝힘
- 받은 가구 쪽에서 캠페인 기간이 다른 캠페인과 겹치는 것은 그대로 둠. 한 가구가 평균 4.6개를 받아 겹치지 않는 경우만 고르면 남는 가구가 너무 적음. 그래서 효과는 '그 캠페인을 받은 시기'의 효과로 읽음
- 이중차분은 '받지 않았다면 받은 가구도 비교 가구와 같은 만큼 바뀌었을 것'이라는 가정에 기댐. 이 가정이 맞는지 보려고, 받기 전 12주를 6주씩 둘로 나눠 두 묶음이 받기 전에도 비슷하게 움직였는지 먼저 확인함
  - 출발점이 같아도 움직이던 방향은 다를 수 있음. 예를 들어 받기 전 12주의 4주당 방문이 두 묶음 모두 4번이어도, 받은 가구는 앞 6주 3번에서 뒤 6주 5번으로 늘어나는 중이었고 받지 않은 가구는 내내 4번이었다면, 받은 가구가 끝난 뒤 늘어난 몫은 캠페인 효과가 아니라 원래 늘던 흐름일 수 있음
  - 6주는 가구 하나로 보면 짧아 흔들리지만, 이 확인은 묶음 전체의 평균 변화를 견주는 것이라 가구가 많으면 흔들림이 줄어듦. 받기 전 기간을 더 길게 잡으면 앞선 다른 캠페인의 영향이 섞이고, 일찍 시작한 캠페인은 분석 기간 밖으로 넘어감

### 3.3 우리 대상에도 통할까

지속 감소 가구는 줄어드는 중인 가구. 원래 잘 오던 가구에 통한 캠페인이 줄어드는 가구에도 통한다는 보장은 없음.

- 받은 가구를 받기 직전 상태로 나눔: 받기 전 12주 가운데 뒤 6주의 구매액이 앞 6주보다 줄어들던 가구와 아닌 가구. 줄어들던 가구에서 효과가 따로 있었는지 봄
- 여기서는 방문 수로 판정함. 지속 감소 가구는 주로 오는 횟수가 줄었으므로, 올 때 더 담게 한 캠페인보다 더 자주 오게 한 캠페인이 맞음
- 01 노트북의 지속 감소 가구로 바로 나누지 않음. 지속 감소 판정에 캠페인 뒤의 구매도 들어 있어, 캠페인이 통하지 않은 가구가 지속 감소로 더 잡히는 순환이 생김. 지속 감소 가구가 몇 가구나 받았는지는 참고로만 셈
- A형은 가구마다 이전 구매를 바탕으로 쿠폰을 골라 줬음. 03 노트북의 방향(가구마다 끊긴 단골 품목을 되찾기)과 가장 닮은 방식이라, A형과 B·C형의 결과를 따로 봄

### 3.4 판정 기준 (결과를 보기 전에 정함)

효과마다 가구를 다시 뽑아 계산을 되풀이하는 방법(부트스트랩)으로 95% 신뢰구간을 구함. 한 가구가 여러 캠페인에 들어가므로, 가구 단위로 뽑음.

| 판정 | 기준 |
|---|---|
| 효과 있음 (대전제 관점) | 끝난 뒤 12주에 4주당 구매액이 늘었고(95% 신뢰구간 전체가 0보다 큼), 4주당 방문 수는 줄지 않음(신뢰구간 전체가 0보다 작지 않음) |
| 반짝 효과 | 캠페인 중에는 위 기준을 넘지만, 끝난 뒤에는 넘지 못함. 쿠폰이 있는 동안만 더 산 것이라 대전제로는 효과 없음 |
| 엇갈림 | 끝난 뒤 12주에 4주당 구매액은 늘었지만, 4주당 방문 수가 줄었음(신뢰구간 전체가 0보다 작음). 대전제의 '계속 오는 것'이 약해진 것이라 대전제로는 효과 없음. 오는 횟수가 줄어든 지속 감소 가구에는 오히려 맞지 않는 쪽 |
| 효과 없음 | 캠페인 중에도 효과 있음의 기준을 넘지 못함 |

- 엇갈림의 예: 4주에 4번 오던 가구가 받은 뒤 3번 오면서, 올 때 쿠폰 상품을 몰아 사서 구매액이 10% 늘어난 경우. 당장 매출은 늘었지만 마트에 오는 습관은 약해짐. 이런 캠페인 방식은 03 노트북에서 착안점이 아니라 피해야 할 방식의 근거로 씀
- '방문 수가 줄었다'는 신뢰구간 전체가 0보다 작을 때만 셈. 조금 줄어 보여도 신뢰구간이 0을 포함하면 '줄지 않음'

- 받기 전 추세 확인에서 두 묶음의 차이가 0과 구별되면(신뢰구간이 0을 포함하지 않으면), 비교 가구를 믿기 어려워 그 결과는 한 단계 낮춰 읽음
- 효과가 있으면 크기를 수치로 바꿈: 가구당 4주에 몇 번 더 왔나, 구매액이 얼마 늘었나, 받은 가구 전체로 얼마나 늘었나, 받은 가구가 쓴 쿠폰 할인 1달러당 구매액이 얼마 늘었나
- 대상 적용(3.3): 줄어들던 가구에서 끝난 뒤 12주의 방문 수 효과가 0보다 크면(신뢰구간 전체가 0보다 큼) 03 노트북에 착안점으로 씀. 신뢰구간이 0을 포함하면 '줄어드는 가구를 더 자주 오게 한다는 근거 없음'으로 적음

```python
# 장보기가 아닌 거래 제외 (01 노트북과 같은 기준)
df = trans.merge(product[["PRODUCT_ID", "DEPARTMENT", "COMMODITY_DESC"]], on="PRODUCT_ID", how="left")
excluded = df["DEPARTMENT"].isin(["KIOSK-GAS", "MISC SALES TRAN"]) | df["COMMODITY_DESC"].eq("COUPON/MISC ITEMS")
x = df[~excluded]

# 가구·날짜별 방문 수(장바구니 수)와 구매액
daily = x.groupby(["household_key", "DAY"]).agg(방문=("BASKET_ID", "nunique"), 구매액=("SALES_VALUE", "sum")).reset_index()

# 가구별 첫 구매일
first_day = x.groupby("household_key")["DAY"].min()
print(f"거래 {len(x):,}줄, 가구·날짜 {len(daily):,}줄, 가구 {daily['household_key'].nunique():,}")
```

```python
# 캠페인별 기간과 쓰임: 끝난 뒤 12주(84일)를 데이터 안에서 볼 수 있으면 판정, 아니면 참고 (데이터 끝난 뒤에 끝나면 제외)
LAST_DAY = x["DAY"].max()
camp["받기 전 시작"] = camp["START_DAY"] - 84
camp["끝난 뒤 끝"] = camp["END_DAY"] + 84
camp["쓰임"] = np.select([camp["끝난 뒤 끝"] <= LAST_DAY, camp["END_DAY"] <= LAST_DAY], ["판정", "참고 (캠페인 중만)"], default="제외")
print("데이터 마지막 날:", LAST_DAY, "/ 가장 이른 받기 전 시작:", camp["받기 전 시작"].min(), "일째")
pd.crosstab(camp["DESCRIPTION"], camp["쓰임"], margins=True, margins_name="합계")
```

표의 '받은 쌍'은 가구와 캠페인의 쌍 수, '받은 가구'는 그 쌍에 들어 있는 가구 수. 한 가구가 같은 유형의 캠페인을 여러 개 받으면 쌍은 여러 개, 가구는 하나로 셈.

## 5. 캠페인은 누구에게 나갔나

00 노트북 1.7절은 첫 캠페인 전 구매액으로 본 것이라, 캠페인마다 받기 직전의 모습은 보지 않았음. 여기서는 판정에 쓰는 21개 캠페인마다 받기 전 12주를 기준으로 세 가지를 봄.

1. 받은 가구와 받지 않은 가구의 받기 전 4주당 방문 수·구매액. 차이가 크면 그대로 견줄 수 없다는 뜻
2. 받기 전 12주 구매액 다섯 단계별로 받은 가구 비율. 6절에서 같은 단계 안에서 견줄 때, 단계마다 비교할 받지 않은 가구가 얼마나 남는지 미리 봄
3. 01 노트북의 지속 감소 가구가 몇 가구나 받았나. 받은 가구가 적으면 '우리 대상에도 통하나'(3.3절)의 근거가 약해짐. 지속 감소 판정에 캠페인 뒤의 구매도 들어 있어(3.3절), 여기서는 몇 가구인지만 셈

```python
# 기간(시작일~끝날, 양 끝 포함)의 가구별 4주당 방문 수·구매액. 기간에 거래가 없는 가구는 0
def window_rate(start, end, households):
    d = daily[daily["DAY"].between(start, end)].groupby("household_key")[["방문", "구매액"]].sum()
    return d.reindex(households, fill_value=0) * 28 / (end - start + 1)

# 판정 캠페인마다, 그때 관측 중인 모든 가구의 받기 전 12주 값과 받음 여부, 받기 전 구매액 다섯 단계
judge = camp[camp["쓰임"].eq("판정")]
received = set(zip(camp_hh["CAMPAIGN"], camp_hh["household_key"]))
rows = []
for _, c in judge.iterrows():
    hhs = first_day.index[first_day < c["받기 전 시작"]]
    w = window_rate(c["받기 전 시작"], c["START_DAY"] - 1, hhs)
    w["받음"] = [(c["CAMPAIGN"], h) in received for h in hhs]
    w["단계"] = pd.qcut(w["구매액"].rank(method="first"), 5, labels=["1단계", "2단계", "3단계", "4단계", "5단계"])
    rows.append(w.assign(CAMPAIGN=c["CAMPAIGN"], 유형=c["DESCRIPTION"]).reset_index())
pre = pd.concat(rows, ignore_index=True)

# 유형·받음별 쌍 수와 받기 전 4주당 값의 중앙값
pre.groupby(["유형", "받음"]).agg(쌍=("household_key", "size"), 방문=("방문", "median"), 구매액=("구매액", "median"))
```

'쌍'은 가구와 캠페인의 쌍 수(받지 않은 쪽은 그 캠페인 때 관측 중이던 나머지 가구), 값은 받기 전 12주의 4주당 방문 수·구매액 중앙값.

- 받은 가구는 받기 전에 4주에 5.3∼7.0번 와서 167∼263달러를 샀음. 받지 않은 가구는 2.0∼2.7번, 50∼74달러. 받은 가구가 원래 두세 배 더 자주 오고 세 배쯤 더 샀음
- 00 노트북 1.7절과 같이, 캠페인마다 받기 직전으로 봐도 많이 사던 가구에 나갔음. 받은 가구와 받지 않은 가구를 그대로 견줄 수 없음

```python
# 받기 전 구매액 단계별 받은 가구 비율(%) (1단계가 가장 적게 삼)
(pre.groupby(["유형", "단계"], observed=True)["받음"].mean() * 100).unstack("단계")
```

단계는 캠페인마다 그때 관측 중인 가구를 받기 전 12주 구매액으로 다섯 등분한 것. 값은 단계 안에서 받은 가구 비율(%).

- A형은 받기 전에 적게 사던 1단계는 3.3%, 많이 사던 5단계는 53.9%가 받았음. 위 단계로 갈수록 받은 비율이 꾸준히 올라감
- B형·C형은 캠페인 하나를 받은 가구가 적어, 5단계에서도 14.6%, 9.9%만 받았음. 방향은 A형과 같음
- 가장 많이 사는 5단계에서도 받지 않은 가구가 A형 46%, B형 85%, C형 90% 남음. 3.2절에서 걱정한 '비교할 가구가 없는 단계'는 다섯 단계로 나누는 한 생기지 않음. 다만 이 수에는 다른 캠페인을 받고 있던 가구도 들어 있어, 실제 비교 가구(그 기간에 어떤 캠페인도 받지 않은 가구)는 6절에서 다시 셈
- 같은 단계 안에서도 받은 가구가 더 많이 사는 쪽에 몰렸을 수 있음. 6절에서 단계 안의 받기 전 값도 함께 견줌

#### 지속 감소 가구는 몇 가구나 받았나

01 노트북 5절과 같은 기준으로 지속 감소 / 뚜렷한 추세 없음 / 지속 증가 유형을 다시 만듦(84주를 4주씩 21구간으로 묶고 계절분을 뺀 뒤 가구마다 맨-켄달 추세 검정, 유의수준 5%).

```python
# 01 노트북 5절과 같은 기준: 18~101주, 1·2분기(18~43주)에 구매가 있던 가구
xq = x[x["WEEK_NO"].between(18, 101)]
hh = xq[xq["WEEK_NO"] <= 43].groupby("household_key")["SALES_VALUE"].sum()
hh = pd.DataFrame(index=hh.index[hh > 0])

# 가구별 4주 구간 구매액 (21구간)과 계절분
week84 = (xq[xq["household_key"].isin(hh.index)]
          .pivot_table(index="household_key", columns="WEEK_NO", values="SALES_VALUE", aggfunc="sum", fill_value=0)
          .reindex(index=hh.index, columns=range(18, 102), fill_value=0))
block = week84.to_numpy().reshape(len(hh), 21, 4).sum(axis=2)
t_idx = np.arange(21)
block_mean = block.mean(axis=0)
slope, intercept = np.polyfit(t_idx, block_mean, 1)
season = block_mean / (slope * t_idx + intercept)

# 가구별 맨-켄달 추세 검정
type_order = ["지속 감소", "뚜렷한 추세 없음", "지속 증가"]
hh["유형"] = my_ts.trend_by_row(pd.DataFrame(block, index=hh.index), alpha=0.05, season=season,
                               labels=("지속 감소", "뚜렷한 추세 없음", "지속 증가"))["유형"]
print(hh["유형"].value_counts().reindex(type_order).to_dict())
```

- 지속 감소 370 / 뚜렷한 추세 없음 1,535 / 지속 증가 500가구로 01 노트북 5절과 같음

```python
# 유형별: 판정 캠페인을 하나라도 받은 가구 수·비율, 받은 쌍 수 (유형별)
jp = pairs[pairs["쓰임"].eq("판정")].merge(hh["유형"], left_on="household_key", right_index=True)
got = hh.assign(받음=hh.index.isin(jp["household_key"]))
summary5 = got.groupby("유형").agg(가구=("받음", "size"), 받은_가구=("받음", "sum")).reindex(type_order)
summary5["받은 비율 %"] = summary5["받은_가구"] / summary5["가구"] * 100
summary5.join(pd.crosstab(jp["유형"], jp["DESCRIPTION"]).add_suffix(" 쌍"))
```

'받은_가구'는 판정에 쓰는 21개 캠페인 가운데 하나라도 받은 가구, 'A형 쌍' 등은 그 유형에서 받은 가구·캠페인 쌍 수.

- 지속 감소 가구 370가구 가운데 272가구(73.5%)가 하나 이상 받았음. 뚜렷한 추세 없음(58.6%), 지속 증가(57.2%)보다 많이 받았음. 지속 감소 가구는 줄기 전 많이 사던 가구가 많아서로 보임(가구의 15.4%가 1·2분기 구매액의 22.1%를 차지, 01 노트북 1절)
- 받은 쌍은 A형 584, B형 297, C형 66쌍. 지속 감소 가구도 캠페인을 충분히 받아 '우리 대상에도 통하나'(3.3절)를 볼 수 있는 수. 다만 C형은 66쌍뿐이라 참고로 읽음
- 이 표는 몇 가구가 받았는지만 보여 줌. 지속 감소 판정에 캠페인 뒤의 구매가 들어 있어, 받은 지속 감소 가구의 효과로 캠페인을 판단하지 않음(3.3절)

## 6. 비교 가구 만들기와 받기 전 추세 확인

5절에서 본 대로 받은 가구는 원래 두세 배 더 자주 오던 가구라, 받지 않은 가구와 그대로 견줄 수 없음. 그래서 세 겹으로 맞춤.

1. 같은 단계 안에서만 견줌: 받기 전 12주 구매액이 비슷한 가구끼리 견줌. 가장 큰 수준 차이가 여기서 줄어듦
2. 수준이 아니라 변화를 견줌(이중차분, 3.2절): 같은 단계 안에서도 남는 수준 차이는, 묶음마다 '끝난 뒤 − 받기 전'을 구해 견주면 빠짐. 예를 들어 같은 단계 안에서 받은 가구가 4주에 6번, 비교 가구가 5.5번 오던 가구라도, 둘의 변화(+1번, +0.5번)를 견주면 원래 0.5번 차이는 결과에 들어가지 않음
3. 받기 전에도 같은 방향으로 움직였는지 확인: 수준 차이는 빠져도, 받기 전부터 받은 가구만 늘어나는 중이었다면 그 흐름이 효과처럼 남음. 그래서 받기 전 12주를 6주씩 나눠 두 묶음의 움직임을 먼저 견줌

#### 판단 기준 (결과를 보기 전에 정함)

- 비교 가구: 3.2절 정의대로, 받기 전 12주 시작부터 끝난 뒤 12주 끝까지와 기간이 겹치는 캠페인을 하나도 받지 않은 가구
- 단계는 5절과 같음(캠페인마다 그때 관측 중인 가구를 받기 전 12주 구매액으로 다섯 등분)
- 캠페인·단계마다 비교 가구가 30가구 미만이면 그 칸은 뺌. 평균이 몇 가구에 크게 흔들리지 않을 최소한으로 흔히 쓰는 수
- 같은 단계 안의 균형은 표준화 평균 차이로 잼. 두 묶음 평균의 차이를 두 묶음이 흩어진 정도(표준편차)로 나눈 값으로, 단위가 다른 방문 수와 구매액을 같은 자로 잴 수 있음. 예를 들어 받은 가구 평균 6번, 비교 가구 5.5번이고 표준편차가 2번이면 0.25. 0.1 미만이면 비슷하다고 보는 것이 흔한 기준(Austin, 2011)
  - 0.1 이상이면 수준 차이가 남은 것. 이중차분이 수준 차이를 빼 주므로 판정을 멈추지는 않지만, 수준이 다른 가구는 움직임도 다를 수 있어 아래 받기 전 추세 확인을 더 무겁게 봄
- 받기 전 추세: 앞 6주 → 뒤 6주의 4주당 방문 수와 구매액 변화를 이중차분으로 견줌. 같은 단계 안에서 견준 뒤 받은 가구 수로 가중 평균함
  - 95% 신뢰구간은 가구를 다시 뽑아 계산을 1,000번 되풀이해 구함(부트스트랩). 한 가구가 여러 캠페인에 들어가므로 가구 단위로 뽑음
  - 신뢰구간이 0을 포함하면 받기 전 움직임이 비슷했다고 보고 통과. 방문 수나 구매액 가운데 하나라도 0을 포함하지 않으면, 그 유형의 효과 판정은 한 단계 낮춰 읽음(3.4절)

```python
# 캠페인별 기간 (양 끝 포함): 받기 전 앞 6주, 뒤 6주, 받기 전 12주, 캠페인 중, 끝난 뒤 12주
def windows(c):
    s, e = c["START_DAY"], c["END_DAY"]
    return {"앞 6주": (s - 84, s - 43), "뒤 6주": (s - 42, s - 1), "받기 전": (s - 84, s - 1),
            "캠페인 중": (s, e), "끝난 뒤": (e + 1, e + 84)}

# 가구별로 받은 캠페인의 기간 (비교 가구 판정용, 30개 캠페인 모두)
got_period = camp_hh.merge(camp[["CAMPAIGN", "START_DAY", "END_DAY"]], on="CAMPAIGN")

# 판정 캠페인마다: 받은 가구(받음) / 비교 가구(겹치는 캠페인을 하나도 받지 않음)와 기간별 4주당 값
rows = []
for _, c in judge.iterrows():
    w = windows(c)
    lo, hi = w["받기 전"][0], w["끝난 뒤"][1]
    busy = set(got_period.loc[(got_period["START_DAY"] <= hi) & (got_period["END_DAY"] >= lo), "household_key"])
    base = pre[pre["CAMPAIGN"].eq(c["CAMPAIGN"])].set_index("household_key")[["받음", "단계"]]
    base = base[base["받음"] | ~base.index.isin(busy)]
    for name, (a, b) in w.items():
        base = base.join(window_rate(a, b, base.index).add_prefix(name + " "))
    rows.append(base.assign(CAMPAIGN=c["CAMPAIGN"], 유형=c["DESCRIPTION"]).reset_index())
panel_all = pd.concat(rows, ignore_index=True)
panel_all["묶음"] = np.where(panel_all["받음"], "받은 가구", "비교 가구")

# 캠페인·단계마다 비교 가구 수, 30가구 미만인 칸은 뺌
n_ctrl = panel_all[~panel_all["받음"]].groupby(["CAMPAIGN", "단계"], observed=True).size()
keep = n_ctrl.index[n_ctrl >= 30]
cell = pd.MultiIndex.from_frame(panel_all[["CAMPAIGN", "단계"]])
dropped = panel_all[~cell.isin(keep)]
panel = panel_all[cell.isin(keep)].copy()
print("뺀 칸의 받은 쌍:", dropped["받음"].sum(), "/ 남은 받은 쌍:", panel["받음"].sum())

# 유형·단계별 받은 쌍과 비교 가구 쌍
pd.crosstab([panel["유형"], panel["묶음"]], panel["단계"])
```

표는 남은 칸의 쌍 수(가구와 캠페인의 쌍). 한 가구가 여러 캠페인의 비교 가구가 될 수 있어 비교 가구도 쌍으로 셈.

- 비교 가구가 30가구 미만인 캠페인·단계 칸을 빼면서 받은 쌍 4,873쌍 가운데 1,459쌍(30%)이 빠지고 3,414쌍이 남음. 비교 가구는 앞뒤 기간에 겹치는 캠페인을 하나도 받지 않아야 해서, 캠페인을 자주 받는 가구가 많은 칸에서는 비교 가구가 모자람
- 남은 받은 쌍은 A형 2,022, B형 1,260, C형 132쌍. C형은 단계마다 받은 쌍이 8∼74쌍뿐이라 결과가 크게 흔들릴 수 있음
- 비교 가구는 아래 단계일수록 많고 위 단계일수록 적음. 5절과 같이 많이 사는 가구일수록 캠페인을 받고 있었기 때문

```python
# 같은 단계 안의 균형: 받기 전 12주 4주당 방문 수·구매액의 표준화 평균 차이 (유형·단계별, 캠페인 합쳐서)
def smd(d, col):
    a, b = d.loc[d["받음"], col], d.loc[~d["받음"], col]
    return (a.mean() - b.mean()) / np.sqrt((a.var() + b.var()) / 2)

bal = panel.groupby(["유형", "단계"], observed=True).apply(
    lambda d: pd.Series({"방문 받은": d.loc[d["받음"], "받기 전 방문"].mean(), "방문 비교": d.loc[~d["받음"], "받기 전 방문"].mean(),
                         "방문 차이": smd(d, "받기 전 방문"),
                         "구매액 받은": d.loc[d["받음"], "받기 전 구매액"].mean(), "구매액 비교": d.loc[~d["받음"], "받기 전 구매액"].mean(),
                         "구매액 차이": smd(d, "받기 전 구매액")}))
with pd.option_context("display.float_format", "{:,.2f}".format):
    display(bal)
```

'방문 받은'·'방문 비교'는 받기 전 12주 4주당 방문 수 평균, '방문 차이'는 표준화 평균 차이. 구매액도 같은 방식.

- 1∼4단계에서 표준화 평균 차이가 방문 0.23∼0.71, 구매액 0.24∼0.88로 기준(0.1)을 모두 넘음. 같은 단계 안에서도 받은 가구가 더 자주 오고 더 많이 샀음. 예를 들어 A형 3단계는 받은 가구 4.54번, 비교 가구 2.97번
- 5단계만 방문 차이가 −0.15∼0.09로 작음. 5단계는 위로 열려 있어 구매액은 여전히 받은 가구 쪽이 큼(0.03∼0.29)
- 판단 기준대로, 수준 차이가 남았으므로 아래 받기 전 추세 확인을 더 무겁게 봄

```python
# 이중차분 추정: 캠페인·단계 칸마다 (받은 가구 평균 변화 − 비교 가구 평균 변화), 받은 가구 수로 가중 평균
# 95% 신뢰구간은 가구 단위 부트스트랩 (한 가구가 여러 캠페인에 들어가 행 단위로 뽑으면 구간이 좁게 나옴).
# helpers의 block_bootstrap_ci는 시간 순서 묶음용이라 가구 단위로 뽑는 것은 직접 계산함
def did(d, before, after, n_boot=1000, seed=0):
    delta = (d[after] - d[before]).to_numpy()
    cell_id = d.groupby(["CAMPAIGN", "단계"], observed=True).ngroup().to_numpy()
    treat = d["받음"].to_numpy()
    hh_id, hh_n = pd.factorize(d["household_key"])[0], d["household_key"].nunique()
    n_cell = cell_id.max() + 1

    def estimate(w):
        nt = np.bincount(cell_id[treat], w[treat], n_cell); st = np.bincount(cell_id[treat], (w * delta)[treat], n_cell)
        nc = np.bincount(cell_id[~treat], w[~treat], n_cell); sc = np.bincount(cell_id[~treat], (w * delta)[~treat], n_cell)
        ok = (nt > 0) & (nc > 0)
        return np.sum(nt[ok] * (st[ok] / nt[ok] - sc[ok] / nc[ok])) / nt[ok].sum()

    rng = np.random.default_rng(seed)
    boot = [estimate(np.bincount(rng.integers(0, hh_n, hh_n), minlength=hh_n)[hh_id].astype(float)) for _ in range(n_boot)]
    lo, hi = np.percentile(boot, [2.5, 97.5])
    return pd.Series({"효과": estimate(np.ones(len(d))), "하한": lo, "상한": hi})

# 받기 전 추세 확인: 앞 6주 → 뒤 6주 변화의 이중차분
trend_rows = {}
for kind, d in panel.groupby("유형"):
    for col in ["방문", "구매액"]:
        trend_rows[(kind, col)] = did(d, f"앞 6주 {col}", f"뒤 6주 {col}")
pretrend = pd.DataFrame(trend_rows).T.rename_axis(["유형", "값"])
pretrend["판정"] = np.where((pretrend["하한"] <= 0) & (pretrend["상한"] >= 0), "통과", "어긋남")
with pd.option_context("display.float_format", "{:,.2f}".format):
    display(pretrend)
```

값은 받기 전 앞 6주 → 뒤 6주의 4주당 변화를 이중차분으로 견준 것(방문은 번, 구매액은 달러). 0이면 두 묶음이 받기 전에 같은 만큼 움직였다는 뜻.

- 여섯 가운데 다섯이 어긋남. 통과는 B형 방문 수 하나(신뢰구간 −0.61∼0.05)
- 모두 음수: 같은 단계 안에서 받은 가구가 비교 가구보다 받기 직전에 덜 오고 덜 사는 쪽으로 움직였음. A형은 6주 사이 4주당 방문이 0.68번, 구매액이 19.7달러 더 줄었음
- 판단 기준대로라면 세 유형 모두 효과 판정을 한 단계 낮춰 읽어야 함

```python
# 받기 전 앞 6주·뒤 6주의 4주당 방문 수 평균 (유형·묶음별): 어긋남이 어느 쪽 움직임에서 왔는지
mv = panel.groupby(["유형", "묶음"])[["앞 6주 방문", "뒤 6주 방문", "앞 6주 구매액", "뒤 6주 구매액"]].mean()
with pd.option_context("display.float_format", "{:,.2f}".format):
    display(mv)
```

값은 유형·묶음별 평균(4주당 방문은 번, 구매액은 달러). 캠페인·단계를 합친 단순 평균이라 위 이중차분 값과 크기는 다름.

- 받은 가구는 받기 직전에 조금씩 줄었음. A형 방문 5.61 → 5.43번, C형 6.03 → 5.30번
- 비교 가구는 거의 그대로이거나 조금 늘었음(A형 2.35 → 2.38번)
- 어긋남은 받은 가구 쪽의 움직임에서 왔음

#### 6절 1차 결과: 단계만 맞춘 비교는 쓸 수 없음

- 같은 단계 안에서도 받은 가구가 더 자주 오고 더 많이 사던 가구였음(표준화 평균 차이 대부분 0.2 이상)
- 받기 전 움직임도 달랐음. 받은 가구는 받기 직전에 줄어드는 중이었고, 비교 가구는 그대로였음
- 이대로 7절에서 효과를 구하면, 받기 전부터 받은 가구가 줄던 흐름이 효과에 섞임. 그 흐름이 이어졌다면 효과가 실제보다 작게 나오고, 받기 직전에 잠깐 줄었다가 원래대로 돌아온 것이라면 효과가 실제보다 크게 나옴. 어느 쪽인지 알 수 없어 한 단계 낮춰 읽는 것만으로는 부족함
- 그래서 비교 가구를 만드는 방법을 고친 뒤 다시 확인함

### 6.1 받기 전 흐름까지 닮은 가구와 짝짓기

받은 가구마다, 같은 캠페인 때의 비교 가구 가운데 받기 전 흐름이 가장 닮은 가구를 골라 짝을 지음. 받기 직전에 줄던 받은 가구는 똑같이 줄던 비교 가구와 견주게 됨. 그 감소가 잠깐 줄었다가 돌아오는 것이었다면 두 쪽 모두 돌아오므로, 효과에 섞이지 않음.

예를 들어 받은 가구 A가 받기 전 앞 6주에 4주당 6번, 뒤 6주에 5번 왔다면, 비교 가구 가운데 앞 6주 6번 안팎, 뒤 6주 5번 안팎인 가구를 고름. 앞 6주 3번, 뒤 6주 3번인 가구는 고르지 않음.

#### 짝짓는 방법 (결과를 보기 전에 정함)

- 닮음을 재는 값: 받기 전 앞 6주와 뒤 6주의 4주당 방문 수, 4주당 구매액(네 값). 흐름(앞 → 뒤)과 수준을 함께 맞춤
- 네 값은 로그(log(1 + 값))로 바꾼 뒤, 캠페인마다 표준편차로 나눠 같은 자로 맞춤. 구매액은 몇몇 가구가 아주 커서, 그대로 쓰면 많이 사는 가구 사이의 차이만 거리로 잡힘. 방문 수와 구매액은 단위가 달라 표준편차로 나눔
- 받은 가구마다 가장 가까운 비교 가구 3가구를 고름(같은 비교 가구가 여러 받은 가구의 짝이 될 수 있음). 1가구만 고르면 그 가구 사정에 크게 흔들리고, 많이 고를수록 덜 닮은 가구까지 들어옴. 1∼5가구 사이에서 흔히 쓰는 중간값
- 짝이 너무 멀면 뺌: 3가구까지의 평균 거리가 0.4를 넘으면 짝이 없는 것으로 봄. 네 값이 모두 표준편차의 0.2배씩 다를 때의 거리로, 짝짓기에서 흔히 쓰는 '표준편차의 0.2배' 기준(Austin, 2011)을 네 값에 맞춘 것
- 5절의 단계와 '비교 가구 30가구' 규칙은 짝짓기로 대신하므로 쓰지 않음

#### 판단 기준 (결과를 보기 전에 정함)

- 짝짓기가 잘 됐나: 유형마다 짝을 지은 받은 가구와 비교 가구의 표준화 평균 차이가 네 값과 받기 전 12주 방문 수·구매액에서 모두 0.1 미만이면 됨. 0.1 이상인 값이 남으면 그 유형의 효과 판정은 한 단계 낮춰 읽음
- 받기 전 흐름을 짝짓기에 썼으므로, 받기 전 추세는 더 이상 따로 검사할 수 없음. 대신 짝을 지은 뒤 앞 6주 → 뒤 6주 변화의 차이가 0에 가까운지 보여 주기만 함
- 효과의 신뢰구간: 받은 가구마다 (받은 가구의 변화 − 짝 3가구의 평균 변화)를 구하고, 받은 가구를 가구 단위로 다시 뽑아 1,000번 되풀이함. 짝은 고정해 두므로 비교 가구 쪽의 흔들림은 구간에 들어가지 않아, 구간이 조금 좁게 나올 수 있음

```python
# 받기 전 흐름으로 짝짓기: 캠페인마다 받은 가구 하나에 가장 가까운 비교 가구 3가구 (같은 비교 가구 재사용 가능)
# helpers에 짝짓기 함수가 없어 sklearn의 최근접 이웃을 직접 씀
from sklearn.neighbors import NearestNeighbors

match_cols = ["앞 6주 방문", "뒤 6주 방문", "앞 6주 구매액", "뒤 6주 구매액"]
K, CALIPER = 3, 0.4

def match_pairs(panel, caliper=CALIPER):
    pair_rows = []
    for camp_id, d in panel.groupby("CAMPAIGN"):
        z = np.log1p(d[match_cols])
        z = (z - z.mean()) / z.std()
        t, c = d[d["받음"]], d[~d["받음"]]
        dist, ind = NearestNeighbors(n_neighbors=K).fit(z.loc[c.index]).kneighbors(z.loc[t.index])
        for i, ti in enumerate(t.index):
            pair_rows.append({"t": ti, "c": list(c.index[ind[i]]), "거리": dist[i].mean()})
    out = pd.DataFrame(pair_rows)
    out["유형"] = panel.loc[out["t"], "유형"].to_numpy()
    out["짝 있음"] = out["거리"] <= caliper
    return out

pairs_m = match_pairs(panel_all)

# 유형별 받은 쌍과 짝이 있는 쌍
summary61 = pairs_m.groupby("유형").agg(받은_쌍=("t", "size"), 짝_있음=("짝 있음", "sum"))
summary61["짝 있음 %"] = summary61["짝_있음"] / summary61["받은_쌍"] * 100
summary61
```

'받은_쌍'은 판정 캠페인 21개의 받은 가구·캠페인 쌍 전체, '짝_있음'은 평균 거리 0.4 안에서 짝 3가구를 찾은 쌍.

- 짝을 찾은 쌍은 A형 1,977쌍(69.5%), B형 1,046쌍(61.0%), C형 150쌍(47.9%). 6절의 단계 비교(3,414쌍)보다 조금 적은 3,173쌍

```python
# 짝이 있는 받은 가구와 짝이 없어 빠진 받은 가구의 받기 전 12주 4주당 값 (중앙값): 어떤 가구가 빠졌나
tv = panel_all.loc[pairs_m["t"], ["받기 전 방문", "받기 전 구매액"]].reset_index(drop=True)
tv[["유형", "짝 있음"]] = pairs_m[["유형", "짝 있음"]]
tv.groupby(["유형", "짝 있음"]).agg(쌍=("받기 전 방문", "size"), 방문=("받기 전 방문", "median"), 구매액=("받기 전 구매액", "median"))
```

값은 받기 전 12주의 4주당 방문 수·구매액 중앙값.

- 짝이 없어 빠진 가구는 4주에 9∼10번 오던 가구. 짝이 있는 가구(4.7∼5.7번)의 두 배쯤. 이렇게 자주 오는 가구는 비교 가구 쪽에 닮은 가구가 거의 없음(5절, 많이 사는 가구일수록 캠페인을 받고 있었음)
- 그래서 아래 효과는 받은 가구 전체가 아니라, 4주에 5번 안팎 오던 가구에 대한 것. 가장 자주 오던 가구의 효과는 이 데이터로 알 수 없음

```python
# 짝이 있는 받은 쌍만: 받은 가구 값과 짝 3가구의 평균 값을 나란히 (짝의 평균은 값이 없는 가구를 빼고 냄)
def matched_values(panel, pairs, cols):
    pm = pairs[pairs["짝 있음"]].reset_index(drop=True)
    tr = panel.loc[pm["t"], ["household_key", "CAMPAIGN", "유형"] + cols].reset_index(drop=True)
    ct = pd.DataFrame([panel.loc[c, cols].mean() for c in pm["c"]]).reset_index(drop=True)
    ct_all = panel.loc[[i for c in pm["c"] for i in c], ["유형"] + cols]
    return tr, ct, ct_all

value_cols = [f"{w} {v}" for w in ["앞 6주", "뒤 6주", "받기 전", "캠페인 중", "끝난 뒤"] for v in ["방문", "구매액"]]
treated, control, control_all = matched_values(panel_all, pairs_m, value_cols)

# 짝을 지은 뒤 균형: 표준화 평균 차이 (받은 가구 대 짝으로 고른 비교 가구, 고른 횟수만큼 셈)
bal_cols = match_cols + ["받기 전 방문", "받기 전 구매액"]
bal61 = pd.DataFrame({kind: {col: (treated.loc[treated["유형"].eq(kind), col].mean() - control_all.loc[control_all["유형"].eq(kind), col].mean())
                                   / np.sqrt((treated.loc[treated["유형"].eq(kind), col].var() + control_all.loc[control_all["유형"].eq(kind), col].var()) / 2)
                             for col in bal_cols} for kind in ["TypeA", "TypeB", "TypeC"]}).T
with pd.option_context("display.float_format", "{:,.3f}".format):
    display(bal61)
```

값은 짝을 지은 뒤의 표준화 평균 차이. 0.1 미만이면 비슷하다고 봄.

- A형·B형은 여섯 값 모두 0.1 미만(0.004∼0.055). 6절의 단계 비교(0.2∼0.9)보다 크게 줄었음
- C형은 앞 6주 구매액만 0.122로 기준을 넘음. 판단 기준대로 C형 효과 판정은 한 단계 낮춰 읽음. C형은 짝이 있는 쌍도 150쌍뿐

```python
# 짝 이중차분: 받은 가구마다 (받은 가구 변화 − 짝 평균 변화), 받은 가구를 가구 단위로 다시 뽑아 95% 신뢰구간
# 값을 잴 수 없는 쌍(예: 장바구니당 구매액에서 그 기간에 오지 않은 가구)은 뺌
# helpers의 block_bootstrap_ci는 시간 순서 묶음용이라 가구 단위로 뽑는 것은 직접 계산함
def matched_did(tr, ct, kind, before, after, n_boot=1000, seed=0):
    m = tr["유형"].eq(kind).to_numpy()
    diff = ((tr[after] - tr[before]) - (ct[after] - ct[before])).to_numpy()[m]
    hh = tr.loc[m, "household_key"].to_numpy()
    ok = ~np.isnan(diff)
    diff, hh = diff[ok], hh[ok]
    hh_id = pd.factorize(hh)[0]
    n_hh = hh_id.max() + 1
    rng = np.random.default_rng(seed)
    boot = []
    for _ in range(n_boot):
        w = np.bincount(rng.integers(0, n_hh, n_hh), minlength=n_hh)[hh_id]
        boot.append(np.sum(w * diff) / w.sum())
    lo, hi = np.percentile(boot, [2.5, 97.5])
    return pd.Series({"쌍": len(diff), "효과": diff.mean(), "하한": lo, "상한": hi})

# 짝을 지은 뒤 받기 전 앞 6주 → 뒤 6주 변화의 차이 (보여 주기만 함)
pre61 = pd.DataFrame({(k, v): matched_did(treated, control, k, f"앞 6주 {v}", f"뒤 6주 {v}")
                      for k in ["TypeA", "TypeB", "TypeC"] for v in ["방문", "구매액"]}).T
with pd.option_context("display.float_format", "{:,.2f}".format):
    display(pre61.rename_axis(["유형", "값"]))
```

값은 짝을 지은 뒤 받기 전 앞 6주 → 뒤 6주 변화의 차이(받은 가구 − 짝 평균, 4주당 방문은 번, 구매액은 달러). 받기 전 흐름을 짝짓기에 썼으므로 검사가 아니라 얼마나 맞췄는지 보여 주는 값.

- 6절의 단계 비교에서 A형 방문 −0.68번이던 차이가 −0.07번으로 줄었음. 받은 가구의 4주당 방문(5번 안팎)에 견주면 1% 남짓
- 그래도 A형은 방문(−0.10∼−0.03)과 구매액(−9.1∼−3.8달러) 모두 신뢰구간이 0 아래에 있음. 짝짓기로도 다 맞추지 못한 작은 감소가 남았다는 뜻. 7절에서 A형 효과를 읽을 때 이만큼은 받기 전 흐름이 섞였을 수 있음을 함께 적음
- B형은 방문·구매액 모두 0을 포함함. C형은 구매액이 −11.9달러로 0 아래(위 균형에서도 C형만 기준을 넘음)

#### 6.1 정리

- 받기 전 흐름까지 닮은 가구와 짝을 지으니, A형·B형은 받은 가구와 비교 가구가 받기 전에 비슷해졌음(표준화 평균 차이 모두 0.1 미만)
- C형은 한 값이 기준을 넘고 짝도 150쌍뿐이라, 효과 판정을 한 단계 낮춰 읽음
- 효과는 4주에 5번 안팎 오던 받은 가구에 대한 것. 가장 자주 오던 가구는 닮은 비교 가구가 없어 빠졌음
- 7절은 이 짝으로 효과를 구함

## 7. 효과 추정

6.1절의 짝으로, 캠페인 유형마다 받기 전 12주에서 캠페인 중, 끝난 뒤 12주로 얼마나 바뀌었는지를 받은 가구와 짝 사이에서 견줌(이중차분). 값은 받은 가구 한 쌍당 4주에 몇 번 더 왔나, 몇 달러 더 샀나.

- 판정은 3.4절대로 4주당 구매액과 4주당 방문 수로 함
  - 효과 있음: 끝난 뒤 구매액 효과의 신뢰구간이 0보다 크고, 방문 효과의 신뢰구간이 0보다 작지 않음
  - 반짝 효과: 캠페인 중에는 위 기준을 넘지만, 끝난 뒤에는 넘지 못함
  - 엇갈림: 끝난 뒤 구매액은 늘었지만 방문이 줄었음
  - 효과 없음: 캠페인 중에도 위 기준을 넘지 못함
- 장바구니당 구매액은 판정에 쓰지 않고, 구매액 효과가 방문과 장바구니 가운데 어디서 왔는지 보는 데 씀. 받은 가구와 짝 가구 모두 두 기간에 한 번 이상 온 경우만 잼
- '한 단계 낮춰 읽음'(C형 균형, 6.1절)은 판정 이름 뒤에 '(약함)'을 붙이고, 03 노트북에서 단독 근거로 쓰지 않는다는 뜻으로 정함
- 효과의 크기를 받은 가구의 받기 전 값과 견준 비율(%)도 함께 보여 줌. 예를 들어 받기 전 4주에 5번 오던 가구에서 효과가 +0.25번이면 5%

```python
# 장바구니당 구매액 (기간별 4주당 구매액 ÷ 4주당 방문 수, 오지 않은 기간은 값 없음)
for w in ["받기 전", "캠페인 중", "끝난 뒤"]:
    panel_all[f"{w} 장바구니당"] = panel_all[f"{w} 구매액"] / panel_all[f"{w} 방문"].replace(0, np.nan)
basket_cols = [f"{w} 장바구니당" for w in ["받기 전", "캠페인 중", "끝난 뒤"]]
tr_b, ct_b, _ = matched_values(panel_all, pairs_m, basket_cols)
treated[basket_cols], control[basket_cols] = tr_b[basket_cols], ct_b[basket_cols]

# 유형 × 기간 × 값별 효과와 95% 신뢰구간, 받은 가구의 받기 전 평균 대비 비율(%)
def effect_table(tr, ct, kinds, periods=("캠페인 중", "끝난 뒤")):
    rows = {}
    for k in kinds:
        for period in periods:
            for v in ["구매액", "방문", "장바구니당"]:
                r = matched_did(tr, ct, k, f"받기 전 {v}", f"{period} {v}")
                base = tr.loc[tr["유형"].eq(k), f"받기 전 {v}"].mean()
                rows[(k, period, v)] = pd.concat([r, pd.Series({"받기 전 대비 %": r["효과"] / base * 100})])
    return pd.DataFrame(rows).T.rename_axis(["유형", "기간", "값"])

kinds = ["TypeA", "TypeB", "TypeC"]
effect = effect_table(treated, control, kinds)
with pd.option_context("display.float_format", "{:,.2f}".format):
    display(effect)
```

'효과'는 받은 가구 한 쌍당 4주에 더 산 금액(구매액·장바구니당은 달러)과 더 온 횟수(방문은 번). '받기 전 대비 %'는 받은 가구의 받기 전 평균에 견준 크기. '쌍'은 계산에 들어간 받은 쌍 수로, 장바구니당은 두 기간에 한 번 이상 온 쌍만 셈.

- A형: 캠페인 중에는 방문이 0.18번(3.5%) 늘었지만 구매액은 +3.9달러로 신뢰구간이 0을 포함함. 끝난 뒤에는 구매액 +0.2달러, 방문 0.00번으로 남은 것이 없음. 장바구니당 구매액만 끝난 뒤 +1.2달러(3.4%) 늘었는데, 구매액 전체로는 이어지지 않음
  - A형은 짝을 지은 뒤에도 받기 전에 조금 줄어드는 중이었음(6.1절, 6주 사이 방문 −0.07번, 구매액 −6.6달러). 이 흐름이 이어졌다면 A형 효과는 실제보다 조금 작게 나왔을 수 있음
- B형: 캠페인 중 구매액 +13.9달러(6.4%), 방문 +0.36번. 끝난 뒤에도 구매액 +13.8달러(6.4%, 신뢰구간 5.6∼21.2달러)가 그대로 남았음. 방문은 +0.14번으로 신뢰구간이 0을 포함해 줄지 않았음
  - 끝난 뒤 효과는 주로 장바구니당 구매액(+2.2달러, 5.4%)에서 왔음. 계속 오던 가구가 올 때 더 담게 된 모습
- C형: 캠페인 중 구매액 +38.6달러(18.5%), 끝난 뒤 +53.5달러(25.7%). 방문도 끝난 뒤 +1.33번(24.7%)으로 세 유형 가운데 가장 크고, 효과가 방문에서 왔음
  - 다만 짝이 150쌍뿐이고 균형 기준을 하나 넘었음(6.1절). 짝을 지은 뒤에도 받은 가구가 받기 전 6주 사이 구매액이 11.9달러 더 줄던 가구라(6.1절), 잠깐 줄었다가 돌아온 몫이 효과에 섞였을 수 있음

```python
# 판정 (3.4절): 구매액이 늘고(하한 > 0) 방문이 줄지 않음(상한 >= 0)
def judge_effect(eff, weak=("TypeC",)):
    out = {}
    for k in eff.index.get_level_values("유형").unique():
        up = {p: eff.loc[(k, p, "구매액"), "하한"] > 0 for p in ["캠페인 중", "끝난 뒤"]}
        down = {p: eff.loc[(k, p, "방문"), "상한"] < 0 for p in ["캠페인 중", "끝난 뒤"]}
        if up["끝난 뒤"] and not down["끝난 뒤"]:
            v = "효과 있음"
        elif up["끝난 뒤"] and down["끝난 뒤"]:
            v = "엇갈림"
        elif up["캠페인 중"] and not down["캠페인 중"]:
            v = "반짝 효과"
        else:
            v = "효과 없음"
        out[k] = v + (" (약함)" if k in weak else "")
    return pd.Series(out, name="판정")

judge_effect(effect).to_frame()
```

- A형은 효과 없음: 캠페인 중 방문이 조금 늘었지만 구매액으로 이어지지 않았고, 끝난 뒤에는 남은 것이 없음
- B형은 효과 있음: 끝난 뒤에도 구매액이 6.4% 높게 유지됐고 방문은 줄지 않았음. 늘어난 몫은 주로 올 때 더 담은 데서 옴
- C형은 효과 있음 (약함): 크기는 가장 크고 방문에서 왔지만, 짝이 적고 받기 전 흐름이 완전히 맞지 않아 03 노트북에서 단독 근거로 쓰지 않음

- 0.6에서 '(약함)'은 6.1절과 같이 균형 기준(0.1)을 넘는 값이 있는 유형에 붙임

### 7.2 참고: 끝난 뒤를 볼 수 없는 캠페인 8개의 캠페인 중 효과

3.1절대로 끝난 뒤 12주를 볼 수 없는 8개(A형 18번, B형 19·21·22·23·25번, C형 15·20번)는 판정하지 않고, 캠페인 중 효과만 참고로 봄. 받기 전 12주, 비교 가구(받기 전 12주 시작부터 캠페인 끝까지 겹치는 캠페인을 받지 않은 가구), 짝짓기(거리 기준 0.4)는 판정 캠페인과 같음.

- 판정 캠페인의 캠페인 중 효과와 방향이 같은지만 봄. 같은 유형인데 방향이 반대면 그 유형의 효과가 캠페인마다 다를 수 있다는 뜻으로 적음

```python
# 참고 캠페인: 6절과 같은 방법, 끝난 뒤 기간 없이
ref = camp[camp["쓰임"].eq("참고 (캠페인 중만)")]
rows_ref = []
for _, c in ref.iterrows():
    w = windows(c)
    w.pop("끝난 뒤")
    lo, hi = w["받기 전"][0], c["END_DAY"]
    busy = set(got_period.loc[(got_period["START_DAY"] <= hi) & (got_period["END_DAY"] >= lo), "household_key"])
    hhs = first_day.index[first_day < lo]
    base = pd.DataFrame({"받음": [(c["CAMPAIGN"], h) in received for h in hhs]}, index=pd.Index(hhs, name="household_key"))
    base = base[base["받음"] | ~base.index.isin(busy)]
    for name, (a, b) in w.items():
        base = base.join(window_rate(a, b, base.index).add_prefix(name + " "))
    rows_ref.append(base.assign(CAMPAIGN=c["CAMPAIGN"], 유형=c["DESCRIPTION"]).reset_index())
panel_ref = pd.concat(rows_ref, ignore_index=True)
for w_ in ["받기 전", "캠페인 중"]:
    panel_ref[f"{w_} 장바구니당"] = panel_ref[f"{w_} 구매액"] / panel_ref[f"{w_} 방문"].replace(0, np.nan)

# 짝짓기와 균형
pairs_ref = match_pairs(panel_ref)
ref_cols = match_cols + [f"{a} {b}" for a in ["받기 전", "캠페인 중"] for b in ["방문", "구매액", "장바구니당"]]
tr_r, ct_r, ctall_r = matched_values(panel_ref, pairs_ref, ref_cols)
bal_r = pd.DataFrame({k: {col: (tr_r.loc[tr_r["유형"].eq(k), col].mean() - ctall_r.loc[ctall_r["유형"].eq(k), col].mean())
                               / np.sqrt((tr_r.loc[tr_r["유형"].eq(k), col].var() + ctall_r.loc[ctall_r["유형"].eq(k), col].var()) / 2)
                         for col in bal_cols} for k in kinds}).T
ref_summary = pairs_ref.groupby("유형").agg(받은_쌍=("t", "size"), 짝_있음=("짝 있음", "sum"))
ref_summary["표준화 평균 차이 최대"] = bal_r.abs().max(axis=1)
with pd.option_context("display.float_format", "{:,.3f}".format):
    display(ref_summary)
```

## 8. 효과를 금액으로

7절의 효과는 4주당 값이라, 캠페인 한 번이 실제로 얼마를 더 팔았는지 바로 보이지 않음. 그래서 세 가지로 바꿈. 짝을 찾은 받은 쌍에 대한 값이고, 거리 기준 0.4(7절)와 0.6(7.1절)을 함께 보여 줌.

1. 받은 쌍 하나당 늘어난 구매액: 캠페인 중과 끝난 뒤 12주를 합친 금액
   - 캠페인 중 = 캠페인 중 4주당 효과 × (캠페인 일수 ÷ 28). 캠페인마다 길이가 달라 쌍마다 그 캠페인 일수를 씀
   - 끝난 뒤 = 끝난 뒤 4주당 효과 × 3 (12주는 4주의 세 배)
   - 예를 들어 캠페인 중 4주당 +14달러, 캠페인 56일, 끝난 뒤 4주당 +14달러면 14 × 2 + 14 × 3 = 70달러
2. 받은 쌍 전체로 늘어난 구매액: 1 × 짝을 찾은 쌍 수. 짝을 못 찾은 가장 자주 오던 가구는 효과를 모르므로 넣지 않음
3. 쿠폰 할인 1달러당 늘어난 구매액: 1 ÷ 받은 쌍 하나당 그 캠페인 쿠폰으로 받은 할인
   - 쿠폰 할인은 받은 가구가 그 캠페인 쿠폰을 쓴 날, 그 쿠폰 상품에 찍힌 쿠폰 할인(제조사 쿠폰 할인 + 마트의 쿠폰 매칭 할인)의 합. 쿠폰 사용 기록에는 금액이 없어 거래 기록에서 찾아 붙인 근사값
   - 쿠폰을 쓰지 않은 쌍은 0으로 넣음. 받은 쌍 전체에 들어간 할인으로 나누기 위해서
- 신뢰구간은 7절과 같이 받은 가구를 가구 단위로 다시 뽑아 1,000번 되풀이해 구함
- 한계: 늘어난 것은 구매액(매출)이지 이익이 아님. 쿠폰 할인 가운데 제조사가 내는 몫이 얼마인지, 캠페인 인쇄·발송 비용이 얼마인지는 데이터에 없음. 그래서 3은 비용 대비 이익이 아니라 '할인 1달러가 매출을 얼마나 끌어왔나'로만 읽음

```python
# 받은 쌍마다 늘어난 구매액: 캠페인 중(4주당 효과 × 일수/28) + 끝난 뒤(4주당 효과 × 3)
camp_days = (camp["END_DAY"] - camp["START_DAY"] + 1).set_axis(camp["CAMPAIGN"])

def pair_totals(tr, ct):
    during = (tr["캠페인 중 구매액"] - tr["받기 전 구매액"]) - (ct["캠페인 중 구매액"] - ct["받기 전 구매액"])
    after = (tr["끝난 뒤 구매액"] - tr["받기 전 구매액"]) - (ct["끝난 뒤 구매액"] - ct["받기 전 구매액"])
    out = tr[["household_key", "CAMPAIGN", "유형"]].copy()
    out["캠페인 중"] = during * tr["CAMPAIGN"].map(camp_days) / 28
    out["끝난 뒤"] = after * 3
    out["합계"] = out["캠페인 중"] + out["끝난 뒤"]
    return out

# 평균과 가구 단위 부트스트랩 95% 신뢰구간 (7절 matched_did와 같은 방식, 비율이면 분모도 함께 다시 뽑음)
def boot_mean(d, num, den=None, n_boot=1000, seed=0):
    hh_id = pd.factorize(d["household_key"])[0]
    n_hh = hh_id.max() + 1
    a = d[num].to_numpy()
    b = np.ones(len(d)) if den is None else d[den].to_numpy()
    rng = np.random.default_rng(seed)
    boot = []
    for _ in range(n_boot):
        w = np.bincount(rng.integers(0, n_hh, n_hh), minlength=n_hh)[hh_id]
        boot.append(np.sum(w * a) / np.sum(w * b))
    lo, hi = np.percentile(boot, [2.5, 97.5])
    return a.sum() / b.sum(), lo, hi

tot = {"0.4": pair_totals(treated, control), "0.6": pair_totals(tr_w, ct_w)}
money_rows = {}
for cal, d in tot.items():
    for k in kinds:
        dk = d[d["유형"].eq(k)]
        m, lo, hi = boot_mean(dk, "합계")
        money_rows[(k, cal)] = {"쌍": len(dk), "쌍당 캠페인 중": dk["캠페인 중"].mean(), "쌍당 끝난 뒤": dk["끝난 뒤"].mean(),
                                "쌍당 합계": m, "하한": lo, "상한": hi, "전체 합계": dk["합계"].sum()}
money = pd.DataFrame(money_rows).T.rename_axis(["유형", "거리 기준"])
with pd.option_context("display.float_format", "{:,.1f}".format):
    display(money)
```

값은 달러. '쌍당 합계'는 받은 쌍 하나가 캠페인 중과 끝난 뒤 12주에 더 산 금액의 평균과 95% 신뢰구간, '전체 합계'는 짝을 찾은 쌍 모두를 더한 금액.

- B형: 받은 쌍 하나당 52.9∼57.9달러(거리 기준 0.6 / 0.4)를 더 샀음. 그 가운데 끝난 뒤 12주가 33.8∼41.3달러로 더 큼. 짝을 찾은 쌍 전체로는 약 6.1만∼7.8만 달러
- C형: 쌍당 224∼255달러, 전체 약 3.8만∼5.8만 달러. 크기는 가장 크지만 7절대로 약함
- A형: 쌍당 6.5달러(−10.8∼24.5)와 −4.9달러로 0과 구별되지 않음. 전체 합계도 거리 기준에 따라 +1.3만 달러에서 −1.3만 달러로 부호가 바뀜

```python
# 받은 쌍마다 그 캠페인 쿠폰으로 받은 할인: 쿠폰을 쓴 날 그 쿠폰 상품에 찍힌 쿠폰 할인 (할인 열은 음수라 부호를 바꿈)
coupon = pd.read_csv(DATA / "coupon.csv")
used = redeem.merge(coupon, on=["COUPON_UPC", "CAMPAIGN"])
disc = (used.merge(x[["household_key", "DAY", "PRODUCT_ID", "COUPON_DISC", "COUPON_MATCH_DISC"]],
                   on=["household_key", "DAY", "PRODUCT_ID"])
            .drop_duplicates(["household_key", "DAY", "PRODUCT_ID", "CAMPAIGN", "COUPON_DISC", "COUPON_MATCH_DISC"]))
disc["쿠폰 할인"] = -(disc["COUPON_DISC"] + disc["COUPON_MATCH_DISC"])
pair_disc = disc.groupby(["household_key", "CAMPAIGN"])["쿠폰 할인"].sum()
print(f"쿠폰 사용 {len(redeem):,}건 가운데 거래 기록에서 할인을 찾은 사용: "
      f"{redeem.set_index(['household_key', 'CAMPAIGN']).index.isin(pair_disc[pair_disc > 0].index).mean() * 100:.1f}%")

# 유형·거리 기준별: 쿠폰을 쓴 쌍 비율, 쌍당 할인, 할인 1달러당 늘어난 구매액
coupon_rows = {}
for cal, d in tot.items():
    d = d.join(pair_disc, on=["household_key", "CAMPAIGN"]).fillna({"쿠폰 할인": 0})
    for k in kinds:
        dk = d[d["유형"].eq(k)]
        m, lo, hi = boot_mean(dk, "합계", "쿠폰 할인")
        coupon_rows[(k, cal)] = {"쿠폰 쓴 쌍 %": (dk["쿠폰 할인"] > 0).mean() * 100, "쌍당 할인": dk["쿠폰 할인"].mean(),
                                 "할인 1달러당 늘어난 구매액": m, "하한": lo, "상한": hi}
coupon_eff = pd.DataFrame(coupon_rows).T.rename_axis(["유형", "거리 기준"])
with pd.option_context("display.float_format", "{:,.2f}".format):
    display(coupon_eff)
```

'쿠폰 쓴 쌍 %'는 받은 쌍 가운데 그 캠페인 쿠폰을 한 장이라도 쓰고 거래 기록에서 할인이 찾아진 쌍의 비율, '쌍당 할인'은 받은 쌍 전체의 평균 쿠폰 할인(달러).

- 쿠폰 사용 2,318건 가운데 84.8%는 거래 기록에서 할인 금액을 찾았음. 나머지는 쿠폰 상품과 거래가 맞춰지지 않아 0으로 들어가, 할인은 조금 작게 잡혔을 수 있음
- 받은 쌍 가운데 쿠폰을 쓴 쌍은 5∼10%뿐이고, 받은 쌍 하나당 할인은 0.05∼0.24달러로 아주 작음
- 그래서 할인 1달러당 늘어난 구매액은 B형 476∼487달러, C형 2,005∼5,473달러로 매우 크고 구간도 넓음. 이 값은 '쿠폰 할인이 효율적'이라는 뜻이라기보다, 늘어난 구매액 대부분이 쿠폰 할인으로 산 것이 아니라는 뜻으로 읽음
  - 예를 들어 B형 효과(쌍당 57.9달러)를 쿠폰을 쓴 쌍(5.7%)만으로 내려면, 쓴 쌍 하나가 약 1,000달러를 더 사야 함. 받기 전 4주에 200달러 안팎을 사던 가구에게는 너무 큰 값이라, 쿠폰을 쓰지 않은 가구도 더 샀을 가능성이 큼
- 캠페인의 실제 비용(인쇄·발송, 제조사가 내는 할인 몫)은 데이터에 없어, 이 표로 캠페인이 이익을 냈는지는 판단하지 않음

#### 8절 정리

- 대전제 관점에서 효과가 있었던 B형은 받은 쌍 하나당 50달러 남짓, 짝을 찾은 쌍 전체로 6만∼8만 달러를 더 팔았음. 절반 넘게가 캠페인이 끝난 뒤 12주에 나왔음
- C형은 쌍당 크기가 더 크지만(220∼260달러) 받은 가구가 적고 7절대로 약함
- 쿠폰을 쓴 가구는 5∼10%뿐이고 할인 금액도 작아, 효과의 대부분은 쿠폰 할인 자체보다 캠페인을 받았다는 것(안내·쿠폰 묶음을 본 것)에서 왔을 가능성이 큼. 이것이 맞는지는 쿠폰을 쓴 가구와 쓰지 않은 가구의 효과를 나눠 봐야 확인할 수 있음

### 8.1 쿠폰을 쓰지 않은 가구도 더 샀나

8절에서 쿠폰을 쓴 쌍이 5∼10%뿐인데도 효과가 컸음. 효과가 쿠폰을 쓴 가구에서만 났는지, 쓰지 않은 가구에서도 났는지 봄. 쓰지 않은 가구에서도 났다면, 효과의 상당 부분은 할인보다 캠페인을 받았다는 것(안내·쿠폰 묶음을 본 것)에서 왔다는 근거가 됨.

- 받은 쌍을 그 캠페인 쿠폰을 한 장이라도 쓴 쌍(쿠폰 사용 기록 기준)과 쓰지 않은 쌍으로 나누고, 7절과 같은 짝(거리 기준 0.4)으로 효과를 따로 구함
- 판단 기준 (결과 보기 전): 쓰지 않은 쌍의 끝난 뒤 4주당 구매액 효과의 95% 신뢰구간이 0보다 크면 '쿠폰을 쓰지 않은 가구도 더 샀음'. 0을 포함하면 '쓰지 않은 가구의 효과는 확인되지 않음'
- 쿠폰을 쓸지는 가구가 스스로 고른 것이라, 쓴 쌍과 쓰지 않은 쌍의 차이를 쿠폰의 효과로 읽지 않음. 예를 들어 원래 그 상품을 사려던 가구가 쿠폰을 썼을 수 있음. 쓰지 않은 쌍에서도 효과가 있는지만 봄
- 효과가 판정된 B형·C형이 중심이고, A형은 참고로 함께 보여 줌

```python
# 받은 쌍마다 그 캠페인 쿠폰을 한 장이라도 썼나 (쿠폰 사용 기록 기준)
used_pairs = set(zip(redeem["household_key"], redeem["CAMPAIGN"]))
use_flag = np.array([(h, c) in used_pairs for h, c in zip(treated["household_key"], treated["CAMPAIGN"])])

# 쓴 쌍 / 쓰지 않은 쌍별 효과 (거리 기준 0.4의 짝)
use_rows = {}
for label, m in [("쓴 쌍", use_flag), ("쓰지 않은 쌍", ~use_flag)]:
    tr_u, ct_u = treated[m].reset_index(drop=True), control[m].reset_index(drop=True)
    for k in kinds:
        for period in ["캠페인 중", "끝난 뒤"]:
            for v in ["구매액", "방문"]:
                use_rows[(k, label, period, v)] = matched_did(tr_u, ct_u, k, f"받기 전 {v}", f"{period} {v}")
by_use = pd.DataFrame(use_rows).T.rename_axis(["유형", "쿠폰", "기간", "값"])
with pd.option_context("display.float_format", "{:,.2f}".format):
    display(by_use.xs("끝난 뒤", level="기간").unstack("값").swaplevel(axis=1).sort_index(axis=1)
            [[("구매액", "쌍"), ("구매액", "효과"), ("구매액", "하한"), ("구매액", "상한"), ("방문", "효과"), ("방문", "하한"), ("방문", "상한")]])
```

값은 끝난 뒤 12주의 4주당 효과(구매액은 달러, 방문은 번)와 95% 신뢰구간. '쌍'은 짝을 찾은 받은 쌍 수.

- B형: 쿠폰을 쓰지 않은 쌍(966쌍)도 끝난 뒤 구매액이 +10.0달러(1.8∼18.1) 늘었음. 판단 기준대로 '쿠폰을 쓰지 않은 가구도 더 샀음'. 방문은 +0.06번으로 0을 포함함
  - 쓴 쌍(80쌍)은 +59.9달러, 방문 +1.07번으로 훨씬 큼. 다만 쿠폰을 쓸 만한 가구가 스스로 쓴 것이라 이 차이를 쿠폰의 효과로 읽지 않음
- C형: 쓰지 않은 쌍(141쌍)이 +51.7달러(32.2∼74.1), 방문 +1.27번. C형 효과는 거의 쓰지 않은 쌍에서 나옴. 쓴 쌍은 9쌍뿐이라 읽지 않음
- A형: 쓴 쌍과 쓰지 않은 쌍 모두 0과 구별되지 않음. 7절의 효과 없음과 같음

```python
# 끝난 뒤 4주당 구매액 효과 합계 가운데 쓰지 않은 쌍에서 나온 몫 (쌍 수 × 쌍당 효과)
after = by_use.xs(("끝난 뒤", "구매액"), level=("기간", "값"))
share = (after["쌍"] * after["효과"]).unstack("쿠폰")
share["쓰지 않은 쌍 몫 %"] = share["쓰지 않은 쌍"] / (share["쓴 쌍"] + share["쓰지 않은 쌍"]) * 100
with pd.option_context("display.float_format", "{:,.1f}".format):
    display(share.loc[["TypeB", "TypeC"]])
```

- 끝난 뒤 효과 합계 가운데 쿠폰을 쓰지 않은 쌍에서 나온 몫은 B형 66.8%, C형 90.8%. 쓴 쌍 하나당 효과는 더 크지만, 쓰지 않은 쌍이 훨씬 많아서

#### 8.1 정리

- 효과가 있었던 B형·C형은 쿠폰을 쓰지 않은 가구도 캠페인 뒤에 더 샀고, 효과 합계의 3분의 2 이상이 쓰지 않은 가구에서 나왔음
- 8절의 '효과 대부분이 쿠폰 할인 자체에서 온 것이 아님'과 같은 방향. 캠페인을 받았다는 것, 곧 안내나 쿠폰 묶음을 보고 이 마트를 다시 떠올린 것이 구매를 끌어왔을 가능성이 큼
- 다만 B형에서 쿠폰을 쓰지 않은 가구는 방문이 늘지 않았고(+0.06번), 올 때 더 담은 쪽. 더 자주 오게 한 것은 쿠폰을 쓴 가구(B형)와 C형

## 9. 줄어들던 가구에도 통했나

지속 감소 가구는 오는 횟수가 줄어든 가구. 7절의 효과는 받은 가구 전체의 평균이라, 줄어드는 중이던 가구에도 통했는지는 따로 봐야 함(3.3절).

- 받은 쌍을 받기 직전 상태로 나눔: 받기 전 12주 가운데 뒤 6주의 구매액이 앞 6주보다 줄었으면 '줄어들던 가구', 아니면 '나머지 가구'
- 짝은 7절과 같음(거리 기준 0.4). 짝도 받기 전 앞 6주와 뒤 6주가 닮은 가구라, 줄어들던 가구의 짝은 똑같이 줄어들던 받지 않은 가구. 그래서 '원래 줄던 흐름'은 빼고 캠페인 효과만 견줄 수 있음
- 판단 기준 (3.4절, 결과 보기 전): 줄어들던 가구에서 끝난 뒤 4주당 방문 효과의 95% 신뢰구간이 0보다 크면 03 노트북에 착안점으로 씀. 0을 포함하면 '줄어드는 가구를 더 자주 오게 한다는 근거 없음'
- 구매액 효과도 함께 보여 주지만 판정은 방문으로 함. 지속 감소 가구는 주로 오는 횟수가 줄었기 때문

```python
# 받기 직전 상태: 뒤 6주 구매액이 앞 6주보다 줄었으면 줄어들던 가구
decl = (treated["뒤 6주 구매액"] < treated["앞 6주 구매액"]).to_numpy()
print("줄어들던 가구 쌍 비율:", treated.assign(줄어듦=decl).groupby("유형")["줄어듦"].mean().mul(100).round(1).to_dict())

# 상태별 효과 (끝난 뒤, 캠페인 중)
state_rows = {}
for label, m in [("줄어들던 가구", decl), ("나머지 가구", ~decl)]:
    tr_s, ct_s = treated[m].reset_index(drop=True), control[m].reset_index(drop=True)
    for k in kinds:
        for period in ["캠페인 중", "끝난 뒤"]:
            for v in ["방문", "구매액"]:
                r = matched_did(tr_s, ct_s, k, f"받기 전 {v}", f"{period} {v}")
                base = tr_s.loc[tr_s["유형"].eq(k), f"받기 전 {v}"].mean()
                state_rows[(k, label, period, v)] = pd.concat([r, pd.Series({"받기 전 대비 %": r["효과"] / base * 100})])
by_state = pd.DataFrame(state_rows).T.rename_axis(["유형", "받기 직전", "기간", "값"])
with pd.option_context("display.float_format", "{:,.2f}".format):
    display(by_state.xs("끝난 뒤", level="기간"))
```

값은 끝난 뒤 12주의 4주당 효과(방문은 번, 구매액은 달러)와 95% 신뢰구간. 받은 쌍의 절반 정도(47∼53%)가 받기 직전에 줄어들던 가구.

- A형: 줄어들던 가구도 나머지 가구도 방문·구매액 모두 0과 구별되지 않음
