# 03. 해결 방안

01 노트북에서 찾은 원인과 02 노트북에서 본 지난 캠페인의 효과를 묶어, 분석 목표(01 노트북 3절)마다 누구에게, 무엇을, 언제 할지 CRM 방안을 세움. 01 노트북 4절 분석 설계의 3단계(해결 방안)에 해당함. 이 데이터는 지난 기록이라 방안의 효과를 직접 증명할 수는 없어, 효과는 4단계(성과 확인)에서 실행했을 때 확인할 방법으로 정함.

> 이 노트북은 교육 과정 강사님의 라이브러리(hossam)를 바탕으로 직접 만든 분석 모듈(`helpers` 패키지의 `my_*`)을 사용함.
> 코드 작성·디버깅과 문장 교정에 생성형 AI(Claude Code)를 보조 도구로 사용함.

## 1. 원인 탐색 결론 (01 노트북 11절)

> 지속 감소 가구(370가구)는 장보기의 일부를 다른 곳으로 옮겼음. 여기서 '다른 곳'은 경쟁사 매장이나 온라인처럼 이 데이터에 기록이 남지 않는 곳. 장보기가 덜 필요해졌거나 이 마트가 나빠진 것은 주된 원인이 아님.

옮겨 간 방식에 따라 가구를 세 묶음으로 나눔. 서서히·한 번에는 5·6분기에도 이 마트에 온 341가구를 나눈 것(01 노트북 10.1절).

| 묶음 | 가구 | 모습 | 근거 (01 노트북) |
|---|---|---|---|
| 서서히 옮겨 감 | 274가구 | 84주에 걸쳐 서서히 줄었고, 그사이 단골 품목 여럿을 통째로 안 사게 됨. 휴지·세제 같은 생활용품이 식사 재료보다 더 줄어듦 | 오는 횟수가 준 것만으로 저절로 안 사게 될 몫(단골 품목의 4.2%)보다 훨씬 많은 단골 품목(18.0%)을 아예 안 삼(10.4절). 생활용품이 더 줄어듦(8.2절) |
| 한 번에 옮겨 감 | 67가구 | 어느 때부터 구매가 계단처럼 한 번에 줄어듦. 대부분 오는 횟수가 크게 줄었음. 이사로 매장이 멀어졌거나, 경쟁사 매장·온라인으로 한꺼번에 옮긴 모습 | 계단 모양으로 줄어든 가구가 우연보다 네 배쯤 많음(10.1절). 이사와 한꺼번에 옮김은 가를 수 없음(10.5절) |
| 끊김 | 29가구 | 5·6분기에 한 번도 오지 않음 | 왜 떠났는지는 가를 수 없음 |

**원인이 아닌 것**

- 이 마트가 나빠짐: 지속 감소 가구가 사던 상품의 값·판매·행사는 다른 가구가 사던 상품과 똑같이 바뀌었음(7절)
- 장보기가 덜 필요해짐(주된 원인이 아님): 싼 상품으로 바꾼 흔적은 아주 작고(6절), 외식이 늘어난 흔적은 없으며(8.2절), 식구가 줄어든 가구는 있더라도 드묾(8.3절, 10.1절)
- 특정 매장 근처의 변화: 지속 감소 가구가 특정 매장에 몰리지 않았고, 같은 매장 손님들이 함께 줄지도 않았음(8.4절)

**확인하지 못한 것**

- 왜 다른 곳을 고르게 됐는지. 다른 곳의 값·거리·온라인 편의는 기록이 없음
- 한 번에 옮겨 간 가구가 이사한 것인지, 경쟁사 매장·온라인으로 옮긴 것인지
- 370가구 가운데 일부(75∼117가구 이상)는 실제로 줄지 않았는데 우연히 지속 감소로 잡혔을 수 있음(10.2절)

## 2. 원인에서 해결 방안으로

원인 탐색 결과마다 해결 방안의 방향. 쿠폰, 묶음 할인, 배송 같은 구체적인 수단은 아직 정하지 않고 방향만 적음.

| 원인 탐색 결과 | 해결 방안의 방향 |
|---|---|
| 장보기 자체는 줄지 않았음 (가지 1이 주된 원인이 아님) | 이 가구들은 줄어든 만큼을 다른 곳에서 여전히 사고 있음. 그래서 소비를 늘리게 하는 것보다, 다른 곳에서 사는 몫을 이 마트로 되돌리는 것이 방향. 예를 들어 가구 A가 휴지를 이제 경쟁사에서 산다면, 휴지를 다시 이 마트에서 사게 하는 것 |
| 이 마트가 나빠지지 않았음 (2-1 아님) | 마트 전체의 값을 내리거나 행사를 늘리는 방식은 맞지 않음. 모든 손님에게 혜택이 돌아가 비용이 크지만, 지속 감소 가구는 손님의 15%뿐이라 대부분의 돈이 원래 잘 오던 손님에게 쓰임. 그래서 지속 감소 가구만 골라 그 가구에 맞는 혜택을 주는 방식(CRM)이 맞음 |
| 서서히 옮겨 간 가구는 단골 품목을 통째로 끊었음 | 가구마다 끊긴 단골 품목이 곧 되찾을 품목. 예를 들어 가구 A가 1·2분기에 장바구니 네 번 넘게 담던 세제를 5·6분기에 한 번도 사지 않았다면, 가구 A에게서 먼저 되찾을 품목은 세제 |
| 생활용품이 식사 재료보다 더 줄었음 | 휴지·세제는 어디서 사도 같은 물건이라, 값이 싸거나 사기 편한 곳으로 옮기기 쉬움. 그래서 값이나 편의를 앞세운 혜택이 맞을 수 있는 품목. 다만 왜 옮겼는지는 확인하지 못해 가정임 |
| 서서히 옮겨 간 가구는 84주에 걸쳐 줄었음 | 늦게 알아챌수록 끊긴 품목이 쌓임. 단골 품목이 끊기기 시작할 때 알아채는 신호가 있으면 더 줄기 전에 대응할 수 있음(목표 3) |
| 한 번에 옮겨 간 가구는 이사인지 한꺼번에 옮긴 것인지 가를 수 없음 | 이사라면 혜택으로 거리를 이기기 어렵고, 한꺼번에 옮긴 것이라면 되찾을 여지가 있음. 그래서 비용이 적은 방식으로 먼저 한 번 다가가고, 반응한 가구에만 더 씀. 반응 여부가 둘을 가르는 단서도 됨 |
| 끊긴 가구는 왜 떠났는지 모름 | 다시 한 번 올 계기만 적은 비용으로 줌 |
| 지난 캠페인은 원래 많이 사던 가구에 몰려 나감 (00 노트북 1.7절) | 지난 캠페인이 효과가 있었는지, 줄어드는 가구에도 통했는지는 02 노트북에서 받기 전 구매가 비슷한 가구끼리 견줘 따로 추정하고, 그 착안점을 수단을 정할 때 더함. 다만 추정치라, 실행한 뒤의 효과는 새로 나누는 실험으로 확인함(4단계) |

## 3. 해결 방안 설계 (초안)

| 목표 (01 노트북 3절) | 대상 | 방향 | 언제 | 우선순위 | 성과 지표 |
|---|---|---|---|---|---|
| 1-가. 서서히 옮겨 간 가구를 되돌림 | 서서히 옮겨 간 274가구 | 끊긴 단골 품목을 다시 이 마트에서 사게 함. 생활용품이 끊겼으면 생활용품부터 | 단골 품목이 끊긴 것을 알아챈 때 (목표 3의 신호와 같음) | 되찾을 몫이 큰 가구부터 | 끊긴 단골 품목을 다시 산 가구 비율, 4주당 방문 수 |
| 1-나. 한 번에 옮겨 간 가구를 되돌림 | 한 번에 옮겨 간 67가구 | 비용이 적은 방식으로 한 번 다시 오게 하고, 반응한 가구에만 1-가와 같은 방향으로 이어감 | 구매가 뚝 떨어진 것을 알아챈 때 | 1-가 다음 | 다가간 뒤 4주 안에 다시 온 가구 비율. 반응이 없으면 이사로 보고 더 쓰지 않음 |
| 2. 끊긴 가구를 다시 오게 함 | 5·6분기에 끊긴 29가구 | 다시 한 번 올 계기를 적은 비용으로 줌 | 끊긴 것을 알아챈 때 | 줄기 전 구매액이 큰 가구부터 | 다시 온 가구 비율 |
| 3. 새로 줄어드는 가구를 막음 | 지금은 지속 감소가 아닌 가구 가운데 단골 품목이 끊기기 시작한 가구 | 1-가와 같은 방향으로 일찍 다가감 | 신호가 보인 때 | 신호가 강하고 되찾을 몫이 큰 가구부터 | 새로 지속 감소에 들어서는 가구 비율 |

- 되찾을 몫: 가구가 1·2분기에 끊긴 단골 품목에 쓴 금액. 예를 들어 가구 A가 1·2분기에 세제·휴지에 26주 동안 60달러를 썼는데 5·6분기에 둘 다 끊었다면 되찾을 몫은 60달러. 방안이 통했을 때 돌아올 수 있는 금액의 상한
- 우선순위에 흔히 쓰는 RFM(최근 방문, 방문 빈도, 구매액)을 쓰지 않는 이유: RFM은 지금 얼마나 사는지를 봄. 이 방안의 목표는 잃은 몫을 되찾는 것이라, 지금 많이 사는 가구보다 많이 잃은 가구가 먼저임. 예를 들어 주 100달러에서 90달러가 된 가구는 RFM으로는 앞이지만 되찾을 몫은 10달러, 주 60달러에서 20달러가 된 가구는 RFM으로는 뒤지만 되찾을 몫은 40달러
- 표의 '방향'은 무엇을 이루려는지만 정한 것. 어떤 수단(쿠폰, 묶음 할인, 배송 등)으로 이룰지는 02 노트북의 착안점을 더해 7절에서 정함
- 목표 3의 신호는 01 노트북의 '왜'와 질문이 다름('언제, 무엇이 먼저 변했나'). 그래서 이 노트북에서 새로 찾음

#### 확인 순서

| 절 | 확인하는 것 | 이어지는 목표 |
|---|---|---|
| 4 | 준비: 01 노트북의 묶음(서서히 274 / 한 번에 67 / 끊김 29)을 다시 만듦 | 전체 |
| 5 | 묶음마다 되찾을 몫이 얼마이고, 어떤 품목이 끊겼나 (가구별 되찾을 품목 목록) | 1-가, 1-나, 2 |
| 6 | 단골 품목이 끊기는 것이 구매액이 줄어든 것보다 먼저 보이나 (목표 3의 신호). 흔히 쓰는 '직전 분기보다 방문이 줄었음' 규칙과 견줌 | 3 |
| 7 | 목표별 CRM 시나리오 정리 (누구에게, 무엇을, 언제, 대상 가구 수, 되찾을 몫). 02 노트북의 착안점으로 방향마다 수단을 정함 | 전체 |

- 성과 확인(실험 설계)은 이 노트북을 마친 뒤 4단계에서 함
