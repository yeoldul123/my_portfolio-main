# 01. 분석 설계와 원인 탐색

00 노트북에서 정한 문제를 이어받아 분석 목적과 목표를 정하고, 이를 이루기 위한 분석 설계를 세움. 원인 탐색과 해결 방안은 이 설계를 따라 진행하되, 결과를 보며 설계를 고쳐 나감. 고친 경우 처음 설계와 고친 이유를 함께 남김.

> 이 노트북은 교육 과정 강사님의 라이브러리(hossam)를 바탕으로 직접 만든 분석 모듈(`helpers` 패키지의 `my_*`)을 사용함.
> 코드 작성·디버깅과 문장 교정에 생성형 AI(Claude Code)를 보조 도구로 사용함.

## 1. 문제 정의

00 노트북 11절에서 정한 문제를 그대로 옮김.

> 분석 대상 2,405가구 가운데 370가구(15.4%, 이하 지속 감소 가구)가 84주 내내 꾸준히 덜 오고 있다. 1·2분기에서 1년 뒤 5·6분기 사이 방문이 57.2%, 구매액이 61.2% 줄었고, 이들은 줄기 전 구매액의 22.1%를 차지하던 가구다.
>
> 이 분석은 업계가 성장의 동력으로 꼽는 가구 수와 충성 가구를 따라, 같은 가구들이 계속 오고 꾸준히 사서 구매액이 유지되거나 늘어나는 것을 건강한 성장으로 본다(3.5절의 대전제). 지속 감소 가구는 바로 그 '계속 오던 가구'가 점점 덜 오게 된 모습이라 이 기준에 어긋난다. 지금은 지속 증가 가구의 증가가 이 손실을 메워 전체 구매액이 5.5% 늘었지만, 그 증가가 없었다면 전체는 13.4% 줄었을 것이다. 지속 감소 가구의 92%는 아직 오고 있으나, 흐름이 이어져 이들이 떠나면 줄기 전 매출의 22.1%가 빠진다.

#### 지속 감소 가구

분석 대상은 00 노트북에서 1·2분기에 구매가 있던 2,405가구. 지속 감소 가구는 이 가운데 분석 기간 84주(18∼101주)의 구매액이 갈수록 줄어드는 추세가 우연으로 보기 어려운 가구(00 노트북 9.3절).

- 84주를 4주씩 21구간으로 묶고, 모든 가구에 공통인 계절 출렁임을 뺀 구간 구매액으로 판단함
- 가구마다 맨-켄달 추세 검정으로 '갈수록 내려가는 경향이 우연히 생길 확률'을 구해, 5% 미만이면 지속 감소로 봄
- 같은 방법으로 '지속 증가'(500가구)와 '뚜렷한 추세 없음'(1,535가구)도 나눔
- 370가구 가운데 75∼117가구는 우연히 걸렸을 수 있고, 어느 가구인지는 가릴 수 없음. 그래서 이 노트북은 지속 감소 가구를 한 묶음으로 나머지 가구와 비교함

## 2. 분석 목적

지속 감소 가구를 다시 이 마트에 오게 해서, 줄기 전 매출의 22.1%를 차지하던 몫을 되찾고 전체 성장이 지속 증가 가구에만 기대지 않게 하는 것.

지속 감소 가구 대부분(341가구, 92%)은 줄었어도 아직 오고 있어 더 줄기 전에 붙잡아야 하고, 5·6분기에 한 번도 오지 않은 29가구(8%)는 다시 돌아오게 해야 함. 이를 위해 이 분석은 지속 감소 가구가 왜 덜 오게 됐는지 찾고, 그 원인에 맞춰 누구에게 무엇을 언제 할지 CRM(고객 관계 관리, 쿠폰·캠페인처럼 가구별로 다르게 하는 마케팅) 방안을 제안함.

## 3. 분석 목표

목적을 이루기 위해 구체적으로 이룰 것.

1. 아직 오고 있는 지속 감소 가구(341가구)의 방문 감소를 멈추고 줄기 전 수준 쪽으로 되돌림. 이 가구들이 줄인 구매액의 90%가 방문 횟수에서 왔기 때문(00 노트북 9.5절)
2. 5·6분기에 한 번도 오지 않은 지속 감소 가구(29가구)를 다시 오게 함
3. 지금은 지속 감소가 아닌 가구가 같은 흐름에 들어서지 않게 함. 줄어드는 가구를 일찍 알아채 더 줄기 전에 대응함

## 4. 분석 설계 (초안)

1절의 문제를 출발점으로, 원인 탐색, 지난 캠페인 효과, 해결 방안, 성과 확인으로 이어지는 흐름. 단계별 방법은 분석을 진행하면서 계속 고침.

이 데이터는 이미 지난 기록이라, 제안한 방안의 효과를 이 데이터로 직접 증명할 수는 없음. 그래서 마지막 단계는 효과를 증명하는 것이 아니라, 실행했을 때 효과를 확인할 방법까지 정하는 것.

| 단계 | 묻는 것 | 어떻게 | 쓰는 데이터 | 이어지는 목표 | 결과물 | 상태 |
|---|---|---|---|---|---|---|
| 1. 원인 탐색 | 지속 감소 가구는 왜 덜 오게 됐나 | 요인 가설 트리로 원인 후보를 펼치고, 가설마다 판단 기준을 먼저 정한 뒤 지속 감소 가구와 나머지 가구를 비교함 | 거래, 상품, 특별 진열·전단, 캠페인·쿠폰, 인구통계 | 1·2·3 (원인을 알아야 세 목표의 수단을 고를 수 있음) | 가설별 판정표(뒷받침됨 / 약함 / 뒷받침되지 않음 / 확인 불가)와 근거 | 완료 (6∼11절) |
| 2. 지난 캠페인 효과 (원인 탐색을 마친 뒤 더함) | 이 마트가 했던 캠페인은 효과가 있었나, 줄어드는 가구에도 통했나 | 받은 가구의 방문·구매액 변화를, 받기 전 구매가 비슷한 받지 않은 가구의 변화와 견줘 효과를 추정함 | 거래, 캠페인·쿠폰 기록 | 1·2·3 (수단을 고를 근거) | 캠페인 유형별 추정 효과, 해결 방안에 줄 착안점 | 원인 탐색 뒤 (02 노트북) |
| 3. 해결 방안 | 누구를 어떻게 다시 오게 할까 | 원인 판정 결과와 캠페인 효과에 맞춰 목표마다 대상 가구의 우선순위와 수단을 정함 | 원인 탐색 결과, 거래(최근 방문·금액), 캠페인 기록 | 목표마다 대상과 수단을 따로 정함 | 목표별 CRM 시나리오(누구에게, 무엇을, 언제), 줄어드는 가구를 일찍 알아챌 신호 | 캠페인 효과 뒤 (03 노트북) |
| 4. 성과 확인 | 방안이 효과가 있는지 어떻게 확인할까 | 목표마다 성과 지표를 정하고, 대상 가구를 무작위로 두 묶음으로 나눠 한쪽에만 쿠폰·안내를 보낸 뒤 두 묶음의 성과 지표를 견주는 실험(A/B 테스트)을 설계함 | 해결 방안 | 목표마다 성과 지표를 따로 둠 | 목표별 성과 지표와 확인 방법(실험 설계) | 해결 방안 뒤 |

> **나중에 덧붙임 (원인 탐색을 마친 뒤):** 처음 설계는 원인 탐색, 해결 방안, 성과 확인의 세 단계였고, 캠페인 기록은 해결 방안에 쓸 자료로만 적어 두었음. 원인 탐색에서 해결 방향이 '다른 곳에서 사는 몫을 이 마트로 되돌리기'로 정해진 뒤, 수단을 고를 근거로 이 마트가 실제로 했던 캠페인의 효과를 따로 보기로 해 2단계를 더함. 지난 캠페인은 원래 많이 사던 가구에 몰려 나가(00 노트북 1.7절) 그대로 견주면 효과가 부풀려지므로, 받기 전 구매가 비슷한 가구끼리 견줌

### 4.1 원인 탐색 설계

#### 요인 가설 트리

이 마트에서 덜 사게 되는 길은 크게 둘. 장보기 자체가 줄었거나, 장보기 양은 그대로인데 일부를 다른 곳에서 하게 된 경우.

```
지속 감소 가구는 왜 이 마트에 덜 오게 됐나
├─ 1. 장보는 양 자체가 줄었다
│   ├─ 1-1. 돈을 아끼게 됐다
│   │   └─ 덜 사거나, 더 싼 상품으로 바꿨다
│   ├─ 1-2. 집에서 먹는 사람이 줄었다
│   │   ├─ 한동안 집을 비웠다 (긴 여행, 입원, 출장)
│   │   └─ 식구가 줄었다 (자녀 독립, 분가)
│   └─ 1-3. 집에서 덜 먹게 됐다
│       └─ 외식이나 배달을 더 한다
└─ 2. 장보기 일부를 다른 곳에서 한다
    ├─ 2-1. 이 마트가 덜 좋아졌다
    │   ├─ 가격: 할인이 줄거나 값이 올랐다
    │   ├─ 상품: 늘 사던 상품이 없어졌다
    │   └─ 행사: 특별 진열·전단 행사를 보고 오던 가구인데, 행사가 줄었다
    ├─ 2-2. 다른 곳이 더 좋아졌다
    │   └─ 근처에 다른 마트가 생겼거나, 온라인으로 장을 본다
    └─ 2-3. 사는 곳이나 다니는 길이 바뀌었다
        └─ 이사하거나 출퇴근 길이 바뀌어 가던 매장이 멀어졌다
```

- 이 데이터에는 다른 곳에서 산 기록이 없어 두 갈래를 직접 가를 수 없음. 갈래를 먼저 고르지 않고, 가설마다 그 가설에서만 나타날 흔적을 하나씩 확인함
- 이 데이터로 확인이 잘 되는 가설부터 확인함(아래 '확인 순서'). 식구 수 변화, 외식, 경쟁 매장, 이사는 직접 기록이 없어 간접 신호로만 봄
- 지속 감소 가구가 줄인 구매액의 90%는 방문 횟수에서 왔음(00 노트북 9.5절). 1-1의 '더 싼 상품으로 바꿨다'는 한 번 올 때 쓰는 돈을 줄이는 길이라 방문 감소의 주된 설명은 되기 어려움. 그래서 한 번 올 때 구매액 쪽의 보조 설명으로 봄
- 1-2의 '한동안 집을 비웠다'는 몇 주 비었다가 돌아오는 모습이라 꾸준한 감소로 잡히기 어려워, 아직 오는 341가구의 설명은 되기 어려움. 끝까지 오지 않은 29가구는 이 데이터 기간 안에서는 잠깐 비운 것과 떠난 것을 가를 수 없음
- 값이 올라 덜 사게 된 경우는 1-1(절약)과 2-1(가격)에 모두 걸림. 2-1의 변화(값, 상품, 행사)는 모든 가구가 똑같이 겪으므로, 지속 감소 가구만 줄었다면 이 가구들이 줄기 전에 그 상품·가격·행사에 더 기대고 있었다는 흔적이 있어야 2-1로 봄. 그런 흔적 없이 품목 전반에서 덜 사거나 싼 상품으로 바꿨으면 1-1로 봄

#### 가설별 흔적과 데이터

가설이 맞다면 지속 감소 가구에서 보일 흔적과 그 흔적을 확인할 데이터. '품목 전반에서 덜 삼' 같은 흔적은 여러 가설에서 함께 나타나므로, 비슷한 흔적을 남기는 가설과 가르는 점을 따로 적음. 가를 때는 주로 두 가지를 봄.

- 줄어든 모양: 서서히 줄었나, 어느 주부터 한 번에 떨어졌나, 같은 매장 가구들이 비슷한 주에 함께 떨어졌나
- 줄어든 범위: 품목 전반인가 일부 품목인가, 방문 횟수가 줄었나 한 번 올 때 구매액이 줄었나

예를 들어 가구 A가 주 50달러에서 1년에 걸쳐 20달러까지 서서히 내려왔다면 1-1·1-3 쪽이고, 가구 B가 40주차부터 갑자기 주 10달러가 됐다면 1-2(식구 감소)·2-3 쪽에 가까움.

'직접'은 그 흔적을 거래 기록에서 바로 셀 수 있는 경우, '간접'은 다른 이유로도 생기는 흔적이라 정황으로만 읽는 경우.

| 가설 | 맞다면 보일 흔적 | 헷갈리는 가설과 가르는 점 | 쓸 데이터 | 확인 정도 |
|---|---|---|---|---|
| 1-1. 돈을 아낌 | 특정 상품·행사와 상관없이 품목 전반에서 서서히 덜 삼(방문 감소). 한 번 올 때는 같은 품목 안에서 자체 상표(PB)·할인 상품 비중이 늘고 사는 상품의 단가가 내려감(보조 흔적) | 2-2·2-3도 품목 전반에서 줄지만 싼 상품으로 바꾸지는 않음. 같은 품목 안에서 싼 상품으로 바꾸고 쿠폰·할인 비중이 늘어난 흔적이 1-1만의 흔적 | 거래(단가, 할인), 상품(제조사 상표 / 자체 상표 구분) | 직접 |
| 1-2. 한동안 집을 비움 | 몇 주 동안 구매가 없다가 다시 나타남 | 다시 원래 수준으로 돌아오는지로 가름. 다만 돌아오는 가구는 꾸준한 감소로 잡히기 어렵고, 끝까지 오지 않은 29가구는 떠난 것과 가를 수 없음 | 거래(주별 구매) | 확인 불가 |
| 1-2. 식구가 줄어듦 | 어느 주부터 한 번에 줄어든 뒤 낮은 수준이 이어짐. 방문보다 한 번 올 때 담는 상품 수가 먼저 줄고, 같은 품목 안에서 작은 포장으로 바꿈 | 2-3도 한 번에 줄지만 장바구니 구성은 그대로이고 방문이 줄어듦. 1-2는 장바구니가 작아지는 쪽이 먼저 | 거래, 상품(품목 분류, 포장 크기는 글자 기록이라 정리 필요), 인구통계(801가구만 있고 한 시점 기록이라 식구 수 변화는 볼 수 없음) | 간접 |
| 1-3. 외식·배달 | 서서히 줄어듦. 고기·채소 같은 식사 재료가 줄고, 세제·휴지 같은 생활용품은 그대로 삼 | 2-2(다른 마트·온라인)는 생활용품까지 함께 줄어듦. 식사 재료와 생활용품의 감소 폭 차이로 가름 | 거래, 상품(품목 분류) | 간접 |

## 5. 준비

00 노트북에서 정한 기준을 그대로 써서 분석 대상 가구, 유형, 가구 크기 단계를 다시 만듦.

- 뺄 거래: 주유(부서 `KIOSK-GAS`), 기타 거래(부서 `MISC SALES TRAN`), 기타 품목(품목 `COUPON/MISC ITEMS`). 장보기가 아닌 거래라서(00 노트북 2절)
- 분석 기간: 18∼101주(84주). 18주부터 13주씩 1∼6분기, 남은 96∼101주는 7로 묶음
- 대상 가구: 1·2분기에 구매가 있던 가구
- 유형: 84주를 4주씩 21구간으로 묶고, 모든 가구에 공통인 계절 출렁임을 뺀 뒤 가구마다 추세 검정(맨-켄달, 유의수준 5%)으로 나눔(00 노트북 9.3절)
- 가구 크기 단계: 84주 주평균 구매액으로 나눈 다섯 단계, 1단계가 가장 적게 삼(00 노트북 9.2절)

````python
# 헬퍼 모듈·분석 라이브러리 불러오기
from helpers import *
import numpy as np
import pandas as pd
from pathlib import Path

# 출력 옵션: 모든 열 표시, 소수 첫째 자리
pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", "{:,.1f}".format)
````

````python
# 원천 CSV 위치 (Mendeley Data 데이터셋)
DATA = Path("원본데이터")

# 거래와 상품 사전
trans   = pd.read_csv(DATA / "transaction_data.csv")
product = pd.read_csv(DATA / "product.csv")
````

````python
# 거래에 부서·품목 붙이고 장보기가 아닌 거래 제외
df = trans.merge(product[["PRODUCT_ID", "DEPARTMENT", "COMMODITY_DESC"]], on="PRODUCT_ID", how="left")
excluded = df["DEPARTMENT"].isin(["KIOSK-GAS", "MISC SALES TRAN"]) | df["COMMODITY_DESC"].eq("COUPON/MISC ITEMS")

# 분석 기간(18~101주) 거래와 13주 단위 분기 (96~101주는 7)
xq = df[~excluded & df["WEEK_NO"].between(18, 101)].copy()
xq["분기"] = ((xq["WEEK_NO"] - 18) // 13 + 1).clip(upper=7)
print(f"제외 전 {len(df):,}줄 → 분석 기간 거래 {len(xq):,}줄")
````

````python
# 대상 가구: 1·2분기에 구매가 있던 가구, 1·2분기와 5·6분기 구매액
hh = (xq[xq["분기"].isin([1, 2, 5, 6])]
      .assign(시기=lambda d: np.where(d["분기"] <= 2, "1·2분기", "5·6분기"))
      .pivot_table(index="household_key", columns="시기", values="SALES_VALUE", aggfunc="sum", fill_value=0))
hh = hh[hh["1·2분기"] > 0]
print(f"대상 가구: {len(hh):,}")
````

````python
# 가구별 4주 구간 구매액 (18~101주, 21구간)
week84 = (xq[xq["household_key"].isin(hh.index)]
          .pivot_table(index="household_key", columns="WEEK_NO", values="SALES_VALUE", aggfunc="sum", fill_value=0)
          .reindex(index=hh.index, columns=range(18, 102), fill_value=0))
block = week84.to_numpy().reshape(len(hh), 21, 4).sum(axis=2)

# 계절분: 구간별 가구 평균을 직선(마트 전체 성장)으로 나눈 나머지
t_idx = np.arange(21)
block_mean = block.mean(axis=0)
slope, intercept = np.polyfit(t_idx, block_mean, 1)
season = block_mean / (slope * t_idx + intercept)

# 가구별 맨-켄달 추세 검정 (계절분을 뺀 4주 구간 구매액, 유의수준 5%)
trend = my_ts.trend_by_row(pd.DataFrame(block, index=hh.index), alpha=0.05, season=season,
                           labels=("지속 감소", "뚜렷한 추세 없음", "지속 증가"))
hh["유형"] = trend["유형"]

# 가구 크기 단계: 84주 주평균 구매액 다섯 단계 (1단계가 가장 적게 삼)
week_avg = week84.sum(axis=1) / 84
hh["크기 단계"] = pd.qcut(week_avg, 5, labels=["1단계", "2단계", "3단계", "4단계", "5단계"])
````

````python
# 유형별 가구 수, 1·2분기 구매액 몫, 1·2분기 대비 5·6분기 변화, 5·6분기 구매가 없는 가구
type_order = ["지속 감소", "뚜렷한 추세 없음", "지속 증가"]
summary = hh.groupby("유형").agg(가구=("유형", "size"), 앞=("1·2분기", "sum"), 뒤=("5·6분기", "sum"),
                                 **{"5·6분기 구매 없음": ("5·6분기", lambda s: (s == 0).sum())}).reindex(type_order)
summary["가구 %"] = summary["가구"] / len(hh) * 100
summary["1·2분기 구매액 몫 %"] = summary["앞"] / hh["1·2분기"].sum() * 100
summary["구매액 변화 %"] = (summary["뒤"] / summary["앞"] - 1) * 100
summary[["가구", "가구 %", "1·2분기 구매액 몫 %", "구매액 변화 %", "5·6분기 구매 없음"]]
````

````python
# 가구 크기 단계별 유형 가구 수
pd.crosstab(hh["크기 단계"], hh["유형"]).reindex(columns=type_order)
````

| 질문 | 재는 값 (가구마다, 1·2분기와 5·6분기 각각) | 맞다면 |

| ③ 할인을 더 챙겼나 | 할인 비중(%): 받은 할인 ÷ 할인 전 금액 | 늘어남 |

상대 단가는 예시로 보면 쉬움. 가구 A가 1·2분기에 우유를 개당 3.0달러에 샀고, 그때 전체 가구가 산 우유가 평균 3.0달러면 100. 5·6분기에 2.7달러짜리를 샀는데 전체 평균이 3.3달러로 올랐다면 82. 마트 전체에서 값이 오른 몫은 빠지고, A가 더 싼 우유로 바꿨다는 것만 남음. 가구가 산 품목마다 이렇게 구한 뒤 품목별 구매액으로 가중 평균함.

- 품목은 세부 품목(`SUB_COMMODITY_DESC`, 예: 흰 우유 `FLUID MILK WHITE ONLY`) 단위. 서로 대신 살 수 있는 상품끼리 묶이는 단위
- 개당 값은 할인 전 값. 할인으로 싸게 산 것은 ③에서 따로 보므로, ②에는 '어떤 상품을 골랐나'만 남김
- 할인 전 금액은 00 노트북 1.2절에서 확인한 식(`SALES_VALUE` − `RETAIL_DISC` − `COUPON_MATCH_DISC`). 받은 할인은 회원카드 할인, 제조사 쿠폰, 쿠폰 매칭 할인의 합

4.1절 표의 1-1 흔적 가운데 '품목 전반에서 서서히 덜 삼'은 2-2·2-3에서도 나타나, 줄어든 모양과 범위를 보는 8절에서 2-2·2-3과 함께 봄. 이 절은 1-1에서만 나타나는 '싼 쪽으로 바꿈'만 봄. 또 1-1은 한 번 올 때 쓰는 돈 쪽의 보조 설명(4.1절)이라, 뒷받침되더라도 방문 감소의 주된 설명은 아님.

#### 비교할 가구

- 지속 감소 가구 가운데 5·6분기에도 산 341가구와, 뚜렷한 추세 없음 가구 가운데 두 시기 모두 산 가구를 비교함. 5·6분기에 사지 않은 가구는 5·6분기 값을 잴 수 없음
- 지속 증가 가구는 4.1절 비교 원칙대로 참고로만 보여 줌

#### 판정 방법 (결과를 보기 전에 정함)

차이의 크기는 클리프의 델타(Cliff's delta)로 잼. 지속 감소 가구 하나와 뚜렷한 추세 없음 가구 하나를 짝지었을 때, 지속 감소 가구가 더 '아끼는 쪽'으로 바뀐 짝의 비율에서 그 반대인 짝의 비율을 뺀 값. 짝 100개 가운데 60개가 아끼는 쪽, 40개가 반대면 0.2이고, 0이면 두 묶음이 같음. 세 지표의 단위(%p, 지수)가 달라도 같은 자로 잴 수 있음.

- 크기 기준은 델타 0.147 이상. 흔히 '작은 효과'의 하한으로 쓰는 값(Romano 외, 2006)으로, 짝 100개 가운데 약 57개 대 43개
- 우연인지는 한쪽 맨-휘트니 검정(p < 0.05)으로 봄. 가구가 많으면 아주 작은 차이도 p-value가 작아져, p-value만으로는 차이가 큰지 알 수 없음

| 지표 판정 | 기준 |
|---|---|
| 뒷받침됨 | 아끼는 쪽으로 우연과 구별되고, 델타 0.147 이상이며, 가구 크기 다섯 단계 가운데 네 단계 이상에서 같은 방향 |
| 약함 | 아끼는 쪽으로 우연과 구별되지만, 델타가 0.147에 못 미치거나 같은 방향인 단계가 세 단계 이하 |
| 뒷받침되지 않음 | 우연과 구별되지 않거나 반대 방향 |

1-1 가설은 ① 자체 상표 비중, ② 상대 단가, ③ 할인 비중 가운데 둘 이상이 '뒷받침됨'이면 뒷받침됨, 하나뿐이거나 '약함'만 있으면 약함, 셋 모두 '뒷받침되지 않음'이면 뒷받침되지 않음으로 판정함.

### 6.1 가구마다 세 지표 구하기

````python
# 1·2분기·5·6분기 거래 (대상 가구), 상표·세부 품목 붙이기
xs = (xq[xq["household_key"].isin(hh.index) & xq["분기"].isin([1, 2, 5, 6])]
      .merge(product[["PRODUCT_ID", "BRAND", "SUB_COMMODITY_DESC"]], on="PRODUCT_ID", how="left"))
xs["시기"] = np.where(xs["분기"] <= 2, "1·2분기", "5·6분기")

# 할인 전 금액(00 노트북 1.2절의 식)과 받은 할인 (할인 열은 음수)
xs["할인 전 금액"] = xs["SALES_VALUE"] - xs["RETAIL_DISC"] - xs["COUPON_MATCH_DISC"]
xs["받은 할인"] = -(xs["RETAIL_DISC"] + xs["COUPON_DISC"] + xs["COUPON_MATCH_DISC"])
xs["자체 상표 금액"] = xs["SALES_VALUE"].where(xs["BRAND"].eq("Private"), 0)

# ① 자체 상표 비중, ③ 할인 비중 (가구·시기별, %)
g = xs.groupby(["household_key", "시기"])
metric = pd.DataFrame({
    "① 자체 상표 비중": g["자체 상표 금액"].sum() / g["SALES_VALUE"].sum() * 100,
    "③ 할인 비중": g["받은 할인"].sum() / g["할인 전 금액"].sum() * 100,
})
````

````python
# ② 상대 단가: 개당 값을 구할 수 있는 줄만 (수량·할인 전 금액이 0보다 크고 세부 품목이 적힌 줄)
unit = xs[(xs["QUANTITY"] > 0) & (xs["할인 전 금액"] > 0)
          & xs["SUB_COMMODITY_DESC"].notna() & xs["SUB_COMMODITY_DESC"].ne("NO SUBCOMMODITY DESCRIPTION")]
print(f"상대 단가에 쓴 줄: {len(unit) / len(xs) * 100:.1f}%")

# 같은 시기 전체 가구의 세부 품목별 평균 개당 값
item = unit.groupby(["시기", "SUB_COMMODITY_DESC"])[["할인 전 금액", "QUANTITY"]].sum()
item_price = (item["할인 전 금액"] / item["QUANTITY"]).rename("전체 개당 값")

# 가구의 세부 품목별 개당 값 ÷ 전체 개당 값 × 100, 가구의 세부 품목 구매액으로 가중 평균
hs = unit.groupby(["household_key", "시기", "SUB_COMMODITY_DESC"])[["할인 전 금액", "QUANTITY"]].sum()
hs = hs.join(item_price, on=["시기", "SUB_COMMODITY_DESC"])
hs["상대 단가"] = hs["할인 전 금액"] / hs["QUANTITY"] / hs["전체 개당 값"] * 100
hs["가중"] = hs["상대 단가"] * hs["할인 전 금액"]
hs_g = hs.groupby(["household_key", "시기"])[["가중", "할인 전 금액"]].sum()
metric["② 상대 단가"] = hs_g["가중"] / hs_g["할인 전 금액"]

# 가구 한 줄로 펼치기, 두 시기 모두 산 가구만
metric_cols = ["① 자체 상표 비중", "② 상대 단가", "③ 할인 비중"]
wide = metric[metric_cols].unstack("시기").dropna()
before = wide.xs("1·2분기", axis=1, level="시기").join(hh["유형"])
change = (wide.xs("5·6분기", axis=1, level="시기") - wide.xs("1·2분기", axis=1, level="시기")).join(hh[["유형", "크기 단계"]])
print("두 시기 모두 산 가구:", change["유형"].value_counts().reindex(type_order).to_dict())
````

- 상대 단가는 거래 줄의 99.3%로 구함. 나머지는 수량이나 할인 전 금액이 0이거나 세부 품목이 적혀 있지 않아 개당 값을 구할 수 없는 줄
- 두 시기 모두 산 가구는 지속 감소 341, 뚜렷한 추세 없음 1,502, 지속 증가 500가구. 이 가구들로 아래를 비교함

#### 자체 상표는 정말 더 싼가

①은 자체 상표가 같은 품목의 제조사 상표보다 싸다는 전제로, 자체 상표 비중이 늘면 돈을 아낀 흔적으로 읽음. 이 전제가 이 마트에서도 맞는지 먼저 봄.

- 같은 세부 품목 안에서 자체 상표(`BRAND` = Private)와 제조사 상표(`BRAND` = National)의 개당 값(할인 전)을 견줌. 예를 들어 흰 우유에서 자체 상표가 개당 2.5달러, 제조사 상표가 3.0달러면 비율은 0.83이고 자체 상표가 더 쌈
- 두 상표가 모두 30줄 이상 팔린 세부 품목만 봄. 줄이 적으면 개당 값이 몇 건의 거래에 크게 흔들림
- 판정 (결과 보기 전): 자체 상표가 더 싼 세부 품목이 세부 품목 수로도, 구매액으로도 절반을 넘으면 전제를 그대로 씀. 그렇지 않으면 ①을 돈을 아낀 흔적으로 읽지 않음
- 한계: 포장 크기는 글자 기록이라 '개당' 값으로 견줌. 같은 세부 품목 안에서 한쪽이 큰 포장을 더 많이 팔면 그쪽이 비싸 보일 수 있음

````python
# 세부 품목별 자체 상표·제조사 상표 개당 값 (두 시기 합침, 할인 전). 상표끼리 값을 견주는 함수가 없어 pandas로 직접 계산
pb = (unit[unit["BRAND"].isin(["Private", "National"])]
      .groupby(["SUB_COMMODITY_DESC", "BRAND"]).agg(금액=("할인 전 금액", "sum"), 수량=("QUANTITY", "sum"), 줄=("BRAND", "size"))
      .unstack("BRAND"))
pb = pb[(pb[("줄", "Private")] >= 30) & (pb[("줄", "National")] >= 30)]
pb_ratio = (pb[("금액", "Private")] / pb[("수량", "Private")]) / (pb[("금액", "National")] / pb[("수량", "National")])
pb_spend = pb[("금액", "Private")] + pb[("금액", "National")]
pb_cover = pb[("금액", "Private")].sum() / unit.loc[unit["BRAND"].eq("Private"), "할인 전 금액"].sum() * 100

print(f"견준 세부 품목: {len(pb)}개 (자체 상표 구매액의 {pb_cover:.1f}%)")
print(f"자체 상표가 더 싼 세부 품목: 수 {(pb_ratio < 1).mean() * 100:.1f}%, 구매액 {pb_spend[pb_ratio < 1].sum() / pb_spend.sum() * 100:.1f}%")
print("개당 값 비율(자체 ÷ 제조사) 분위:", pb_ratio.quantile([0.25, 0.5, 0.75]).round(2).to_dict())
````

- 두 상표가 모두 30줄 이상 팔린 세부 품목 573개를 견줌. 자체 상표 구매액의 90.7%가 이 품목들에 들어 있어, 자체 상표 대부분을 본 셈
- 자체 상표가 더 싼 세부 품목은 수로 86.7%, 구매액으로 87.8%. 기준(절반)을 크게 넘음
- 개당 값 비율의 중앙값은 0.72. 같은 세부 품목에서 자체 상표가 제조사 상표보다 대체로 30% 가까이 쌈. 절반의 세부 품목이 0.59∼0.84 사이
- 그래서 ①의 전제(자체 상표 = 싼 쪽)를 그대로 씀. 다만 열 품목 가운데 한 품목 남짓은 자체 상표가 더 비싸, 자체 상표 비중의 변화는 ② 상대 단가보다 덜 정확한 신호

### 6.2 줄기 전에는 비슷했나

줄어드는 동안의 변화를 보기 전에, 줄기 전(1·2분기) 출발점이 같았는지 봄. 지속 감소 가구가 원래부터 더 아끼며 사던 가구라면 변화를 읽는 방식이 달라짐. 값은 유형별 중앙값.

````python
# 유형별 1·2분기 값의 중앙값 (①·③은 %, ②는 전체 평균 = 100)
before.groupby("유형")[metric_cols].median().reindex(type_order)
````

- 세 유형이 거의 같음. 자체 상표 비중 20.4∼21.1%, 할인 비중 14.8∼15.6%, 상대 단가 111∼113. 지속 감소 가구가 원래 더 아끼며 사던 가구는 아님
- 상대 단가가 100보다 조금 큰 것은 가구의 품목별 구매액으로 가중해서임. 같은 품목에서 비싼 상품을 산 가구일수록 그 품목 구매액이 커 무게가 더 실림. 세 유형을 같은 방식으로 재므로 유형끼리 비교에는 지장 없음

### 6.3 줄어드는 동안 싼 쪽으로 바꿨나

1·2분기에서 5·6분기 사이에 세 지표가 얼마나 바뀌었나. '변화'는 가구마다 5·6분기 값 − 1·2분기 값의 중앙값(①·③은 %p, ②는 지수 차이). 델타는 '아끼는 쪽'이 양수가 되게 부호를 맞춤.

````python
# 지표별 아끼는 쪽: ①·③은 늘어남(greater), ②는 내려감(less)
expect = {"① 자체 상표 비중": "greater", "② 상대 단가": "less", "③ 할인 비중": "greater"}
expect_text = {"greater": "늘어남", "less": "내려감"}
dec = change[change["유형"].eq("지속 감소")]
flat = change[change["유형"].eq("뚜렷한 추세 없음")]
med = change.groupby("유형")[metric_cols].median()

rows = []
for col, alt in expect.items():
    res = my_stats.cliffs_delta(dec[col], flat[col]).loc[alt]
    rows.append({"지표": col, "맞다면": expect_text[alt],
                 "지속 감소 변화": med.loc["지속 감소", col], "뚜렷한 추세 없음 변화": med.loc["뚜렷한 추세 없음", col],
                 "지속 증가 변화 (참고)": med.loc["지속 증가", col],
                 "델타 (아끼는 쪽 +)": res["delta"] * (1 if alt == "greater" else -1), "한쪽 p-value": res["p-value"]})
compare = pd.DataFrame(rows).set_index("지표")
with pd.option_context("display.float_format", "{:,.3f}".format):
    display(compare)
````

- 세 지표 모두 변화가 0에 가까움. 지속 감소 가구는 자체 상표 비중 −0.007%p, 상대 단가 −2.0, 할인 비중 +0.136%p. 뚜렷한 추세 없음 가구는 +0.352%p, +0.3, −0.179%p
- ① 자체 상표 비중: 델타 −0.022로 오히려 아끼지 않는 쪽이고, p-value 0.737로 우연과 구별되지 않음
- ② 상대 단가: 지속 감소 가구는 지수가 2.0 내려가, 전체 평균 값의 약 2%만큼 싼 쪽으로 옮겨 갔고, p-value 0.021로 우연과는 구별됨. 하지만 델타 0.070은 짝 100개 가운데 약 53.5개 대 46.5개로, 크기 기준 0.147(57개 대 43개)의 절반에 못 미침
- ③ 할인 비중: 델타 0.044, p-value 0.104로 우연과 구별되지 않음

### 6.4 가구 크기 단계마다 같은 방향인가

많이 사는 가구와 적게 사는 가구는 사는 방식이 다를 수 있어(4.1절 비교 원칙), 같은 델타를 가구 크기 단계마다 다시 구함. 양수면 그 단계에서도 지속 감소 가구가 아끼는 쪽.

````python
# 가구 크기 단계별 델타 (아끼는 쪽 +)와 양수인 단계 수
tiers = ["1단계", "2단계", "3단계", "4단계", "5단계"]
tier_delta = pd.DataFrame({
    col: {t: my_stats.cliffs_delta(dec.loc[dec["크기 단계"].eq(t), col],
                                   flat.loc[flat["크기 단계"].eq(t), col]).loc[alt, "delta"] * (1 if alt == "greater" else -1)
          for t in tiers}
    for col, alt in expect.items()}).T
tier_delta["같은 방향 단계 (5개 중)"] = (tier_delta[tiers] > 0).sum(axis=1)
with pd.option_context("display.float_format", "{:,.3f}".format):
    display(tier_delta)
````

- 단계마다 들쭉날쭉함. ① 자체 상표 비중은 5단계에서만 0.239이고 나머지 네 단계는 음수. 같은 방향(0보다 큼)은 1개 단계
- ② 상대 단가는 2·4단계에서만 0.14∼0.19이고, 나머지는 0 근처이거나 음수. 같은 방향은 3개 단계(1단계 0.019 포함)
- ③ 할인 비중은 네 단계가 양수지만 모두 0.1 아래. 같은 방향은 4개 단계

### 6.5 판정

판정 방법의 세 조건을 지표마다 확인함.

````python
# 판정 방법의 세 조건과 지표 판정
cond = pd.DataFrame({
    "아끼는 쪽으로 우연과 구별 (p < 0.05)": (compare["한쪽 p-value"] < 0.05) & (compare["델타 (아끼는 쪽 +)"] > 0),
    "델타 0.147 이상": compare["델타 (아끼는 쪽 +)"] >= 0.147,
    "네 단계 이상 같은 방향": tier_delta["같은 방향 단계 (5개 중)"] >= 4,
})
cond["판정"] = np.select([cond.all(axis=1), cond.iloc[:, 0]], ["뒷받침됨", "약함"], default="뒷받침되지 않음")
display(cond.replace({True: "예", False: "아니오"}))

# 1-1 가설 판정: '뒷받침됨' 지표가 둘 이상이면 뒷받침됨
n_support = cond["판정"].eq("뒷받침됨").sum()
verdict_11 = ("뒷받침됨" if n_support >= 2 else
              "약함" if n_support == 1 or cond["판정"].eq("약함").any() else "뒷받침되지 않음")
print(f"뒷받침됨 지표 {n_support}개, 약함 지표 {cond['판정'].eq('약함').sum()}개 → 1-1 판정: {verdict_11}")
````

#### 1-1 판정: 약함

- '뒷받침됨' 지표는 없고, ② 상대 단가 하나가 '약함'이라 판정 방법대로 약함
- 지속 감소 가구가 싼 쪽으로 바꾼 흔적은 상대 단가에서만 아주 작게 보이고(전체 평균 값의 약 2%), 자체 상표와 할인에서는 보이지 않음. 1-1은 처음부터 방문 감소의 주된 설명 후보가 아니었고(4.1절), 한 번 올 때 쓰는 돈 쪽의 보조 설명으로도 약함

## 7. 2-1 이 마트가 덜 좋아졌나

1-1 다음으로, 같은 직접 확인 가설인 2-1을 봄(4.1절 확인 순서). 1-2·1-3은 8절에서 2-2·2-3과 함께 봄.

값이 오르거나, 늘 사던 상품이 없어지거나, 행사가 줄면 장보기 일부를 다른 곳으로 옮길 수 있음.

그런데 값·상품·행사는 마트가 바꾸는 것이라 한 가구에만 바뀌지 않음. 우유 값이 오르면 우유를 사는 모든 가구가 오른 값을 봄. 그래서 '마트가 나빠졌다'만으로는 왜 하필 지속 감소 가구만 덜 오게 됐는지 설명이 안 됨.

설명이 되려면, 나빠진 것이 지속 감소 가구가 사던 상품에 몰려 있어야 함(4.1절). 예를 들어 가구 A는 주로 우유와 빵을, 가구 B는 주로 고기와 채소를 산다고 하자. 1년 사이 우유와 빵 값만 크게 올랐다면, 같은 마트에서도 A가 겪은 값 인상이 B보다 큼. 그런데 A만 덜 오게 됐다면 값 인상이 이유일 수 있음.

그래서 가구마다 줄기 전(1·2분기)에 사던 상품이 1년 뒤(5·6분기)에 얼마나 나빠졌는지(값이 올랐나, 없어졌나, 행사에서 빠졌나) 재고, 지속 감소 가구와 뚜렷한 추세 없음 가구를 비교함.

| 질문 | 재는 값 (가구마다, 1·2분기에 산 상품 기준) | 맞다면 |
|---|---|---|
| ① 가격: 사던 상품의 값이 더 올랐나 | 가구 물가 변화(%): 1·2분기에 산 상품마다 마트 전체 개당 값의 1·2분기 → 5·6분기 변화율을, 가구의 상품별 1·2분기 구매액으로 가중 평균 | 더 많이 오름 |
| ② 상품: 사던 상품이 더 많이 사라졌나 | 사라진 상품 몫(%): 1·2분기 구매액 가운데, 5·6분기에 마트 어느 매장에서도 팔린 기록이 없는 상품의 몫 | 더 큼 |
| ③ 특별 진열·전단: 사던 상품이 행사에 덜 오르게 됐나 | 행사 주 비율 변화(%p): 1·2분기에 산 상품마다 가구의 주 매장에서 특별 진열이나 전단에 오른 주의 비율을 시기별로 구하고(5·6분기 − 1·2분기), 가구의 상품별 1·2분기 구매액으로 가중 평균 | 더 많이 줄어듦 |

예시로 보면 쉬움. 가구 A가 1·2분기에 우유 10달러어치, 빵 5달러어치를 샀다고 하자.

- ① 1년 사이 우유 값이 3.0달러에서 3.3달러로 10% 오르고 빵 값은 그대로면, A의 가구 물가 변화는 (10 × 10% + 5 × 0%) ÷ 15 = 6.7%

- ③ A의 주 매장에서 우유가 1·2분기 26주 가운데 13주(50%) 행사에 올랐고 5·6분기에는 5주(19%)라면, 우유의 변화는 −31%p. 빵도 같은 식으로 구해 구매액으로 가중 평균

- 개당 값은 마트가 받은 금액 ÷ 수량(`SALES_VALUE` ÷ `QUANTITY`). 회원카드 할인이 빠진 값이라, 정가가 그대로여도 할인이 줄면 값이 오른 것으로 잡힘(4.1절 표의 '값이 오르거나 할인이 줄어든')

### 7.1 가구마다 세 지표 구하기

````python
# 대상 가구의 1·2분기 상품별 구매액 (세 지표의 가중치)
base = (xq[xq["household_key"].isin(hh.index) & xq["분기"].isin([1, 2])]
        .groupby(["household_key", "PRODUCT_ID"])["SALES_VALUE"].sum().rename("구매액").reset_index())
base = base[base["구매액"] > 0]

# ② 5·6분기(70~95주)에 마트 어느 매장에서도 팔린 기록이 없는 상품 (원천 거래 전체 기준)
sold_56 = set(trans.loc[trans["WEEK_NO"].between(70, 95), "PRODUCT_ID"])
base["사라짐"] = ~base["PRODUCT_ID"].isin(sold_56)

# ① 마트 전체 상품별 개당 값 (마트가 받은 금액 ÷ 수량, 모든 가구)
sold = xq[xq["분기"].isin([1, 2, 5, 6]) & (xq["QUANTITY"] > 0) & (xq["SALES_VALUE"] > 0)]
sold = sold.assign(시기=np.where(sold["분기"] <= 2, "1·2분기", "5·6분기"))
price = sold.groupby(["PRODUCT_ID", "시기"])[["SALES_VALUE", "QUANTITY"]].sum()
price = (price["SALES_VALUE"] / price["QUANTITY"]).unstack("시기")
base = base.join((price["5·6분기"] / price["1·2분기"] - 1).mul(100).rename("값 변화"), on="PRODUCT_ID")

# 가구별 ① 가구 물가 변화(%, 두 시기 모두 값이 있는 상품의 구매액 가중 평균), ② 사라진 상품 몫(%)
priced = base.dropna(subset=["값 변화"])
exposure = pd.DataFrame({
    "① 가구 물가 변화": (priced["값 변화"] * priced["구매액"]).groupby(priced["household_key"]).sum()
                       / priced.groupby("household_key")["구매액"].sum(),
    "② 사라진 상품 몫": base["구매액"].where(base["사라짐"], 0).groupby(base["household_key"]).sum()
                       / base.groupby("household_key")["구매액"].sum() * 100,
})
print(f"1·2분기 상품 구매액 가운데 두 시기 모두 값이 있는 상품: {priced['구매액'].sum() / base['구매액'].sum() * 100:.1f}%")
````

````python
# 가구의 주 매장: 1·2분기 구매액이 가장 큰 매장
main_store = (xq[xq["household_key"].isin(hh.index) & xq["분기"].isin([1, 2])]
              .groupby(["household_key", "STORE_ID"])["SALES_VALUE"].sum()
              .groupby(level=0).idxmax().str[1].rename("주 매장"))

# 특별 진열·전단 기록 (1·2분기 18~43주, 5·6분기 70~95주). 같은 상품·매장·주가 두 줄인 경우는 한 번만 셈(00 노트북 1.8절)
causal = pd.read_csv(DATA / "causal_data.csv",
                     dtype={"PRODUCT_ID": "int32", "STORE_ID": "int32", "WEEK_NO": "int16",
                            "display": "category", "mailer": "category"})
causal = causal[(causal["WEEK_NO"].between(18, 43) | causal["WEEK_NO"].between(70, 95))
                & (causal["display"].ne("0") | causal["mailer"].ne("0"))]
causal = causal[["PRODUCT_ID", "STORE_ID", "WEEK_NO"]].drop_duplicates()
causal["시기"] = np.where(causal["WEEK_NO"] <= 43, "1·2분기", "5·6분기")

# 매장·상품별 행사 주 비율(%): 시기마다 26주 가운데 특별 진열이나 전단에 오른 주
promo_weeks = causal.groupby(["STORE_ID", "PRODUCT_ID", "시기"]).size().unstack("시기", fill_value=0) / 26 * 100

# ③ 주 매장에 기록이 있는 가구만, 1·2분기에 산 상품마다 행사 주 비율 변화(%p)를 구매액으로 가중 평균
b3 = base.join(main_store, on="household_key")
b3 = b3[b3["주 매장"].isin(causal["STORE_ID"].unique())]
b3 = b3.join(promo_weeks, on=["주 매장", "PRODUCT_ID"]).fillna({"1·2분기": 0, "5·6분기": 0})
b3["행사 변화"] = b3["5·6분기"] - b3["1·2분기"]
exposure["③ 행사 주 비율 변화"] = ((b3["행사 변화"] * b3["구매액"]).groupby(b3["household_key"]).sum()
                                 / b3.groupby("household_key")["구매액"].sum())

exposure = exposure.join(hh[["유형", "크기 단계"]])
exposure["끊긴 가구"] = hh["5·6분기"].eq(0)
exp_cols = ["① 가구 물가 변화", "② 사라진 상품 몫", "③ 행사 주 비율 변화"]
print(f"주 매장에 특별 진열·전단 기록이 있는 가구: {b3['household_key'].nunique():,} / {len(hh):,}")
print("지표를 구한 가구:", exposure[exp_cols].notna().sum().to_dict())
````

- ① 가구 물가 변화는 1·2분기 구매액의 86.5%를 차지하는 상품으로 구함. 나머지는 5·6분기에 팔리지 않았거나(②에서 봄) 개당 값을 구할 수 없는 상품

### 7.3 줄기 전에 사던 상품이 더 나빠졌나

세 지표를 유형별로 견줌. 값은 유형별 중앙값(①·②는 %, ③은 %p). 델타는 '2-1 쪽'(①·② 더 큼, ③ 더 줄어듦)이 양수가 되게 부호를 맞춤. 5·6분기에 오지 않은 29가구는 지속 감소 가구에 들어 있고, 따로 떼어 낸 값을 참고로 붙임.

````python
# 지표별 2-1 쪽: ①·②는 더 큼(greater), ③은 더 줄어듦(less)
expect2 = {"① 가구 물가 변화": "greater", "② 사라진 상품 몫": "greater", "③ 행사 주 비율 변화": "less"}
expect2_text = {"greater": "더 큼", "less": "더 줄어듦"}
dec2 = exposure[exposure["유형"].eq("지속 감소")]
flat2 = exposure[exposure["유형"].eq("뚜렷한 추세 없음")]
med2 = exposure.groupby("유형")[exp_cols].median()
gone = exposure.loc[exposure["끊긴 가구"], exp_cols].median()

rows = []
for col, alt in expect2.items():
    res = my_stats.cliffs_delta(dec2[col].dropna(), flat2[col].dropna()).loc[alt]
    rows.append({"지표": col, "맞다면": expect2_text[alt],
                 "지속 감소": med2.loc["지속 감소", col], "뚜렷한 추세 없음": med2.loc["뚜렷한 추세 없음", col],
                 "지속 증가 (참고)": med2.loc["지속 증가", col], "끊긴 29가구 (참고)": gone[col],
                 "델타 (2-1 쪽 +)": res["delta"] * (1 if alt == "greater" else -1), "한쪽 p-value": res["p-value"]})
compare2 = pd.DataFrame(rows).set_index("지표")
with pd.option_context("display.float_format", "{:,.3f}".format):
    display(compare2)
````

- ① 가구 물가 변화: 1년 사이 지속 감소 가구가 사던 상품은 2.19%, 뚜렷한 추세 없음 가구가 사던 상품은 2.24% 올라 거의 같음. 델타 −0.013으로 오히려 반대 방향이고, p-value 0.656
- ② 사라진 상품 몫: 지속 감소 11.7%, 뚜렷한 추세 없음 11.3%. 두 묶음 모두 줄기 전에 사던 상품의 약 11%가 1년 뒤 팔리지 않았음. 델타 0.034, p-value 0.153으로 우연과 구별되지 않음
- ③ 행사 주 비율 변화: 두 묶음 모두 1%p 안쪽으로 조금 줄었고, 중앙값은 지속 감소 가구가 오히려 덜 줄었음(−0.87%p 대 −0.98%p). 델타 0.017, p-value 0.303
- 5·6분기에 오지 않은 29가구도 ① 2.57%, ② 11.5%, ③ −0.59%p로 다른 가구와 비슷함. 끊긴 가구가 사던 상품이 특별히 나빠진 것도 아님

### 7.4 가구 크기 단계마다 같은 방향인가

6.4절과 같이 가구 크기 단계마다 델타를 다시 구함. 양수면 그 단계에서도 지속 감소 가구가 더 많이 나빠진 쪽.

````python
# 가구 크기 단계별 델타 (2-1 쪽 +)와 양수인 단계 수
tier_delta2 = pd.DataFrame({
    col: {t: my_stats.cliffs_delta(dec2.loc[dec2["크기 단계"].eq(t), col].dropna(),
                                   flat2.loc[flat2["크기 단계"].eq(t), col].dropna()).loc[alt, "delta"] * (1 if alt == "greater" else -1)
          for t in tiers}
    for col, alt in expect2.items()}).T
tier_delta2["같은 방향 단계 (5개 중)"] = (tier_delta2[tiers] > 0).sum(axis=1)
with pd.option_context("display.float_format", "{:,.3f}".format):
    display(tier_delta2)
````

## 8. 1-2·1-3·2-2·2-3 어떻게 줄었나

직접 확인할 수 있는 가설은 6·7절에서 봄(1-1 약함, 2-1 뒷받침되지 않음). 남은 넷은 식구 수, 외식, 다른 가게, 이사처럼 마트 밖에서 생긴 일이라 거래 기록에 원인이 남지 않음. 대신 4.1절 표대로 원인마다 '줄어든 모양'과 '줄어든 범위'가 다름. 이 절은 지속 감소 가구가 어떤 모양과 범위로 줄었는지 보고, 그 모양에 맞는 가설을 고름.

머리가 길어 순서를 먼저 적음: 가구 A·B 예시 → 네 가지 질문과 가설마다 예상되는 모습(표) → 질문마다 재는 방법(①∼④) → 비교할 가구 → 판정 방법(결과 보기 전에 정함).

예를 들어 두 가구가 1년 사이 똑같이 주 40달러에서 25달러로 줄었다고 하자.

- 가구 A는 40, 39, 37, 36 … 25달러처럼 매달 조금씩 줄었고, 고기·채소는 줄었지만 세제·휴지는 그대로 삼. 외식이 늘어 집밥 재료만 덜 사게 된 모습(1-3)
- 가구 B는 40달러로 이어지다가 어느 달부터 25달러로 뚝 떨어졌고, 오는 횟수는 그대로인데 한 번에 담는 상품 수가 줄었음. 식구가 줄어 한 번에 덜 담게 된 모습(1-2)

같은 15달러 감소라도 A와 B의 원인은 다르게 읽힘. 이렇게 네 가지 질문으로 가설을 가름.

| 질문 | 재는 값 | 1-2 식구 감소 | 1-3 외식·배달 | 2-2 다른 곳이 좋아짐 | 2-3 생활 반경이 바뀜 |
|---|---|---|---|---|---|
| ① 모양: 서서히 줄었나, 어느 때 한 번에 줄었나 | 계단 점수 (아래 설명) | 한 번에 | 서서히 | 가르지 못함 (서서히·한 번에 모두 가능) | 한 번에 |
| ② 범위: 식사 재료만 줄었나, 생활용품도 줄었나 | 식사 재료 몫 변화(%p) | 함께 줄어듦 | 식사 재료가 더 줄어듦 | 생활용품이 더 줄어듦 | 함께 줄어듦 |
| ③ 한 번에 줄어든 가구는 무엇이 줄었나 | 구매액이 뚝 떨어진 때 앞뒤로, 오는 횟수(방문 수)와 한 번에 담는 상품 수가 각각 얼마나 줄었나 | 장바구니당 상품 수 | | | 방문 수 |
| ④ 같은 매장 손님들도 같이 줄었나 | 지속 감소 가구가 유난히 많은 매장이 있나, 한 가구가 뚝 떨어질 때 같은 매장 손님들도 같이 줄었나 | 따로 (그 가구만 줄어듦) | 따로 (그 가구만 줄어듦) | 함께 (매장 손님들이 같이 줄어듦) | 따로 (그 가구만 줄어듦) |

- 표에 1-1(돈을 아낌)이 없는 이유: 돈을 아끼려는 가구는 특정 품목이 아니라 이것저것 조금씩 덜 삼. 그래서 매달 조금씩(① 서서히), 식사 재료와 생활용품이 비슷하게(② 함께), 그 가구만(④ 따로) 줄어듦. 그런데 다른 가게로 조금씩 옮겨 가는 가구(2-2)도 매장 손님들과 함께 줄지 않고 생활용품이 특별히 더 줄지 않았다면 똑같은 모습이라, 이 모습만으로는 둘을 가를 수 없음. 그래서 이런 가구가 몇 가구인지만 참고로 보여 주고, 1-1 판정은 싼 상품으로 바꿨는지를 직접 본 6절 결과를 그대로 씀
- 2-2(다른 곳이 좋아짐)는 흔적이 두 갈래로 나타날 수 있음
  - 매장 근처에 새 가게가 생긴 경우: 그 매장 손님 여럿이 비슷한 때에 같이 줄어듦(④ 함께)
  - 다른 가게나 온라인이 특정 품목을 더 싸게 파는 경우: 휴지·세제 같은 생필품은 어디서 사도 같은 물건이라 먼저 그쪽으로 옮겨 감. 그래서 식사 재료보다 생활용품이 더 줄어듦(② 생활용품이 더 줄어듦)
  - 두 경우는 따로 일어날 수 있어, 둘 가운데 하나만 나타나도 2-2의 흔적으로 봄
- 식구 수가 바뀌었는지를 직접 보여 주는 기록은 없음. 가구원 수 같은 인구통계는 801가구에만 있고, 한 시점에 한 번 적은 값이라 식구가 줄었는지는 알 수 없음

#### ① 서서히 줄었나, 한 번에 줄었나 (계단 점수)

가구마다 84주를 4주씩 21구간으로 나눈 구매액(5절에서 계절에 따른 오르내림을 뺀 값)을 놓고, 아래 두 선 가운데 어느 쪽에 더 가까운지 봄.

- 직선: 처음부터 끝까지 일정하게 내려가는 선. 가구 A(40, 39, 37 … 25달러)처럼 서서히 줄면 이 선에 가까움
- 계단: 어느 때까지는 높은 칸, 그 뒤로는 낮은 칸인 두 칸짜리 선. 가구 B(40, 40, 40 … 25, 25달러)처럼 한 번에 줄면 이 선에 가까움. 두 칸이 바뀌는 자리를 '뚝 떨어진 때'라고 부름
  - 두 칸이 바뀌는 자리는 정해져 있지 않으므로, 가능한 자리마다 계단을 그려 보고 실제 값과 가장 덜 어긋나는 곳을 고름. 다만 맨 앞이나 맨 끝을 고르면 한쪽 칸에 값이 한두 개만 남아, 마지막 달에 우연히 덜 산 것만으로도 '뚝 떨어졌다'고 잡힘. 그래서 앞과 뒤에 각각 3구간(12주, 약 석 달) 이상 남는 자리만 고름. 낮아진 상태가 적어도 석 달은 이어져야 뚝 떨어졌다고 보는 셈

계단 점수는 계단이 직선보다 실제 값에 얼마나 더 가까운지를 나타냄. 직선으로 그렸을 때 실제 값과 어긋난 정도를 100이라 하면, 계단으로 그려 어긋난 정도가 30으로 줄 때 계단 점수는 0.7. 클수록 계단에 가깝고, 0 이하면 직선이 더 잘 맞음(계산: 1 − 계단 오차 제곱합 ÷ 직선 오차 제곱합).

그런데 계단은 바뀌는 자리를 골라 그릴 수 있어서, 서서히 줄어든 가구에도 어느 정도 맞음. 예를 들어 가구 A는 매달 조금씩 줄었지만 구매액은 달마다 들쭉날쭉함(40, 42, 37, 39, 35, 38 …). 이런 가구도 우연히 낮았던 달이 뒤쪽에 몰린 자리를 골라 계단을 그리면, 직선보다 조금은 더 잘 맞을 수 있음. 그래서 계단 점수가 0보다 크다는 것만으로는 한 번에 줄었다고 할 수 없음.

그래서 가구마다 '이 가구가 정말 서서히 줄었다면 계단 점수가 얼마까지 나올 수 있나'를 직접 만들어 보고, 실제 점수가 그보다 높을 때만 한 번에 줄었다고 봄.

1. 가구의 21구간 값을 '직선'과 '직선에서 벗어난 만큼(들쭉날쭉한 부분)'으로 나눔. 예를 들어 직선이 40인 달에 실제가 42면 +2, 직선이 39인 달에 실제가 37이면 −2
2. 들쭉날쭉한 부분(+2, −2 …)을 카드처럼 섞어 직선 위에 다시 얹음. 직선은 그대로라 서서히 줄어든 모양이고, 들쭉날쭉한 정도도 실제와 같음. 다만 낮은 달이 한쪽에 몰려 있던 흔적은 섞이면서 사라짐. 이것이 '서서히 줄었다면 나올 법한 가짜 가구'
3. 섞기를 200번 해서 가짜 가구 200개를 만들고, 각각 계단 점수를 구함
4. 실제 점수가 가짜 200개 가운데 높은 쪽 10개(상위 5%) 안에 들 만큼 높으면 '한 번에 줄어든 가구'로 봄. 서서히 줄어든 가구라면 그렇게 높은 점수가 나오기 어렵기 때문

모든 가구가 실제로는 서서히 줄었더라도, 20가구 가운데 1가구(5%)쯤은 우연히 이 기준을 넘음. 341가구라면 17가구쯤. 그래서 '한 번에' 가구가 5%를 얼마나 넘는지를 봄. 이보다 훨씬 많이 잡히면 실제로 한 번에 줄어든 가구가 있다는 뜻.

- 한계: 21구간뿐이라 크게 떨어진 가구는 잘 잡히지만, 조금 떨어진 가구는 들쭉날쭉함에 묻혀 '서서히'로 남기 쉬움. 그래서 한 번에 줄어든 가구 수는 실제보다 적게 나오는 쪽

#### ② 식사 재료만 줄었나, 생활용품도 줄었나 (식사 재료 몫 변화)

가구가 식사 재료와 생활용품에 쓴 돈 가운데 식사 재료가 차지하는 몫(%)을 1·2분기와 5·6분기에 각각 구하고, 얼마나 바뀌었는지(5·6분기 − 1·2분기) 봄.

예를 들어 가구 A가 1·2분기에 식사 재료 60달러, 생활용품 40달러를 샀다면 식사 재료 몫은 60%.

- 외식이 늘어 5·6분기에 식사 재료 30달러, 생활용품 40달러를 샀다면 몫은 43%로 17%p 내려감 → 1-3 쪽
- 생필품을 다른 곳에서 사게 되어 식사 재료 60달러, 생활용품 15달러를 샀다면 몫은 80%로 20%p 올라감 → 2-2 쪽
- 둘 다 절반으로 줄면(30달러, 20달러) 몫은 60% 그대로 → 1-2·2-3처럼 함께 줄어든 모습

두 묶음은 아래와 같이 정함.

- 식사 재료: 채소·과일, 고기, 생선 같은 신선 식품(부서 `DEPARTMENT` = PRODUCE, MEAT, MEAT-PCKGD, SEAFOOD, SEAFOOD-PCKGD). 집에서 요리할 때 사는 재료
- 생활용품: 세제, 휴지·키친타월, 비닐·호일, 비누 등(품목 `COMMODITY_DESC` = LAUNDRY DETERGENTS, LAUNDRY ADDITIVES, DISHWASH DETERGENTS, HOUSEHOLD CLEANG NEEDS, BATH TISSUES, PAPER TOWELS, PAPER HOUSEWARES, FD WRAPS/BAGS/TRSH BG, DISPOSIBLE FOILWARE, SOAP - LIQUID & BAR). 외식이 늘어도 집에서 계속 쓰는 물건
- 과자·음료·가공식품처럼 두 묶음에 들지 않는 상품은 뺌. 외식이 늘어서도, 다른 가게로 옮겨서도 줄 수 있어 둘을 가르는 데 도움이 안 됨

#### ③ 오는 횟수가 줄었나, 한 번에 담는 양이 줄었나

①에서 '한 번에 줄어든 가구'로 잡힌 가구만, 뚝 떨어진 때의 앞과 뒤를 견줌. 구매액은 '오는 횟수 × 한 번에 담는 양'이라, 둘 가운데 어느 쪽이 더 줄었는지 보면 이유를 짐작할 수 있음.

- 오는 횟수: 4주에 몇 번 왔나(장바구니 수)
- 한 번에 담는 양: 한 번 올 때 상품을 몇 줄 샀나(영수증 한 장에 찍힌 상품 줄 수)

예를 들어 가구 B가 떨어지기 전에는 4주에 8번 와서 한 번에 상품 20개를 담았다고 하자.

- 떨어진 뒤에도 8번 오는데 12개씩 담았다면, 담는 양이 줄었음 → 1-2 쪽(식구가 줄어 덜 담음)
- 떨어진 뒤 5번 와서 20개씩 담았다면, 오는 횟수가 줄었음 → 2-3 쪽(이사 등으로 매장이 멀어짐)
- 둘 다 줄었으면 더 크게 줄어든 쪽으로 봄. 예를 들어 8번에서 5번(38% 감소), 20개에서 15개(25% 감소)로 줄었다면 오는 횟수 쪽. 줄어든 비율은 로그 변화율로 견줌

#### ④ 같은 매장 손님들도 같이 줄었나

가구마다 가장 많이 산 매장(1·2분기 구매액 기준, 7절과 같음)을 그 가구의 '주 매장'으로 보고, 두 가지를 봄.

**지속 감소 가구가 유난히 많은 매장이 있나 (매장 몰림)**

전체로 보면 대상 가구 약 7가구 가운데 1가구가 지속 감소 가구(370 ÷ 2,405 = 15%). 근처에 다른 가게가 생긴 매장이 있다면 그 매장 손님 여럿이 함께 줄어, 그 매장의 지속 감소 비율이 15%보다 훨씬 높아짐. 예를 들어 손님 40가구인 매장에서 지속 감소 가구가 6가구면 보통이고, 20가구면 몰린 것.

- 손님이 적은 매장은 우연만으로도 비율이 크게 흔들림. 손님 10가구인 매장은 지속 감소 가구가 1가구만 늘어도 비율이 10%p 뛰어, 보통이 15%여도 30%(3가구)쯤은 쉽게 나옴. 그래서 '몰렸다'고 하려면 우연으로 나올 만한 정도와 견줘야 함
  - 우연의 정도는 섞어서 만듦. 가구마다 붙은 유형 이름표(지속 감소 / 뚜렷한 추세 없음 / 지속 증가)를 떼어 무작위로 다시 붙임. 그러면 지속 감소 가구 수(370)는 그대로이고, 어느 매장에 있는지는 매장과 상관없어짐. 이것을 1,000번 해서 '매장과 상관없이 줄었다면' 나올 몰림을 만들고, 실제 몰림이 그보다 큰지 봄
  - 이름표는 같은 크기 단계 안에서만 섞음. 지속 감소 가구가 많이 사는 가구나 적게 사는 가구 한쪽에 더 많고, 그런 가구가 특정 매장에 모여 있다면, 매장 사정이 아니라 가구 크기 때문에 몰려 보일 수 있음. 같은 단계 안에서만 섞으면 이 몫이 섞은 결과에도 똑같이 들어가 빠짐
  - 손님이 10가구 이상인 매장만 봄. 그보다 적으면 비율이 너무 흔들려 읽기 어려움
- 몰림의 크기는 '우연을 넘어 몰린 가구 수'로 잼
  - 매장마다 보통이면 있을 지속 감소 가구 수(섞은 1,000번의 평균, 대략 손님 수 × 15%)보다 실제로 몇 가구 더 있는지 셈. 위 예시 매장은 보통 6가구인데 20가구라 14가구가 넘침. 보통보다 적은 매장은 0으로 둠. 이렇게 넘친 수를 모든 매장에서 더함
  - 그런데 섞은 결과에서도 우연히 보통보다 조금씩 넘치는 매장이 생김. 예를 들어 섞은 결과에서 넘친 수의 합이 평균 40가구라면, 그 40가구는 우연으로도 생기는 몫. 실제 합이 70가구면 70 − 40 = 30가구가 '우연을 넘어 몰린 가구 수'

**한 가구가 뚝 떨어질 때 같은 매장 손님들도 같이 줄었나 (함께 / 따로)**

①에서 한 번에 줄어든 가구마다, 같은 주 매장의 다른 손님들이 산 금액이 뚝 떨어진 때의 앞 12주보다 뒤 12주에 얼마나 바뀌었는지를 마트 전체의 변화와 견줌(매장 변화 ÷ 마트 전체 변화). 12주는 ①에서 '낮아진 상태가 석 달은 이어져야 뚝 떨어졌다'고 정한 것과 같은 길이.

이 값이 그 매장으로서도 드물게 낮으면 '함께', 아니면 '따로'로 봄. 매장마다 원래 흔들리는 폭이 달라서(손님이 적은 매장은 아무 일이 없어도 석 달 사이 10%쯤은 쉽게 오르내림), 기준을 매장마다 따로 만듦.

1. 그 매장에 대해, 84주 안에서 한 주씩 옮겨 가며 앞뒤 12주를 잡을 수 있는 모든 자리(61자리)마다 같은 값을 구함. 그 매장이 평소 석 달 사이에 얼마나 오르내리는지가 됨
2. 뚝 떨어진 때의 값이 61자리 가운데 가장 크게 줄어든 쪽 5%(3자리) 안에 들면 '함께'

- 예를 들어 어느 매장이 평소에는 마트 전체와 견줘 석 달 사이 ±5% 안에서 오르내리는데, 뚝 떨어진 때에는 12% 줄었다면 '함께'. 그 가구 혼자의 사정보다 매장 쪽 사정일 가능성이 큼. 같은 12% 감소라도 평소 ±15%씩 오르내리는 작은 매장이라면 드문 일이 아니라 '따로'
- 마트 전체와 견주므로, 계절이나 마트 전체의 변화 때문에 함께 줄어 보이는 것은 빠짐
- 5%는 ①의 '한 번에' 기준, 바로 위 매장 몰림의 우연 기준과 같은 값

#### 비교할 가구

- ①·③(모양, 무엇이 줄었나): 지속 감소 가구 가운데 5·6분기에도 산 341가구로 봄. 끝까지 오지 않은 29가구는 마지막에 0으로 떨어져 '한 번에, 오는 횟수가 줄어듦'으로 잡히기 쉬움. 그런데 이 가구들이 마트를 떠났는지, 이사했는지, 다른 곳으로 옮겼는지는 가를 수 없어(4.1절) 따로 떼어 참고로만 봄
- ②(범위): 6절과 같이 두 시기 모두 산 지속 감소 가구와 뚜렷한 추세 없음 가구를 견줌
- ④ 매장 몰림: 매장마다 손님 가운데 지속 감소 가구가 얼마나 되는지 보는 것이라 대상 가구 2,405가구 전체를 씀

#### 판정 방법 (결과를 보기 전에 정함)

질문에 따라 판정 방법이 두 가지임.

**②: 줄지 않은 가구와 견줌**

②는 뚜렷한 추세 없음 가구도 같은 값을 잴 수 있음. '줄지 않은 가구도 이 정도는 바뀌었나'를 기준으로 삼을 수 있어, 6·7절과 같은 방법을 씀.

| 지표 판정 (②) | 기준 |
|---|---|
| 뒷받침됨 | 가설 쪽으로 우연과 구별되고(한쪽 맨-휘트니 검정 p < 0.05), 델타 0.147 이상이며, 가구 크기 다섯 단계 가운데 네 단계 이상에서 같은 방향 |
| 약함 | 가설 쪽으로 우연과 구별되지만, 델타가 0.147에 못 미치거나 같은 방향인 단계가 세 단계 이하 |
| 뒷받침되지 않음 | 우연과 구별되지 않거나 반대 방향 |

**①·③·④: 우연과 견주고, 몇 가구인지 셈**

①·③·④는 견줄 상대가 없음. '뚝 떨어진 때'는 줄어든 가구에만 있어서, 줄지 않은 가구에게는 '한 번에 줄었나'를 물을 수 없음. 대신 두 가지를 봄.

- 우연보다 많은가: 모두 서서히 줄었거나(① 5%) 매장과 상관없이 줄었을 때(④ 섞은 결과) 나올 수보다 많은지 검정함
- 얼마나 많은가: 실제 비율이 우연일 때의 비율보다 얼마나 많은지를 코언의 h로 잼

코언의 h는 두 비율의 차이를 재는 표준 척도로, 0.2가 흔히 쓰는 '작은 효과'의 하한(Cohen, 1988). 6·7절의 델타 0.147도 같은 '작은 효과'에 맞춘 값이라, 두 판정 방법이 같은 잣대를 씀. h는 비율이 0%나 100%에 가까울수록 같은 %p 차이를 더 크게 침. 예를 들어 5%에서 10%로 늘면 h 0.19로 작은 효과(0.2)에 가깝지만, 50%에서 55%로 늘면 h 0.10. 우연일 때가 5%처럼 낮은 비교에 맞음.

- ①과 1-2·2-3: 우연일 때 비율은 5%. 가설의 모습을 모두 갖춘 가구가 지속 감소 가구(341가구)의 약 10.2% 이상이면 h 0.2를 넘음. 1-2·2-3은 ③·④ 조건이 더 붙어 우연일 때 비율이 5%보다 낮지만, 5%로 두어 기준을 엄격한 쪽으로 잡음
- ④ 매장 몰림: 실제로 넘친 가구 비율과 섞었을 때 넘친 가구 비율(평균)을 견줌. 위 예시 숫자라면 70 ÷ 370 = 19%와 40 ÷ 370 = 11%로 h 0.23

| 지표 판정 (①·③·④) | 기준 |
|---|---|
| 뒷받침됨 | 우연보다 많고(① 한쪽 이항 검정, ④ 섞기 검정 p < 0.05), 우연일 때보다 h 0.2 이상 많음 |
| 약함 | 우연보다 많지만, h가 0.2에 못 미침 |
| 뒷받침되지 않음 | 우연과 구별되지 않음 |

- 가구 크기 단계별 비율도 참고로 함께 보여 줌

### 8.1 ① 서서히 줄었나, 한 번에 줄었나

8절 머리의 계단 점수와 섞기 기준으로 지속 감소 가구를 '한 번에 / 서서히'로 나눔. 5·6분기에도 산 341가구로 판정하고, 끊긴 29가구는 따로 보여 줌.

````python
# 계절분을 뺀 4주 구간 구매액 (5절의 block ÷ season), 지속 감소 가구만
adj = pd.DataFrame(block / season, index=hh.index)
dec_hh = hh.index[hh["유형"].eq("지속 감소")]

# 계단 점수와 '한 번에' 판정: 앞뒤 3구간 이상 남는 경계, 직선 둘레 오르내림을 낱개로 섞은 가짜 가구 200개
step = my_ts.step_vs_line(adj.loc[dec_hh], min_seg=3, block=1, n_fake=200, random_state=0)
shape = step[["계단 점수", "경계 위치", "가짜가 같거나 높은 수", "모양"]].rename(columns={"경계 위치": "뚝 떨어진 구간"})
shape["끊긴 가구"] = hh.loc[dec_hh, "5·6분기"].eq(0)
shape["크기 단계"] = hh.loc[dec_hh, "크기 단계"]
live = shape[~shape["끊긴 가구"]]
gone_shape = shape[shape["끊긴 가구"]]
````

````python
# 우연(5%)과 견준 한 번에 가구 비율: 한쪽 이항 검정과 코언의 h
def vs_chance(k, n, chance=0.05):
    res = my_stats.prop_vs_chance(k, n, chance).loc["greater"]
    return {"가구": n, "해당 가구": k, "비율 %": k / n * 100, "우연 %": chance * 100,
            "한쪽 p-value": res["p-value"], "코언의 h": res["h"]}

shape_test = pd.DataFrame([vs_chance(int(live["모양"].eq("한 번에").sum()), len(live)),
                           vs_chance(int(gone_shape["모양"].eq("한 번에").sum()), len(gone_shape))],
                          index=["341가구 (판정)", "끊긴 29가구 (참고)"])
with pd.option_context("display.float_format", "{:,.3f}".format):
    display(shape_test)

# 크기 단계별 한 번에 가구 비율 (참고, 341가구)
live.groupby("크기 단계", observed=True)["모양"].agg(가구="size", 한_번에=lambda s: s.eq("한 번에").sum()).assign(
    **{"한 번에 %": lambda d: d["한_번에"] / d["가구"] * 100}).rename(columns={"한_번에": "한 번에"}).T
````

표의 '비율 %'는 가구 가운데 '한 번에 줄어든 가구'로 잡힌 몫이고, '우연 %'는 모두 서서히 줄었을 때 우연히 잡힐 몫. 코언의 h가 0.2 이상이면 우연보다 '작은 효과' 이상으로 많은 것.

- 5·6분기에도 산 341가구 가운데 88가구(25.8%)가 한 번에 줄었음. 우연이면 5%(약 17가구)라 다섯 배쯤 많고, p-value는 0.001 아래, h는 0.615
- 크기 단계마다 18.0∼30.7%로, 다섯 단계 모두 5%를 크게 넘음. 많이 사는 가구든 적게 사는 가구든 비슷하게 나타남
- 끊긴 29가구는 16가구(55.2%)만 한 번에로 잡힘. 끝까지 오지 않은 가구라고 모두 계단 모양으로 끊긴 것은 아님
- 다만 이 비율은 많게 잡혔을 수 있음. 실제 구매액은 이웃한 구간끼리 비슷하게 움직이는데(한 달 많이 사면 다음 달도 많이 사는 식), 가짜 가구는 순서를 낱개로 섞어 이 성질이 없음. 가짜 가구가 실제보다 매끈하면 실제 가구가 '한 번에'로 더 자주 잡힘

### 8.2 ② 식사 재료만 줄었나, 생활용품도 줄었나

가구마다 식사 재료 몫(식사 재료 ÷ (식사 재료 + 생활용품), %)의 1·2분기 → 5·6분기 변화를 구하고, 지속 감소 가구와 뚜렷한 추세 없음 가구를 견줌. 델타는 가설 쪽이 양수가 되게 부호를 맞춤(1-3 쪽 = 더 내려감, 2-2 쪽 = 더 올라감). 5·6분기에 사지 않은 29가구는 5·6분기 몫을 잴 수 없어 빠짐.

````python
# 식사 재료(신선 식품 부서)와 생활용품(10개 품목), 가구·시기별 구매액
fresh = ["PRODUCE", "MEAT", "MEAT-PCKGD", "SEAFOOD", "SEAFOOD-PCKGD"]
home = ["LAUNDRY DETERGENTS", "LAUNDRY ADDITIVES", "DISHWASH DETERGENTS", "HOUSEHOLD CLEANG NEEDS", "BATH TISSUES",
        "PAPER TOWELS", "PAPER HOUSEWARES", "FD WRAPS/BAGS/TRSH BG", "DISPOSIBLE FOILWARE", "SOAP - LIQUID & BAR"]
xr = xq[xq["household_key"].isin(hh.index) & xq["분기"].isin([1, 2, 5, 6])].copy()
xr["묶음"] = np.select([xr["DEPARTMENT"].isin(fresh), xr["COMMODITY_DESC"].isin(home)], ["식사 재료", "생활용품"], default="")
xr["시기"] = np.where(xr["분기"] <= 2, "1·2분기", "5·6분기")
amt = xr[xr["묶음"].ne("")].pivot_table(index="household_key", columns=["시기", "묶음"], values="SALES_VALUE",
                                         aggfunc="sum", fill_value=0)

# 식사 재료 몫(%)과 변화(%p), 두 시기 모두 두 묶음 합이 0보다 큰 가구만
share = pd.DataFrame({t: amt[t]["식사 재료"] / (amt[t]["식사 재료"] + amt[t]["생활용품"]) * 100
                      for t in ["1·2분기", "5·6분기"]})
share = share.replace([np.inf, -np.inf], np.nan).dropna()
share["변화"] = share["5·6분기"] - share["1·2분기"]
share = share.join(hh[["유형", "크기 단계"]])
print("두 시기 모두 값이 있는 가구:", share["유형"].value_counts().reindex(type_order).to_dict())

# 유형별 식사 재료 몫과 변화의 중앙값 (%, %p)
share.groupby("유형")[["1·2분기", "5·6분기", "변화"]].median().reindex(type_order)
````

값은 유형별 중앙값. 식사 재료 몫 86%는 식사 재료와 생활용품에 쓴 돈 100달러 가운데 86달러가 식사 재료라는 뜻. 변화가 양수면 생활용품이 더 많이 줄었고, 음수면 식사 재료가 더 많이 줄었음.

- 두 시기 모두 값이 있는 가구는 지속 감소 312, 뚜렷한 추세 없음 1,436, 지속 증가 464가구
- 지속 감소 가구는 식사 재료 몫이 1·2분기 86.1%, 5·6분기 89.5%이고, 가구마다 잰 변화의 중앙값은 +2.0%p. 생활용품이 식사 재료보다 더 많이 줄었음
- 뚜렷한 추세 없음 가구는 −0.1%p로 거의 그대로이고, 지속 증가 가구는 −1.7%p
- 외식이 늘어 집밥 재료를 덜 산 것(1-3)이라면 식사 재료 몫이 내려가야 하는데, 지속 감소 가구는 반대로 올라감

````python
# 두 방향으로 견줌: 1-3 쪽(더 내려감, less)과 2-2 쪽(더 올라감, greater)
dec3 = share[share["유형"].eq("지속 감소")]
flat3 = share[share["유형"].eq("뚜렷한 추세 없음")]
expect3 = {"1-3 쪽 (식사 재료 몫이 더 내려감)": "less", "2-2 쪽 (식사 재료 몫이 더 올라감)": "greater"}

rows, tier_rows = [], {}
for name, alt in expect3.items():
    sign = 1 if alt == "greater" else -1
    res = my_stats.cliffs_delta(dec3["변화"], flat3["변화"]).loc[alt]
    tier_rows[name] = {t: my_stats.cliffs_delta(dec3.loc[dec3["크기 단계"].eq(t), "변화"],
                                                flat3.loc[flat3["크기 단계"].eq(t), "변화"]).loc[alt, "delta"] * sign
                       for t in tiers}
    rows.append({"방향": name, "델타 (가설 쪽 +)": res["delta"] * sign, "한쪽 p-value": res["p-value"],
                 "같은 방향 단계 (5개 중)": sum(v > 0 for v in tier_rows[name].values())})
compare3 = pd.DataFrame(rows).set_index("방향")

# 8절 머리의 ② 지표 판정
ok_p = (compare3["한쪽 p-value"] < 0.05) & (compare3["델타 (가설 쪽 +)"] > 0)
compare3["판정"] = np.select([ok_p & (compare3["델타 (가설 쪽 +)"] >= 0.147) & (compare3["같은 방향 단계 (5개 중)"] >= 4), ok_p],
                            ["뒷받침됨", "약함"], default="뒷받침되지 않음")
with pd.option_context("display.float_format", "{:,.3f}".format):
    display(compare3)
    display(pd.DataFrame(tier_rows).T.rename_axis("단계별 델타 (가설 쪽 +)"))
````

델타는 지속 감소 가구가 가설 쪽으로 더 바뀌었으면 양수. 0.147 이상이 '작은 효과'의 하한.

- 1-3 쪽(식사 재료 몫이 더 내려감): 델타 −0.125로 반대 방향이고, 다섯 단계 모두 음수. 뒷받침되지 않음
- 2-2 쪽(식사 재료 몫이 더 올라감, 곧 생활용품이 더 줄어듦): 델타 0.125, p-value 0.001 아래로 우연과 구별되고, 다섯 단계 모두 같은 방향(0.068∼0.222). 다만 델타가 0.147에 못 미쳐 약함
- 정리하면 지속 감소 가구는 생활용품을 식사 재료보다 조금 더 많이 줄였음. 생필품처럼 다른 곳에서 사기 쉬운 품목부터 옮겨 간 2-2의 모습과 방향은 맞지만, 차이는 작음

### 8.3 ③ 한 번에 줄어든 가구는 무엇이 줄었나

8.1에서 '한 번에'로 잡힌 가구(341가구 가운데)만, 뚝 떨어진 구간의 앞과 뒤를 견줌. 오는 횟수는 4주당 방문 수(장바구니 수), 한 번에 담는 양은 장바구니당 상품 줄 수. 변화는 로그 변화율(뒤 ÷ 앞의 로그)로 잼. 비율을 그대로 견주면 절반으로 준 것(0.5배)과 두 배로 는 것(2배)이 1에서 떨어진 거리가 달라 보이는데, 로그를 쓰면 둘이 같은 크기(−0.69와 +0.69)가 되어 방문 수와 담는 양의 변화를 공정하게 견줄 수 있음. 둘 가운데 더 크게 줄어든 쪽으로 나눔.

````python
# 한 번에 줄어든 가구의 거래를 뚝 떨어진 주(경계 구간의 첫 주) 앞뒤로 나눔
jump = live[live["모양"].eq("한 번에")].copy()
jump["경계 주"] = 18 + 4 * jump["뚝 떨어진 구간"]
xj = xq[xq["household_key"].isin(jump.index)].merge(jump[["경계 주"]], left_on="household_key", right_index=True)
xj["앞뒤"] = np.where(xj["WEEK_NO"] < xj["경계 주"], "앞", "뒤")
g = (xj.groupby(["household_key", "앞뒤"]).agg(방문=("BASKET_ID", "nunique"), 줄=("BASKET_ID", "size"))
       .unstack("앞뒤").reindex(jump.index).fillna(0))

# 4주당 방문 수, 장바구니당 상품 줄 수와 로그 변화율
ch = pd.DataFrame({
    "4주당 방문 앞": g[("방문", "앞")] / (jump["경계 주"] - 18) * 4,
    "4주당 방문 뒤": g[("방문", "뒤")] / (102 - jump["경계 주"]) * 4,
    "장바구니당 줄 앞": g[("줄", "앞")] / g[("방문", "앞")],
    "장바구니당 줄 뒤": g[("줄", "뒤")] / g[("방문", "뒤")].replace(0, np.nan),
})
ch["방문 로그 변화"] = np.log(ch["4주당 방문 뒤"].replace(0, np.nan) / ch["4주당 방문 앞"])
ch["줄 수 로그 변화"] = np.log(ch["장바구니당 줄 뒤"] / ch["장바구니당 줄 앞"])
jump = jump.join(ch)
jump["더 줄어든 쪽"] = np.where(jump["줄 수 로그 변화"] < jump["방문 로그 변화"], "담는 양", "오는 횟수")
print("뚝 떨어진 뒤 방문이 없는 가구:", int(g[("방문", "뒤")].eq(0).sum()))

# 더 줄어든 쪽별 가구 수와 중앙값
summary3 = jump.groupby("더 줄어든 쪽").agg(
    가구=("모양", "size"), **{c: (c, "median") for c in ["4주당 방문 앞", "4주당 방문 뒤", "장바구니당 줄 앞", "장바구니당 줄 뒤"]})
summary3
````

값은 가구 수와 중앙값. 4주당 방문 8.5 → 1.8은 뚝 떨어지기 전 4주에 8.5번 오던 가구가 뒤에는 1.8번 온다는 뜻.

- 한 번에 줄어든 88가구 가운데 75가구는 오는 횟수가 더 줄었고, 13가구만 한 번에 담는 양이 더 줄었음
- 오는 횟수 쪽 가구는 방문이 4주에 8.5번에서 1.8번으로 약 5분의 1이 됨. 한 번에 담는 양은 8.0줄에서 5.9줄로 덜 줄었음. 장보기 대부분을 이 마트 밖으로 옮긴 모습
- 담는 양 쪽 13가구는 장바구니당 14.1줄에서 6.4줄로 절반 아래가 됐지만, 방문도 5.8번에서 3.0번으로 줄었음
- 감소의 90%가 방문 감소였던 00 노트북 9.5절 결과와 같이, 한 번에 줄어든 가구도 대부분 오는 횟수가 줄었음

### 8.4 ④ 같은 매장 손님들도 같이 줄었나

먼저 지속 감소 가구가 특정 매장에 몰렸는지(매장 몰림)를 대상 가구 전체로 보고, 다음으로 8.1의 한 번에 줄어든 가구마다 같은 매장 손님들도 같은 때 줄었는지(함께 / 따로) 봄.

````python
# 대상 가구가 10가구 이상인 주 매장만
store_df = hh[["유형", "크기 단계"]].join(main_store)
sd = store_df[store_df["주 매장"].map(store_df["주 매장"].value_counts()) >= 10]
store_code, store_ids = pd.factorize(sd["주 매장"])
is_dec = sd["유형"].eq("지속 감소").to_numpy()
actual = np.bincount(store_code, weights=is_dec, minlength=len(store_ids))

# 같은 크기 단계 안에서 유형 이름표를 1,000번 섞어 매장별 지속 감소 가구 수
# helpers에 섞기 검정 함수가 없어 numpy로 직접 계산
rng = np.random.default_rng(1)
n_shuf = 1000
tier_idx = [np.flatnonzero(sd["크기 단계"].eq(t).to_numpy()) for t in tiers]
shuf = np.empty((n_shuf, len(store_ids)))
for i in range(n_shuf):
    lab = is_dec.copy()
    for idx in tier_idx:
        lab[idx] = rng.permutation(lab[idx])
    shuf[i] = np.bincount(store_code, weights=lab, minlength=len(store_ids))

# 매장마다 보통(섞은 평균)보다 넘친 수의 합: 실제와 섞은 결과
expected = shuf.mean(axis=0)
excess = np.clip(actual - expected, 0, None).sum()
excess_shuf = np.clip(shuf - expected, 0, None).sum(axis=1)
n_dec_big = int(is_dec.sum())
crowd = pd.Series({
    "매장 수": len(store_ids), "가구": len(sd), "지속 감소 가구": n_dec_big,
    "넘친 가구 (실제)": excess, "넘친 가구 (섞은 평균)": excess_shuf.mean(),
    "우연을 넘어 몰린 가구": excess - excess_shuf.mean(),
    "섞기 p-value": ((excess_shuf >= excess).sum() + 1) / (n_shuf + 1),
    "코언의 h": my_stats.cohen_h(excess / n_dec_big, excess_shuf.mean() / n_dec_big)})
with pd.option_context("display.float_format", "{:,.3f}".format):
    display(crowd.to_frame("매장 몰림"))

# 지속 감소 가구가 보통보다 많이 넘친 매장 (참고, 상위 5개)
(pd.DataFrame({"주 매장": store_ids, "가구": np.bincount(store_code), "지속 감소": actual, "보통": expected})
   .assign(넘친_수=lambda d: d["지속 감소"] - d["보통"]).sort_values("넘친_수", ascending=False)
   .rename(columns={"넘친_수": "넘친 수"}).head(5).set_index("주 매장"))
````

'넘친 가구'는 매장마다 보통(섞은 1,000번의 평균)보다 많은 지속 감소 가구 수를 모두 더한 값. 실제와 섞은 결과의 차이가 '우연을 넘어 몰린 가구'.

- 손님이 10가구 이상인 100개 매장(2,298가구, 지속 감소 354가구)을 봄
- 실제로 넘친 가구는 75.1가구, 섞었을 때도 평균 67.4가구가 넘침. 우연을 넘어 몰린 가구는 7.7가구뿐
- 섞기 p-value 0.058로 우연과 구별되지 않고, h도 0.054로 작은 효과(0.2)에 한참 못 미침
- 가장 많이 넘친 매장도 30가구 가운데 9가구(보통 4.7가구) 정도. 지속 감소 가구가 특정 매장에 몰려 있지 않음. 근처에 새 가게가 생겨 한 매장 손님들이 함께 빠져나간 흔적은 보이지 않음

````python
# 대상 가구의 매장별 주별 구매액(18~101주)과 마트 전체 주별 구매액
xt = xq[xq["household_key"].isin(hh.index)]
weeks = range(18, 102)
store_week = xt.pivot_table(index="STORE_ID", columns="WEEK_NO", values="SALES_VALUE", aggfunc="sum",
                            fill_value=0).reindex(columns=weeks, fill_value=0)
mart_week = store_week.sum().to_numpy()

# 앞 12주 대비 뒤 12주 변화를 마트 전체와 견준 값, 경계 12~72번째 주(61자리)
pos = np.arange(12, 73)
def rel_change(s, m):
    cs, cm = np.r_[0, np.cumsum(s)], np.r_[0, np.cumsum(m)]
    with np.errstate(divide="ignore", invalid="ignore"):
        return ((cs[pos + 12] - cs[pos]) / (cs[pos] - cs[pos - 12])) / ((cm[pos + 12] - cm[pos]) / (cm[pos] - cm[pos - 12]))

# 한 번에 줄어든 가구마다: 같은 주 매장의 다른 손님 구매액(그 가구 몫을 뺀 값)
own_week = (xt[xt["household_key"].isin(jump.index)].groupby(["household_key", "STORE_ID", "WEEK_NO"])["SALES_VALUE"].sum())
together = {}
for key, row in jump.iterrows():
    st = main_store[key]
    own = own_week.get(key, pd.Series(dtype=float))
    own = own.xs(st, level="STORE_ID").reindex(weeks, fill_value=0) if st in own.index.get_level_values(0) else pd.Series(0, index=weeks)
    rel = rel_change(store_week.loc[st].to_numpy() - own.to_numpy(), mart_week)
    at = rel[4 * row["뚝 떨어진 구간"] - 12]
    # 앞 12주에 다른 손님 구매가 없어 값을 구할 수 없으면 순위를 비워 '따로'로 둠
    together[key] = {"매장 변화 % (마트 대비)": (at - 1) * 100,
                     "61자리 가운데 낮은 순위": int((rel <= at).sum()) if np.isfinite(at) else np.nan}
together = pd.DataFrame(together).T
jump = jump.join(together)
jump["매장"] = np.where(jump["61자리 가운데 낮은 순위"] <= 3, "함께", "따로")
pd.crosstab(jump["더 줄어든 쪽"], jump["매장"], margins=True, margins_name="합")
````

'함께'는 그 가구가 뚝 떨어진 때 같은 매장의 다른 손님들도 그 매장으로서는 드물게 크게 줄어든 경우.

- 한 번에 줄어든 88가구 가운데 '함께'는 4가구(4.5%), 나머지 84가구는 '따로'. 그 매장의 다른 손님 구매가 앞 12주에 없어 값을 구할 수 없는 1가구는 '따로'에 넣었음
- 기준이 61자리 가운데 가장 낮은 3자리라, 아무 일이 없어도 약 5%(3 ÷ 61)는 '함께'로 잡힘. 4.5%는 우연 수준
- 한 번에 줄어든 가구는 같은 매장 손님들과 상관없이 혼자 줄었음. 매장 근처의 변화보다 가구 자신의 사정 쪽

### 10.1 '한 번에' 판정 다시 하기 (질문 A)

8.1절은 가구마다 '서서히 줄었다면 나올 법한 가짜 가구' 200개를 만들고, 실제 가구의 계단 점수가 그보다 높으면 '한 번에 줄었다'고 봤음. 가짜 가구는 직선 둘레의 들쭉날쭉한 부분을 한 구간씩 낱개로 섞어 만들었음.

그런데 실제 구매액은 이웃한 구간끼리 비슷하게 움직일 수 있음. 예를 들어 손님이 많은 철이나 집에 일이 있는 몇 달은 연달아 많이 사고, 그 뒤 몇 달은 연달아 적게 삼. 낱개로 섞으면 이런 '연달아'가 깨져 가짜 가구는 실제보다 매끈해짐. 매끈한 가짜와 견주면 실제 가구는 조금만 덩어리져도 계단처럼 보여 '한 번에'로 너무 쉽게 잡힘.

그래서 두 단계로 확인함.

1. 이웃끼리 정말 같이 움직이는지 잼(이웃 상관). 가구마다 직선 둘레의 들쭉날쭉한 값이 바로 옆 구간과 얼마나 같이 움직이는지 구함. 0이면 상관없고, 0보다 크면 많이 산 구간 다음에 또 많이 사는 식
2. 같이 움직인다면 가짜 가구를 다시 만듦(묶어 섞기). 21구간을 3구간(12주)씩 7묶음으로 나누고 묶음의 순서만 섞음. 묶음 안의 순서는 그대로라 '연달아'가 남음. 묶음 길이 3구간은 8절에서 '낮아진 상태가 석 달은 이어져야 뚝 떨어졌다'고 정한 길이와 같음

결과를 보기 전에 정한 것:

- 이웃 상관이 0보다 큰 가구가 341가구의 절반보다 뚜렷하게 많으면(한쪽 부호 검정 p < 0.05) 묶어 섞기로 다시 판정하고, 1-2·2-3 모습 가구도 다시 셈. 아니면 8절 판정을 그대로 씀
- 다시 판정한 '한 번에' 가구가 8절과 달라도 바로 셀 수 있게, ③ 무엇이 줄었나와 ④ 같은 매장 손님은 341가구 모두에 미리 구해 둠(8.3·8.4절과 같은 계산)

````python
# 직선 둘레 오르내림의 이웃 상관(1차 자기상관), 5·6분기에도 산 지속 감소 341가구
y_live = adj.loc[live.index].to_numpy()
X_line = np.c_[np.ones(21), np.arange(21)]
r_c = y_live - (X_line @ np.linalg.lstsq(X_line, y_live.T, rcond=None)[0]).T
r_c = r_c - r_c.mean(axis=1, keepdims=True)
ac1 = (r_c[:, 1:] * r_c[:, :-1]).sum(axis=1) / (r_c ** 2).sum(axis=1)

# 한쪽 부호 검정: 0보다 큰 가구가 절반(0.5)보다 많은가
ac_pos = int((ac1 > 0).sum())
ac_p = my_stats.prop_vs_chance(ac_pos, len(ac1), 0.5).loc["greater", "p-value"]
use_block = ac_p < 0.05
print(f"이웃 상관 중앙값 {np.median(ac1):.3f}, 0보다 큰 가구 {ac_pos} / {len(ac1)} ({ac_pos / len(ac1) * 100:.1f}%), "
      f"한쪽 부호 검정 p-value {ac_p:.4f} → 묶어 섞기 {'씀' if use_block else '안 씀'}")
````

이웃 상관이 0보다 크면 많이 산 구간 다음에 또 많이 사는 식이라는 뜻.

- 341가구 가운데 191가구(56.0%)가 0보다 큼. 절반(50%)보다 많고, 우연과 구별됨(p-value 0.015)
- 다만 중앙값은 0.042로 작음. 이웃끼리 강하게 묶인 가구가 많은 것이 아니라, 약하게 같이 움직이는 가구가 조금 더 많은 정도
- 정한 규칙대로 묶어 섞기로 다시 판정함

### 10.3 1·2분기에만 많이 샀나 (3-2, 질문 C)

1·2분기에 평소보다 많이 산 가구는, 그 뒤 평소대로만 사도 줄어든 것처럼 보임. 예를 들어 늘 주 30달러를 사던 가구가 1·2분기에만 손님 치레로 주 40달러를 샀다면, 5·6분기에 원래대로 30달러를 사도 25% 줄어든 것으로 잡힘.

이 가설이 맞다면 지속 감소 가구는 1·2분기 구매액이 그 전보다 높았을 것. 분석 기간(18주부터) 앞의 1∼17주를 '그 전(앞 시기)'으로 씀.

- 재는 값: 1·2분기 주평균 ÷ 앞 시기 주평균. 1보다 크면 1·2분기에 그 전보다 많이 샀다는 뜻. 앞 시기는 그 가구의 첫 구매 주부터 17주까지
- 1∼17주는 데이터가 모이기 시작한 때라 가구가 차례로 나타남(1주 88가구 → 17주 1,371가구). 앞 시기가 너무 짧으면 값이 흔들려, 첫 구매가 9주 이전인 가구(앞 시기 9주 이상)만 봄
- 앞 시기를 산 기록이 있는 첫 주부터 세므로 앞 시기 주평균이 조금 높게 잡힘. 두 유형에 똑같이 들어가 비교에는 영향이 적음
- 판정 (결과 보기 전): 델타 규칙. 지속 감소 가구의 값이 뚜렷한 추세 없음 가구보다 크면 3-2 쪽

````python
# 데이터가 모이기 시작한 1~17주의 주별 가구 수 (전체 가구)
print("주별 가구 수:", df.groupby("WEEK_NO")["household_key"].nunique().loc[[1, 9, 17]].to_dict())

# 첫 구매 주, 앞 시기(첫 구매 주~17주) 주평균, 1·2분기(18~43주) 주평균
early = df[~excluded & df["household_key"].isin(hh.index) & df["WEEK_NO"].le(43)]
first_week = early.groupby("household_key")["WEEK_NO"].min()
pre_amt = early[early["WEEK_NO"].le(17)].groupby("household_key")["SALES_VALUE"].sum()
bump = hh[["유형", "크기 단계", "1·2분기"]].join(first_week.rename("첫 구매 주"))
bump = bump[bump["첫 구매 주"].le(9)].copy()
bump["앞 시기 주평균"] = pre_amt.reindex(bump.index, fill_value=0) / (18 - bump["첫 구매 주"])
bump["1·2분기 주평균"] = bump["1·2분기"] / 26
bump = bump[bump["앞 시기 주평균"].gt(0)]
bump["1·2분기 ÷ 앞 시기"] = bump["1·2분기 주평균"] / bump["앞 시기 주평균"]
bump["5·6분기 주평균"] = hh.loc[bump.index, "5·6분기"] / 26
bump["5·6분기 ÷ 앞 시기 (참고)"] = bump["5·6분기 주평균"] / bump["앞 시기 주평균"]
print("첫 구매가 9주 이전인 가구:", bump["유형"].value_counts().reindex(type_order).to_dict())

# 유형별 중앙값 (5·6분기는 참고: 1·2분기에서 줄어든 뒤 그 전 수준과 견주면 어디쯤인지)
with pd.option_context("display.float_format", "{:,.2f}".format):
    display(bump.groupby("유형")[["앞 시기 주평균", "1·2분기 주평균", "5·6분기 주평균", "1·2분기 ÷ 앞 시기",
                                "5·6분기 ÷ 앞 시기 (참고)"]].median().reindex(type_order))
````

값은 유형별 중앙값(달러, 배). 앞 시기는 1∼17주 가운데 그 가구의 첫 구매 주부터 잰 주평균.

- 1∼17주는 1주 88가구에서 17주 1,371가구로 가구가 차례로 나타남(위 출력). 첫 구매가 9주 이전인 가구는 지속 감소 168, 뚜렷한 추세 없음 672, 지속 증가 212가구
- 지속 감소 가구는 앞 시기 주 28.0달러, 1·2분기 33.2달러로, 1·2분기가 앞 시기의 1.10배(가구별 비율의 중앙값). 1·2분기에 조금 높았음
- 뚜렷한 추세 없음 가구는 0.96배로 거의 그대로
- 지속 증가 가구는 거꾸로 1·2분기가 앞 시기보다 낮았음(0.78배). 1·2분기가 높았던 가구가 감소로, 낮았던 가구가 증가로 잡히는 경향이 함께 보임
- 지속 감소 가구의 5·6분기는 앞 시기의 0.32배. 1·2분기가 높았던 만큼(1.10배)을 훨씬 넘어, 그 전 수준보다도 크게 내려갔음

````python
# 3-2 쪽(greater) 델타, 크기 단계별 델타와 판정
def delta_verdict(a, b, col, alt):
    sign = 1 if alt == "greater" else -1
    res = my_stats.cliffs_delta(a[col], b[col]).loc[alt]
    # 어느 한쪽에 가구가 없는 단계는 델타를 비워 두고 같은 방향으로 세지 않음
    tier_d = {t: (my_stats.cliffs_delta(a.loc[a["크기 단계"].eq(t), col], b.loc[b["크기 단계"].eq(t), col]).loc[alt, "delta"] * sign
                  if a["크기 단계"].eq(t).any() and b["크기 단계"].eq(t).any() else np.nan)
              for t in tiers}
    d, pv, same_dir = res["delta"] * sign, res["p-value"], sum(v > 0 for v in tier_d.values())
    verdict = ("뒷받침됨" if pv < 0.05 and d >= 0.147 and same_dir >= 4 else "약함" if pv < 0.05 and d > 0 else "뒷받침되지 않음")
    return {"델타 (가설 쪽 +)": d, "한쪽 p-value": pv, "같은 방향 단계 (5개 중)": same_dir, "판정": verdict, **tier_d}

v32 = delta_verdict(bump[bump["유형"].eq("지속 감소")], bump[bump["유형"].eq("뚜렷한 추세 없음")], "1·2분기 ÷ 앞 시기", "greater")
with pd.option_context("display.float_format", "{:,.3f}".format):
    display(pd.Series(v32, name="3-2 (1·2분기만 많았나)").to_frame().T)
````

- 1·2분기가 앞 시기보다 높았던 정도는 지속 감소 가구가 뚜렷한 추세 없음 가구보다 큼. 델타 0.169, p-value 0.001 아래, 다섯 단계 모두 같은 방향이라 판정 방법대로 3-2는 뒷받침됨
- 다만 3-2가 설명하는 몫은 작음. 1·2분기가 높았다가 제자리로만 돌아왔다면 5·6분기는 앞 시기 수준(1배)이어야 하는데, 실제로는 0.32배로 내려갔음. 지속 감소 가구는 대부분 그 전 수준보다도 크게 줄었음
- 지속 증가 가구에서 반대 모습(1·2분기가 낮았음)이 함께 보이는 것은, 추세 검정이 처음에 우연히 높았던 가구를 감소로 잡는 쪽으로 기울었을 수 있다는 뜻. 이 몫은 3-1(우연)과 겹침

### 10.4 서서히 줄어든 가구는 단골 품목을 통째로 끊었나 (가지 1과 가지 2 가르기, 질문 C)

서서히 줄어든 가구가 트리의 어느 큰 갈래인지 가름.

- 가지 1(장보는 양 자체가 줄었다): 덜 필요해져 모든 품목을 고르게 조금씩 덜 삼
- 가지 2(장보기 일부를 다른 곳에서 한다): 일부 품목은 통째로 다른 곳에서 사서, 이 마트에서는 그 품목을 아예 안 삼

다만 덜 오게 되면 어떤 품목은 저절로 빠짐. 그래서 '덜 와서 저절로 빠질 만큼'을 먼저 구하고, 실제로 그보다 많이 빠졌는지 봄. 예를 들어 1·2분기에 50번 와서 그 가운데 20번 우유를 산 가구가 5·6분기에 25번 왔다면, 장보기 습관이 그대로일 때 우유를 한 번도 안 살 일은 거의 없음. 그런데 우유를 아예 안 샀다면 우유는 다른 곳에서 산다는 뜻.

- 단골 품목: 1·2분기에 4번(장바구니 4개) 이상 산 품목(`COMMODITY_DESC`)
- 실제 끊긴 몫(%): 단골 품목 1·2분기 구매액 가운데, 5·6분기에 한 번도 사지 않은 품목의 몫
- 기대 끊긴 몫(%): 장보기 습관은 그대로이고 오는 횟수만 바뀌었을 때 빠질 몫. 1·2분기 장바구니 가운데 5·6분기 방문 수만큼을 무작위로 다시 뽑아(같은 장바구니를 또 뽑을 수 있음), 그 안에 한 번도 나오지 않은 단골 품목의 몫을 구함. 50번 뽑아 평균
- 재는 값: 실제 − 기대(%p). 0보다 크면 덜 와서 저절로 빠질 만큼보다 더 많이 빠졌다는 뜻
- 줄지 않은 가구도 입맛이나 계절 때문에 품목이 조금씩 바뀌므로, 서서히 줄어든 지속 감소 가구와 뚜렷한 추세 없음 가구를 견줌. 한 번에 줄어든 가구는 참고로 보여 줌
- 판정 (결과 보기 전): 델타 규칙. 지속 감소 가구의 값이 더 크면 가지 2 쪽. 뒷받침되지 않으면 가지 1에 가깝지만, 다른 곳에서 모든 품목을 조금씩 나눠 사는 경우와는 가를 수 없음

````python
# 견줄 가구: 서서히 줄어든 지속 감소, 한 번에 줄어든 지속 감소(참고), 뚜렷한 추세 없음(5·6분기에도 산 가구)
groups = pd.concat([
    pd.Series("지속 감소 · 서서히", index=live.index[live[shape_col].eq("서서히")]),
    pd.Series("지속 감소 · 한 번에 (참고)", index=live.index[live[shape_col].eq("한 번에")]),
    pd.Series("뚜렷한 추세 없음", index=hh.index[hh["유형"].eq("뚜렷한 추세 없음") & hh["5·6분기"].gt(0)])])
xc = xq[xq["분기"].isin([1, 2, 5, 6]) & xq["household_key"].isin(groups.index)]
b12 = xc[xc["분기"].le(2)]
n56 = xc[xc["분기"].ge(5)].groupby("household_key")["BASKET_ID"].nunique()
c56 = xc[xc["분기"].ge(5)].groupby("household_key")["COMMODITY_DESC"].agg(lambda s: set(s.dropna()))

# 가구마다 실제·기대 끊긴 몫 (helpers에 맞는 함수가 없어 numpy로 직접 뽑기)
rng = np.random.default_rng(3)
n_draw = 50
drop_rows = {}
for key, d in b12.groupby("household_key"):
    cnt = d.groupby("COMMODITY_DESC")["BASKET_ID"].nunique()
    regular = cnt.index[cnt >= 4]
    if len(regular) == 0 or key not in n56.index:
        continue
    spend = np.clip(d[d["COMMODITY_DESC"].isin(regular)].groupby("COMMODITY_DESC")["SALES_VALUE"].sum()
                    .reindex(regular).to_numpy(), 0, None)
    if spend.sum() <= 0:
        continue
    actual_drop = spend[~regular.isin(list(c56[key]))].sum() / spend.sum() * 100
    mat = pd.crosstab(d["BASKET_ID"], d["COMMODITY_DESC"]).reindex(columns=regular, fill_value=0).to_numpy() > 0
    present = mat[rng.integers(0, len(mat), size=(n_draw, n56[key]))].any(axis=1)
    expected_drop = ((~present) * spend).sum(axis=1).mean() / spend.sum() * 100
    drop_rows[key] = {"단골 품목 수": len(regular), "실제 끊긴 몫": actual_drop, "기대 끊긴 몫": expected_drop}

drop = pd.DataFrame(drop_rows).T.join(groups.rename("묶음")).join(hh["크기 단계"])
drop["실제 − 기대"] = drop["실제 끊긴 몫"] - drop["기대 끊긴 몫"]
group_order = ["지속 감소 · 서서히", "지속 감소 · 한 번에 (참고)", "뚜렷한 추세 없음"]
drop.groupby("묶음")[["단골 품목 수", "실제 끊긴 몫", "기대 끊긴 몫", "실제 − 기대"]].median().reindex(group_order).assign(
    가구=drop["묶음"].value_counts())
````

값은 묶음별 중앙값. 실제 끊긴 몫 18.0%는 단골 품목에 쓰던 돈 100달러 가운데 18달러어치 품목을 5·6분기에 한 번도 사지 않았다는 뜻. 기대 끊긴 몫은 습관 그대로 덜 오기만 했을 때 빠질 몫. '실제 − 기대'는 가구마다 뺀 값의 중앙값이라 두 중앙값의 차와 다름.

- 서서히 줄어든 지속 감소 260가구: 실제 18.0%, 기대 4.2%, 차이 9.5%p. 덜 와서 저절로 빠질 만큼보다 훨씬 많은 단골 품목을 아예 끊었음
- 뚜렷한 추세 없음 1,244가구: 실제 2.9%, 기대 0.3%, 차이 2.1%p. 줄지 않은 가구도 품목이 조금은 바뀜
- 한 번에 줄어든 지속 감소 65가구(참고): 실제 28.4%, 기대 8.8%, 차이 12.7%p
- 10.1절에서 서서히로 판정된 274가구와 한 번에 67가구 가운데, 단골 품목(4번 이상 산 품목)이 있는 가구만 들어감

````python
# 가지 2 쪽(greater): 서서히 줄어든 지속 감소 대 뚜렷한 추세 없음
v_split = delta_verdict(drop[drop["묶음"].eq("지속 감소 · 서서히")], drop[drop["묶음"].eq("뚜렷한 추세 없음")], "실제 − 기대", "greater")
v_split_jump = delta_verdict(drop[drop["묶음"].eq("지속 감소 · 한 번에 (참고)")], drop[drop["묶음"].eq("뚜렷한 추세 없음")], "실제 − 기대", "greater")
with pd.option_context("display.float_format", "{:,.3f}".format):
    display(pd.DataFrame({"서서히 (판정)": v_split, "한 번에 (참고)": v_split_jump}).T)
````

- 서서히 줄어든 가구: 델타 0.296, p-value 0.001 아래, 다섯 단계 모두 같은 방향이라 뒷받침됨. 모든 품목을 고르게 덜 산 것(가지 1)이 아니라, 일부 단골 품목을 통째로 끊은 것(가지 2)
- 이 마트가 그 품목을 안 팔게 된 것은 아님. 7.3절에서 사던 상품이 사라진 몫은 다른 가구와 같았음. 끊은 품목은 다른 곳에서 사게 됐다고 보는 것이 자연스러움
- 8.2절에서 생활용품이 식사 재료보다 더 줄었던 것(델타 0.125)과도 같은 방향. 다른 곳에서 사기 쉬운 품목부터 옮겨 간 모습
- 한 번에 줄어든 가구(참고)도 델타 0.337로 같은 모습
- 질문 C의 답(첫 번째 점): 서서히 줄어든 가구도 가지 2(장보기 일부를 다른 곳에서 한다)

### 10.5 한 번에 줄어든 가구는 다른 지점으로 옮겨 갔나 (2-3과 2-2 가르기, 질문 B)

2-3 모습 가구(어느 때부터 한 번에, 오는 횟수가 더 줄고, 같은 매장 손님들은 그대로)가 어디로 옮겼는지는 이 데이터로 볼 수 없음. 경쟁사 매장이나 온라인에서 산 기록은 남지 않기 때문. 그래서 옮겨 간 곳 대신, 뚝 떨어진 뒤에도 이 마트에 조금 남아 있는 구매가 어느 지점에서 일어났는지 봄.

- 이사한 가구(2-3)라면, 새 집 근처에 이 마트의 다른 지점이 있을 때 남은 장보기를 그 지점에서 할 것. 예를 들어 A동 지점에 다니던 가구가 B동으로 이사했다면, 그 뒤에는 B동 지점에서 가끔 사는 식
- 집은 그대로인데 경쟁사 매장이나 온라인으로 옮긴 가구(2-2)라면, 가끔 이 마트에 올 때도 늘 가던 원래 지점에 갈 것
- 그래서 이 확인은 한쪽으로만 답할 수 있음. 남은 구매가 다른 지점으로 옮겨 갔으면 이사(2-3)의 근거가 되지만, 원래 지점에 머물렀으면 2-2일 수도, 새 동네에 이 마트 지점이 없어 경쟁사로 간 2-3일 수도 있어 가를 수 없음

- 재는 값: 뚝 떨어진 뒤 가장 많이 산 매장이 1·2분기 주 매장과 다른 가구의 비율. 뚝 떨어진 뒤 구매가 없는 가구는 뺌
- 기준 비율: 줄지 않은 가구도 가끔 다니는 매장이 바뀜. 그래서 뚜렷한 추세 없음 가구 가운데 5·6분기에 가장 많이 산 매장이 1·2분기 주 매장과 다른 가구의 비율과 견줌
- 판정 (결과 보기 전): 가구 수 규칙. 기준 비율보다 많고(한쪽 이항 검정 p < 0.05) 코언의 h 0.2 이상이면 뒷받침됨 → 2-3 모습 가구 가운데 적어도 일부는 이사 등으로 다른 지점 근처로 옮겨 간 것. 뒷받침되지 않으면 원래 지점에 남아 덜 온 것이라, 위에서 밝힌 대로 2-2와 2-3을 가를 수 없음
- 이사를 더 직접 보여 줄 인구통계(주거 형태 등)는 801가구에만 있고 세입자는 53가구뿐이라, 2-3 모습 가구 가운데 기록이 있는 가구가 너무 적어 쓰지 않음

### 10.6 지속 감소 판정 다시 하기 (이웃 상관 보정, 질문 A·C)

10.1∼10.5절을 마친 뒤 더한 확인. 10.1절에서 본 '이웃 구간끼리 같이 움직이는 성질'은 '한 번에' 판정뿐 아니라, 5절에서 지속 감소 가구를 고른 맨-켄달 검정에도 영향을 줌. 이 분석의 결론은 모두 그 370가구에서 출발하므로, 출발점이 흔들리는지 확인함.

맨-켄달 검정은 21구간의 값이 서로 상관없다고 가정함. 이웃끼리 같이 움직이면, 앞쪽 몇 달 연달아 많이 사고 뒤쪽 몇 달 연달아 적게 산 것만으로도 꾸준히 줄어든 것처럼 보여 지속 감소가 많게 잡힘. 그래서 이웃 상관을 반영한 맨-켄달 검정(Hamed & Rao, 1998)으로 다시 가름.

- 생각: 동전을 21번 던져 앞면이 15번 나온 것과, 7번 던진 결과를 세 번씩 베껴 적어 21칸 가운데 15칸이 앞면인 것은 믿을 정도가 다름. 21구간이 이웃끼리 묶여 있으면 사실상 21개보다 적은 정보라, 같은 기울기라도 우연일 가능성을 더 크게 봐야 함
- 계산: 가구마다 추세를 뺀 값으로 시차별 이웃 상관(바로 옆, 두 칸 옆 … 구간끼리의 상관)을 구하고, 우연과 구별되는 상관만 모아 검정의 우연 폭을 넓힘(보정 배수). 추세는 센 기울기(구간을 둘씩 짝지어 구한 기울기들의 중앙값)로 잡아 튀는 값에 덜 흔들리게 하고, 이웃 상관은 값 대신 순위(가장 작은 값 1, 그다음 2 …)로 구해 맨-켄달 검정처럼 값의 크기가 아니라 순서만 씀. 보정 배수가 1보다 작으면(이웃끼리 반대로 움직임) 1로 둠. 원래 검정보다 쉽게 잡히는 일이 없게 하기 위함
- 5절과 같은 값(계절분을 뺀 4주 구간 구매액 21구간), 같은 유의수준 5%, 감소·증가 한쪽 검정을 따로 함
- 5절 검정은 표본이 작을 때 정확한 분포를 쓰고 이 검정은 정규 근사를 써서, 보정이 없어도 판정이 몇 가구 다를 수 있음. 그래서 보정 전(정규 근사) 결과도 함께 보여 줌

결과를 보기 전에 정한 것:

1. 두 검정 모두 지속 감소인 가구를 '확실한 지속 감소', 두 검정 모두 뚜렷한 추세 없음인 가구를 비교 상대로 둠
2. 6·7·8.2·10.1·10.3·10.4·10.5절의 지표를 이 두 묶음으로, 같은 판정 규칙에 따라 다시 판정함. 판정이 같으면 결론이 보정에 흔들리지 않는 것이고, 달라지면 그 지표는 '보정에 따라 달라짐'으로 적음
3. 8.4절 매장 몰림은 두 판정 모두 '뒷받침되지 않음'이었고, 지속 감소 가구가 줄면 몰림을 찾기 더 어려워지므로 다시 하지 않음

````python
# 이웃 상관을 반영한 맨-켄달 검정 (Hamed & Rao, 1998), 5절과 같은 값·유의수준
hr = my_ts.trend_by_row(adj, alpha=0.05, labels=("지속 감소", "뚜렷한 추세 없음", "지속 증가"), autocorr=True)
hr = hr.rename(columns={"유형": "보정 뒤"})

# 보정 전(정규 근사) 판정: 보정 배수를 되돌린 검정 통계량으로 구함
from scipy.stats import norm
z_raw = norm.ppf(hr["감소 p-value"]) * np.sqrt(hr["보정 배수"])
hr["보정 전 (정규 근사)"] = np.select([z_raw < norm.ppf(0.05), z_raw > norm.ppf(0.95)], ["지속 감소", "지속 증가"],
                                    default="뚜렷한 추세 없음")

# 5절 유형별 보정 배수 중앙값과 가구 수
print("5절 유형별 보정 배수 중앙값:", hr["보정 배수"].groupby(hh["유형"]).median().reindex(type_order).round(2).to_dict())
pd.concat({"보정 전 (정규 근사)": pd.crosstab(hh["유형"], hr["보정 전 (정규 근사)"]),
           "보정 뒤": pd.crosstab(hh["유형"], hr["보정 뒤"])}, axis=1).reindex(index=type_order).rename_axis("5절 유형")
````

보정 배수는 이웃 상관 때문에 검정을 얼마나 엄격하게 했는지를 나타냄. 1이면 그대로이고, 2면 우연의 폭을 두 배로 넓혀 봄. 표는 5절 유형(행)이 다시 한 검정에서 어떻게 나뉘는지 보여 줌.

- 세 유형 모두 보정 배수의 중앙값이 1.0. 대부분 가구는 우연과 구별되는 이웃 상관이 잡히지 않아 보정이 걸리지 않았음
- 보정 없이 정규 근사만 쓰면 지속 감소 370가구 가운데 366가구가 그대로. 4가구는 계산 방식(정확한 분포 대 정규 근사) 차이
- 보정 뒤에는 350가구가 지속 감소로 남고 20가구가 뚜렷한 추세 없음으로 바뀜. 지속 증가는 500가구 가운데 461가구가 남음. 뚜렷한 추세 없음 가구는 모두 그대로
- 다만 21구간으로는 이웃 상관이 꽤 커야(0.43 이상) 우연과 구별됨. 10.1절에서 본 것처럼 약하게 같이 움직이는 가구(중앙값 0.042)는 이 검정으로도 보정되지 않아, 우연히 걸린 가구(3-1)가 모두 걸러진 것은 아님

````python
# 확실한 지속 감소(두 검정 모두 지속 감소)와 비교 상대(두 검정 모두 뚜렷한 추세 없음)
sure_dec = hh.index[hh["유형"].eq("지속 감소") & hr["보정 뒤"].eq("지속 감소")]
sure_flat = hh.index[hh["유형"].eq("뚜렷한 추세 없음") & hr["보정 뒤"].eq("뚜렷한 추세 없음")]
print(f"확실한 지속 감소 {len(sure_dec)}가구 (5·6분기 구매 없음 {int(hh.loc[sure_dec, '5·6분기'].eq(0).sum())}가구), 비교 상대 {len(sure_flat)}가구")

# 델타 규칙 지표를 두 묶음으로 다시 판정
def rejudge(table, col, alt, dec_keys=sure_dec):
    t = table.dropna(subset=[col])
    r = delta_verdict(t[t.index.isin(dec_keys)], t[t.index.isin(sure_flat)], col, alt)
    return {"가구 (감소)": int(t.index.isin(dec_keys).sum()), "델타·h": r["델타 (가설 쪽 +)"], "한쪽 p-value": r["한쪽 p-value"],
            "같은 방향 단계": r["같은 방향 단계 (5개 중)"], "보정 뒤 판정": r["판정"]}

rows = {}
for col, alt in expect.items():
    rows[f"6절 1-1 {col}"] = {**rejudge(change, col, alt), "원래 판정": cond.loc[col, "판정"]}
for col, alt in expect2.items():
    rows[f"7절 2-1 {col}"] = {**rejudge(exposure, col, alt), "원래 판정": cond2.loc[col, "판정"]}
for name, alt in expect3.items():
    rows[f"8.2절 {name}"] = {**rejudge(share, "변화", alt), "원래 판정": compare3.loc[name, "판정"]}
rows["10.3절 3-2 1·2분기만 많았나"] = {**rejudge(bump, "1·2분기 ÷ 앞 시기", "greater"), "원래 판정": v32["판정"]}
slow_keys = drop.index[drop["묶음"].eq("지속 감소 · 서서히")]
rows["10.4절 서서히: 단골 품목이 통째로 빠졌나"] = {**rejudge(drop, "실제 − 기대", "greater", sure_dec.intersection(slow_keys)), "원래 판정": v_split["판정"]}
````

````python
# 가구 수 규칙 지표: 확실한 지속 감소 가운데 5·6분기에도 산 가구로 다시 셈 (10.1절 묶어 섞기 판정)
sure_live = live.index.intersection(sure_dec)
jump_s = live.loc[sure_live, shape_col].eq("한 번에")
cond23 = jump_s & allj.loc[sure_live, "더 줄어든 쪽"].eq("오는 횟수") & allj.loc[sure_live, "매장"].eq("따로")
cond12 = jump_s & allj.loc[sure_live, "더 줄어든 쪽"].eq("담는 양") & allj.loc[sure_live, "매장"].eq("따로")
for name, k, orig in [("10.1절 ① 한 번에 줄어든 가구", int(jump_s.sum()), recheck.loc["① 한 번에 줄어든 가구", "판정"]),
                      ("10.1절 1-2 모습", int(cond12.sum()), recheck.loc["1-2 모습 (한 번에 + 담는 양 + 따로)", "판정"]),
                      ("10.1절 2-3 모습", int(cond23.sum()), recheck.loc["2-3 모습 (한 번에 + 오는 횟수 + 따로)", "판정"])]:
    r = vs_chance(k, len(sure_live))
    rows[name] = {"가구 (감소)": len(sure_live), "델타·h": r["코언의 h"], "한쪽 p-value": r["한쪽 p-value"],
                  "같은 방향 단계": np.nan, "보정 뒤 판정": count_verdict(r["한쪽 p-value"], r["코언의 h"]), "원래 판정": orig,
                  "해당 가구": k}

# 10.5절: 2-3 모습 가운데 다른 지점으로 옮긴 가구, 기준은 비교 상대 가구
moved_s = moved[moved.index.isin(sure_dec)]
base_s = s56[s56.index.isin(sure_flat)].ne(main_store.reindex(s56[s56.index.isin(sure_flat)].index)).mean()
r = vs_chance(int(moved_s.sum()), len(moved_s), chance=base_s)
rows["10.5절 2-3 모습이 다른 지점으로 옮겼나"] = {"가구 (감소)": len(moved_s), "델타·h": r["코언의 h"], "한쪽 p-value": r["한쪽 p-value"],
                                         "같은 방향 단계": np.nan, "보정 뒤 판정": count_verdict(r["한쪽 p-value"], r["코언의 h"]),
                                         "원래 판정": moved_test["판정"].iloc[0], "해당 가구": int(moved_s.sum())}

robust = pd.DataFrame(rows).T[["가구 (감소)", "해당 가구", "델타·h", "한쪽 p-value", "같은 방향 단계", "원래 판정", "보정 뒤 판정"]]
num_cols = ["가구 (감소)", "해당 가구", "델타·h", "한쪽 p-value", "같은 방향 단계"]
robust[num_cols] = robust[num_cols].astype(float)
robust["같음"] = np.where(robust["원래 판정"].eq(robust["보정 뒤 판정"]), "예", "아니오")
with pd.option_context("display.float_format", "{:,.3f}".format, "display.max_colwidth", 40):
    display(robust)
````

확실한 지속 감소 350가구(이 가운데 5·6분기에 오지 않은 가구 25가구)와 비교 상대 1,535가구로 지금까지의 지표를 다시 판정함. '델타·h' 열은 두 묶음을 견주는 지표에서는 클리프의 델타, 가구 수를 세는 지표에서는 코언의 h.

### 10.7 계속 써야 하는 생활용품도 통째로 끊었나 (10.4절 보충)

#### 결론이 흔들리지 않는지

- '한 번에' 판정을 실제 가구와 더 닮은 가짜 가구로 엄격하게 다시 해도(10.1절) 판정은 같고, 2-3 모습 가구 수만 72가구에서 54가구로 줄었음
- 지속 감소 판정 자체를 이웃 상관을 반영한 검정으로 다시 해도(10.6절) 370가구 가운데 350가구가 남았고, 다시 판정한 지표 14개가 모두 원래와 같음
- 다만 21구간으로는 약한 이웃 상관까지 걸러 내지 못해, 우연히 지속 감소로 걸린 가구(3-1)가 남아 있을 수 있음
- 10.4절의 '단골 품목을 통째로 끊음'은 조금 크게 나왔을 수 있음. 지속 감소 가구는 1·2분기에 평소보다 조금 더 샀고(10.3절, 1.10배) 단골 품목도 많아(23.5개 대 14.0개), 그때만 산 품목이 단골로 잡혔다가 빠진 몫이 섞였을 수 있음. 다만 크기 단계 다섯 개 모두 같은 방향이라(10.4절) 결론은 유지됨
