# 03. 해결 방안

01 노트북에서 찾은 원인과 02 노트북에서 본 지난 캠페인의 효과를 묶어 분석 목표(01 노트북 3절)마다 누구에게, 무엇을, 언제 할지 CRM 방안을 세움. 01 노트북 4절 분석 설계의 3단계(해결 방안)에 해당함. 이 데이터는 지난 기록이므로 방안의 효과를 직접 증명할 수는 없음. 효과는 4단계(성과 확인)에서 실행했을 때 확인할 방법으로 정함.

> 이 노트북은 교육 과정 강사님의 라이브러리(hossam)를 바탕으로 직접 만든 분석 모듈(`helpers` 패키지의 `my_*`)을 사용함.
> 코드 작성·디버깅과 문장 교정에 생성형 AI(Claude Code)를 보조 도구로 사용함.

## 1. 원인 탐색 결론 (01 노트북 11절)

> 지속 감소 가구(370가구)는 장보기의 일부를 다른 곳으로 옮겼을 가능성이 큼. 다른 곳에서 산 기록이 없어 직접 확인한 것은 아님. 단순히 장보기를 줄였다고 볼 때보다 옮겼다고 볼 때 더 잘 들어맞는 흔적을 근거로 한 결론임. 여기서 '다른 곳'은 경쟁사 매장이나 온라인처럼 이 데이터에 기록이 남지 않는 곳을 뜻함. 장보기가 덜 필요해진 흔적은 약하고, 이 마트의 값·상품·행사가 이 가구들에게 더 나빠지거나 이 가구들이 그 변화에 더 크게 반응한 흔적도 없음. 왜 다른 곳이 이 마트보다 나아졌는지(다른 곳이 좋아짐, 가구 사정이 바뀜, 기록에 없는 이 마트의 변화)는 이 데이터로 구분할 수 없음.

줄어든 방식에 따라 가구를 세 묶음으로 나눔. 서서히·한 번에는 5·6분기에도 이 마트에 온 341가구를 구분한 결과임(01 노트북 10.1절).

| 묶음 | 가구 | 모습 | 근거 (01 노트북) |
|---|---|---|---|
| 서서히 감소 | 274가구 | 84주에 걸쳐 서서히 줄었고, 그사이 단골 품목 여럿을 통째로 안 사게 됨. 휴지·세제 같은 생활용품이 식사 재료보다 더 줄어듦 | 오는 횟수가 준 것만으로 저절로 안 사게 될 몫(단골 품목의 4.2%)보다 훨씬 많은 단골 품목(18.0%)을 아예 안 삼(10.4절). 줄기 전 크기가 비슷한 가구와 견줘도 같음(델타 0.324). 생활용품이 더 줄어듦(8.2절) |
| 한 번에 감소 | 67가구 | 어느 때부터 구매가 계단처럼 한 번에 줄어듦. 대부분 오는 횟수가 크게 줄었음. 이사로 매장이 멀어졌거나, 경쟁사 매장·온라인으로 한꺼번에 옮긴 모습 | 계단 모양으로 줄어든 가구가 우연보다 네 배쯤 많음(10.1절). 이사와 한꺼번에 옮김은 가를 수 없음(10.5절) |
| 끊김 | 29가구 | 5·6분기에 한 번도 오지 않음 | 왜 떠났는지는 가를 수 없음 |

**주된 원인으로 보기 어려운 것**

- 이 마트의 값·상품·행사: 지속 감소 가구가 사던 상품도 다른 가구가 사던 상품과 똑같이 바뀌었고(7.3∼7.5절), 값이 오르거나 행사에서 빠진 상품을 다른 가구보다 더 줄이지도 않았음(7.7절). 품질·서비스처럼 기록에 없는 변화는 확인하지 못함
- 장보기가 덜 필요해짐(주된 원인이 아님): 싼 상품으로 바꾼 흔적은 아주 작고(6절), 외식이 늘어난 흔적은 없으며(8.2절), 식구가 줄어든 가구는 있더라도 드묾(8.3절, 10.1절)
- 특정 매장 근처의 변화: 지속 감소 가구가 특정 매장에 몰리지 않았고, 같은 매장 손님들이 함께 줄지도 않았음(8.4절)

**확인하지 못한 것**

- 왜 다른 곳을 고르게 됐는지. 다른 곳의 값·거리·온라인 편의는 기록이 없음
- 한 번에 줄어든 가구가 이사한 것인지, 경쟁사 매장·온라인으로 옮긴 것인지
- 이 마트의 품질·서비스(신선도, 일시 품절, 혼잡) 변화
- 기록에 없는 가구 사이의 차이(가구원 수, 소득, 매장까지 거리 등). 01 노트북은 줄기 전 구매액·방문 수만 맞춰 비교했음(01 노트북 5.1절)
- 370가구 가운데 일부(75∼117가구 이상)는 실제로 줄지 않았는데 우연히 지속 감소로 잡혔을 수 있음(10.2절)

## 2. 원인에서 해결 방안으로

원인 탐색 결과별로 해결 방안의 방향을 정리함. 쿠폰, 묶음 할인, 배송 같은 구체적인 수단은 아직 정하지 않고 방향만 적음.

원인 결론이 방안의 무엇을 정했는지 보이도록 결론이 달랐다면 방향이 어떻게 달라졌을지를 오른쪽 열에 함께 적음. 오른쪽 열은 데이터로 확인한 결과가 아니라 논리상의 가정임.

| 원인 탐색 결과 | 해결 방안의 방향 | 다른 결론이었다면 (가정) |
|---|---|---|
| 장보기 자체가 줄었다는 흔적은 약함 (가지 1을 주된 원인으로 보기 어려움) | 이 가구들은 줄어든 만큼을 다른 곳에서 사고 있을 가능성이 큼. 그래서 소비를 늘리게 하는 것보다, 다른 곳에서 사고 있을 몫을 이 마트로 되돌리는 것이 방향. 예를 들어 가구 A가 휴지를 이제 경쟁사에서 산다면, 휴지를 다시 이 마트에서 사게 하는 것 | 장보기 양 자체가 줄었다면 끊긴 품목 가운데 상당수는 필요가 줄거나 끝난 품목일 수 있어, 그 품목을 다시 사게 하는 제안과 '되찾을 몫'이라는 기준의 근거가 약해짐. 이 가구들에 비용을 쓸 이유도 약해짐 |
| 이 마트의 값·상품·행사가 이 가구들에게 더 나빠지지도, 더 크게 작용하지도 않았음 (2-1 뒷받침되지 않음) | 지금 근거로는 마트 전체의 값을 내리거나 행사를 늘리는 방식보다, 지속 감소 가구만 골라 그 가구에 맞게 다가가는 방식(CRM)을 먼저 검토함. 마트 전체 방식은 모든 손님에게 혜택이 돌아가 비용이 크지만, 지속 감소 가구는 손님의 15%뿐이라 대부분의 돈이 원래 잘 오던 손님에게 쓰임. 마트 전체 방식이 효과가 없다고 확인한 것은 아님 | 이 마트의 값·상품·행사가 이 가구들에게 더 나빠졌다면, 가구를 골라 다가가는 것과 함께 그 값·상품 구성을 손보는 방안도 검토해야 함 |
| 돈을 아낀 흔적은 약함 (1-1, 여러 번 검정을 보정하면 뒷받침되지 않음) | 큰 할인을 앞세울 근거는 없음. 쿠폰의 할인 폭 자체는 데이터로 정할 수 없음 | 돈을 아끼려는 흔적이 강했다면 자체 상표나 할인 상품을 앞세운 제안을 검토 |
| 서서히 줄어든 가구는 단골 품목을 통째로 끊었음 | 가구마다 끊긴 단골 품목이 곧 되찾을 품목. 예를 들어 가구 A가 1·2분기에 장바구니 네 번 넘게 담던 세제를 5·6분기에 한 번도 사지 않았다면, 가구 A에게서 먼저 되찾을 품목은 세제 | 모든 품목을 고르게 조금씩 덜 샀다면 '끊긴 품목'이 없어, 품목을 골라 제안하는 대신 오는 횟수 자체를 늘리는 제안 |
| 생활용품이 식사 재료보다 더 줄었음 | 휴지·세제는 어디서 사도 같은 물건이라, 값이 싸거나 사기 편한 곳으로 옮기기 쉬움. 그래서 값이나 편의를 앞세운 혜택이 맞을 수 있는 품목. 다만 01 노트북 7.7절에서 이 가구들은 이 마트의 값 변화에 다른 가구보다 더 민감하지 않았으므로, 값보다 편의 쪽이 맞을 수도 있음. 왜 옮겼는지는 확인하지 못해 가정임 | 식사 재료가 더 줄었다면(외식·배달, 1-3) 장보기 품목보다 간편식처럼 집밥을 덜 하는 가구에 맞춘 제안 |
| 한 번에 줄어든 가구는 이사인지 한꺼번에 옮긴 것인지 가를 수 없음 | 이사라면 혜택으로 거리를 이기기 어렵고, 한꺼번에 옮긴 것이라면 되찾을 여지가 있음. 그래서 비용이 적은 방식으로 먼저 한 번 다가가고, 반응한 가구에만 더 씀. 반응 여부는 참고가 되지만, 반응이 없다고 이사라고 볼 수는 없음 | 모든 가구가 서서히 줄었다면 단계를 나눠 다가갈 이유가 줄어듦 |
| 끊긴 가구는 왜 떠났는지 모름 | 다시 한 번 올 계기만 적은 비용으로 줌 | 떠난 이유가 데이터에 없어 다른 결론을 가정하기 어려움 |
| 지난 캠페인은 원래 많이 사던 가구에 몰려 나감 (00 노트북 1.7절) | 지난 캠페인이 효과가 있었는지, 줄어드는 가구에도 통했는지는 02 노트북에서 받기 전 구매가 비슷한 가구끼리 견줘 따로 추정하고, 그 착안점을 수단을 정할 때 더함. 다만 추정치라, 실행한 뒤의 효과는 새로 나누는 실험으로 확인함(4단계) | - |

- 정리하면, 대상(지속 감소 370가구)은 00 노트북이 찾았고, 01 노트북의 결론은 줄어든 모양에 따른 묶음(서서히·한 번에·끊김)과 되찾을 품목(끊긴 단골 품목), 그리고 마트 전체가 아니라 가구를 골라 다가갈 근거를 정함. 큰 할인을 앞세울 근거가 없다는 것도 01에서 나옴(1-1, 7.7절). 어떤 수단으로 전할지(쿠폰이 든 안내)는 02 노트북의 지난 캠페인 효과로 정함(6절). 묶음은 원인이 아니라 줄어든 모양으로 나눈 것이라, 묶음별 방안은 원인별이 아니라 감소 양상별 대응임
- 원인에 맞춘 제안(끊긴 품목별 쿠폰)이 모두에게 같은 쿠폰보다 나은지는 아직 확인되지 않음. 02에서 가구마다 고른 쿠폰(A형)은 효과가 남지 않았기 때문. 7절에서 실험할 가설로 둠

## 3. 해결 방안 설계 (초안)

| 목표 (01 노트북 3절) | 대상 | 방향 | 언제 | 우선순위 | 성과 지표 |
|---|---|---|---|---|---|
| 1-가. 서서히 줄어든 가구를 되돌림 | 서서히 줄어든 274가구 | 끊긴 단골 품목을 다시 이 마트에서 사게 함. 생활용품이 끊겼으면 생활용품부터 | 지금. 5·6분기에 단골 품목이 끊긴 것이 확인됨 | 되찾을 몫이 큰 가구부터 | 끊긴 단골 품목을 다시 산 가구 비율, 4주당 방문 수 |
| 1-나. 한 번에 줄어든 가구를 되돌림 | 한 번에 줄어든 67가구 | 비용이 적은 방식으로 한 번 다시 오게 하고, 반응한 가구에만 1-가와 같은 방향으로 이어감 | 지금. 구매가 계단처럼 떨어진 것이 확인됨 | 1-가 다음 | 다가간 뒤 4주 안에 다시 온 가구 비율. 반응이 없으면 더 쓰지 않음(반응이 없다는 것만으로 이사인지는 알 수 없음) |
| 2. 끊긴 가구를 다시 오게 함 | 5·6분기에 끊긴 29가구 | 다시 한 번 올 계기를 적은 비용으로 줌 | 지금. 5·6분기에 한 번도 오지 않은 것이 확인됨 | 줄기 전 구매액이 큰 가구부터 | 다시 온 가구 비율 |

- 되찾을 몫: 가구가 1·2분기에 끊긴 단골 품목에 쓴 금액. 예를 들어 가구 A가 1·2분기에 세제·휴지에 26주 동안 60달러를 썼는데 5·6분기에 둘 다 끊었다면 되찾을 몫은 60달러. 지난 구매액으로 잰 값이므로 방안이 통했을 때 실제로 돌아올 금액을 뜻하지는 않음
- 표의 '방향'은 무엇을 이루려는지만 정한 것임. 어떤 수단(쿠폰, 묶음 할인, 배송 등)으로 이룰지는 02 노트북의 착안점을 더해 6절에서 정함

#### 확인 순서

| 절 | 확인하는 것 | 이어지는 목표 |
|---|---|---|
| 4 | 준비: 01 노트북의 묶음(서서히 274 / 한 번에 67 / 끊김 29)을 다시 만듦 | 전체 |
| 5 | 묶음마다 되찾을 몫이 얼마이고, 어떤 품목이 끊겼나 (가구별 되찾을 품목 목록) | 1-가, 1-나, 2 |
| 6 | 목표별 CRM 시나리오 정리 (누구에게, 무엇을, 언제, 대상 가구 수, 되찾을 몫). 02 노트북의 착안점으로 방향마다 수단을 정함 | 전체 |
| 7 | 효과 확인 설계: 무작위 비교 집단, 성과 지표, 이 가구 수로 가려낼 수 있는 효과 크기 | 전체 |

- 각 절의 판단 기준은 그 절에서 결과를 보기 전에 적음
- 실제 실행과 효과 확인(01 노트북 4절의 4단계)은 이 노트북 밖의 일. 7절은 그 방법만 정함

> **참고 (5.1)** 되찾을 몫 가운데 생활용품은 1.6∼3.3%뿐이었음. 그래서 표의 '생활용품이 끊겼으면 생활용품부터'는 우선순위가 아니라 다가갈 때 앞세울 품목의 후보로만 쓰고, 가구의 우선순위는 5.1절의 금액 기준을 따름.

## 4. 준비

## 5. 묶음마다 되찾을 몫과 끊긴 품목

3절의 우선순위 기준인 '되찾을 몫'을 가구마다 구함. 용어는 01 노트북 10.4절과 같음.

- 단골 품목: 1·2분기에 4번(장바구니 4개) 이상 산 품목(`COMMODITY_DESC`)
- 끊긴 단골 품목: 단골 품목 가운데 5·6분기 26주 동안 한 번도 사지 않은 품목
- 되찾을 몫: 끊긴 단골 품목에 1·2분기 26주 동안 쓴 금액(달러). 지난 구매액으로 잰 값이라, 방안이 통했을 때 실제로 돌아올 금액을 뜻하지는 않음
- 생활용품: 01 노트북 8.2절에서 정한 10개 품목(세제, 휴지·키친타월, 비닐·호일, 비누 등). 3절에서 생활용품이 끊겼으면 생활용품부터 되찾기로 해서, 되찾을 몫 가운데 생활용품 몫을 따로 봄

예를 들어 가구 A가 1·2분기에 세제를 장바구니 6개에 담아 30달러, 우유를 10개에 담아 40달러, 과자를 2개에 담아 5달러를 썼다면 단골 품목은 세제·우유(과자는 4개 미만). 5·6분기에 우유만 샀다면 끊긴 단골 품목은 세제이고, 되찾을 몫은 30달러, 그 가운데 생활용품 몫은 100%.

이 절에서는 판정하지 않고 규모를 잼. 결과를 보기 전에 각 값의 쓰임을 다음과 같이 정함.

- 묶음 안의 우선순위: 되찾을 몫이 큰 가구부터(3절)
- 묶음 사이의 순서: 3절 표 그대로(서서히 → 한 번에 → 끊김). 되찾을 몫의 크기로 바꾸지 않음. 묶음마다 다가가는 방식과 비용이 달라 금액만으로 견줄 수 없기 때문
- 끊김 29가구는 5·6분기에 산 것이 없어 단골 품목이 모두 끊긴 것으로 잡힘. 그래서 이 묶음의 되찾을 몫은 '줄기 전 단골 품목 구매액 전체'라는 뜻

```python
# 생활용품 10개 품목 (01 노트북 8.2절)
home = ["LAUNDRY DETERGENTS", "LAUNDRY ADDITIVES", "DISHWASH DETERGENTS", "HOUSEHOLD CLEANG NEEDS", "BATH TISSUES",
        "PAPER TOWELS", "PAPER HOUSEWARES", "FD WRAPS/BAGS/TRSH BG", "DISPOSIBLE FOILWARE", "SOAP - LIQUID & BAR"]

# 지속 감소 가구의 1·2분기·5·6분기 거래
xg = xq[xq["household_key"].isin(groups.index) & xq["분기"].isin([1, 2, 5, 6])]
b12 = xg[xg["분기"].le(2)]
bought56 = xg[xg["분기"].ge(5)].groupby("household_key")["COMMODITY_DESC"].agg(lambda s: set(s.dropna()))

# 가구·품목별 1·2분기 장바구니 수와 구매액 → 단골 품목만 (helpers에 맞는 함수가 없어 pandas로 직접 집계)
item = b12.groupby(["household_key", "COMMODITY_DESC"]).agg(장바구니=("BASKET_ID", "nunique"), 구매액=("SALES_VALUE", "sum"))
reg = item[item["장바구니"].ge(4)].reset_index()
reg["끊김"] = [c not in bought56.get(k, set()) for k, c in zip(reg["household_key"], reg["COMMODITY_DESC"])]
reg["생활용품"] = reg["COMMODITY_DESC"].isin(home)

# 가구마다 단골 품목 수, 끊긴 단골 품목 수, 되찾을 몫, 그 가운데 생활용품 몫
cut = reg[reg["끊김"]]
per_hh = pd.DataFrame({
    "단골 품목 수": reg.groupby("household_key").size(),
    "끊긴 단골 품목 수": cut.groupby("household_key").size(),
    "되찾을 몫": cut.groupby("household_key")["구매액"].sum(),
    "생활용품 되찾을 몫": cut[cut["생활용품"]].groupby("household_key")["구매액"].sum(),
}).reindex(groups.index).fillna(0).join(groups)
per_hh.head()
```

```python
# 묶음별 규모: 끊긴 단골 품목이 있는 가구, 되찾을 몫(중앙값·합계), 생활용품 몫
g = per_hh.groupby("묶음")
size = pd.DataFrame({
    "가구": g.size(),
    "단골 품목 있는 가구": g["단골 품목 수"].apply(lambda s: (s > 0).sum()),
    "끊긴 품목 있는 가구": g["끊긴 단골 품목 수"].apply(lambda s: (s > 0).sum()),
    "끊긴 품목 수 (중앙값)": g["끊긴 단골 품목 수"].median(),
    "되찾을 몫 중앙값 ($)": g["되찾을 몫"].median(),
    "되찾을 몫 합계 ($)": g["되찾을 몫"].sum(),
    "생활용품 몫 %": g["생활용품 되찾을 몫"].sum() / g["되찾을 몫"].sum() * 100,
}).reindex(group_order)
size["끊긴 품목 있는 가구 %"] = size["끊긴 품목 있는 가구"] / size["가구"] * 100
size
```

값은 묶음별. 되찾을 몫은 1·2분기 26주 동안의 금액(달러).

- 끊긴 단골 품목이 있는 가구: 서서히 줄어든 가구 243가구(88.7%), 한 번에 줄어든 가구 61가구(91.0%). 두 묶음 모두 대부분의 가구에 되찾을 품목이 있음
- 가구당 되찾을 몫(중앙값)은 한 번에 줄어든 가구가 169.0달러로 서서히 줄어든 가구(81.9달러)의 두 배쯤. 한 번에 줄어든 가구는 끊긴 품목도 더 많음(중앙값 8개 대 4개)
- 묶음 전체의 되찾을 몫은 서서히 42,655.6달러, 한 번에 19,539.2달러, 끊김 5,352.4달러
- 끊김 29가구 가운데 단골 품목이 있는 가구는 18가구뿐. 나머지 11가구는 1·2분기에 장바구니 4개 이상 산 품목이 하나도 없으므로 이 기준으로는 되찾을 품목이 잡히지 않음. 가구당 되찾을 몫도 21.9달러로 작음
- 되찾을 몫 가운데 생활용품 몫은 세 묶음 모두 1.6∼3.3%. 01 노트북 8.2절에서 생활용품이 식사 재료보다 '비율로' 더 줄었지만, 금액으로는 되찾을 몫의 작은 부분

```python
# 서서히 줄어든 가구에서 자주 끊긴 단골 품목: 끊은 가구 수와 되찾을 몫 합계 (상위 10개)
slow_cut = cut[cut["household_key"].isin(groups.index[groups.eq("서서히 감소")])]
top_items = (slow_cut.groupby("COMMODITY_DESC")
             .agg(끊은_가구=("household_key", "nunique"), 되찾을_몫=("구매액", "sum"), 생활용품=("생활용품", "first"))
             .sort_values("되찾을_몫", ascending=False).head(10))
top_items.columns = ["끊은 가구", "되찾을 몫 합계 ($)", "생활용품"]
top_items
```

- 금액이 큰 품목은 소고기, 맥주, 냉동 육류·피자, 탄산음료, 델리 육류처럼 장보기에서 자주 사는 품목이 많음
- 분유(INFANT FORMULA), 기저귀(DIAPERS & DISPOSABLES)도 상위에 있음. 아이가 자라 더는 필요 없어진 것일 수 있으므로 이런 품목은 다른 곳으로 옮겨 간 몫이 아니라 필요가 끝난 몫일 수 있음. 이 데이터로는 둘을 가를 수 없음
- 담배(CIGARETTES), 선불 휴대폰 충전(PREPAID WIRELESS&ACCESSORIES)은 장보기 품목과 성격이 달라, 되찾을 품목으로 쓸지 따로 정해야 함

### 5.1 CRM 우선순위용 되찾을 몫

5절의 품목 목록을 보고 더한 확인. 끊긴 품목 가운데 이 마트로 되돌린다는 방향(2절)과 맞지 않는 품목이 섞여 있으므로 우선순위를 정할 때 쓸 금액을 따로 만듦. 5절의 값은 그대로 둠.

- 빼는 품목: 담배(CIGARETTES, CIGARS, TOBACCO OTHER)와 선불 휴대폰 충전(PREPAID WIRELESS&ACCESSORIES). 장보기와 성격이 다르고, 쿠폰·안내로 되돌릴 장보기 몫으로 보기 어려움
- 넣되 따로 보는 품목: 아기용품(INFANT FORMULA, DIAPERS & DISPOSABLES, BABY FOODS, BABYFOOD, BABY HBC, INFANT CARE PRODUCTS). 아이가 자라 필요가 끝났을 가능성이 있지만 이 데이터로는 확인할 수 없음. 짐작으로 빼지 않고 뺐을 때 순위가 얼마나 바뀌는지만 확인함
- 판단 기준 (결과 보기 전): 묶음마다 '아기용품 포함'과 '아기용품 제외' 금액으로 가구 순위를 매겨, 상위 절반에 드는 가구가 90% 이상 같으면 아기용품을 넣든 빼든 우선순위는 사실상 같다고 봄. 그보다 적게 겹치면 두 순위를 모두 6절로 넘김
- 끊김 29가구 가운데 단골 품목이 없는 11가구는 '단골 품목 기준으로는 되찾을 몫을 잴 수 없는 가구'로 따로 표시함. 다가갈 가치가 없다는 뜻은 아님

```python
# 우선순위용 금액: 담배·선불 충전 제외(기본), 아기용품까지 제외(확인용)
non_grocery = ["CIGARETTES", "CIGARS", "TOBACCO OTHER", "PREPAID WIRELESS&ACCESSORIES"]
baby = ["INFANT FORMULA", "DIAPERS & DISPOSABLES", "BABY FOODS", "BABYFOOD", "BABY HBC", "INFANT CARE PRODUCTS"]
cut_g = cut[~cut["COMMODITY_DESC"].isin(non_grocery)]
per_hh["우선순위용 몫"] = cut_g.groupby("household_key")["구매액"].sum().reindex(per_hh.index).fillna(0)
per_hh["아기용품 제외 몫"] = (cut_g[~cut_g["COMMODITY_DESC"].isin(baby)].groupby("household_key")["구매액"].sum()
                          .reindex(per_hh.index).fillna(0))
per_hh["단골 품목 없음"] = per_hh["단골 품목 수"].eq(0)

# 묶음마다 상위 절반이 얼마나 겹치나 (helpers에 맞는 함수가 없어 pandas로 직접 계산)
def top_half(s):
    return set(s.sort_values(ascending=False).index[: (len(s) + 1) // 2])

rows = {}
for name, d in per_hh[~per_hh["단골 품목 없음"]].groupby("묶음"):
    a, b = top_half(d["우선순위용 몫"]), top_half(d["아기용품 제외 몫"])
    rows[name] = {"가구 (단골 품목 있음)": len(d),
                  "우선순위용 몫 합계 ($)": d["우선순위용 몫"].sum(),
                  "빠진 담배·충전 ($)": (d["되찾을 몫"] - d["우선순위용 몫"]).sum(),
                  "아기용품 몫 ($)": (d["우선순위용 몫"] - d["아기용품 제외 몫"]).sum(),
                  "상위 절반 겹침 %": len(a & b) / len(a) * 100}
check = pd.DataFrame(rows).T.reindex(group_order)
check["단골 품목 없는 가구"] = per_hh.groupby("묶음")["단골 품목 없음"].sum().reindex(group_order)
check
```

- 담배·선불 충전을 빼면 우선순위용 몫은 서서히 40,717.3달러, 한 번에 18,970.7달러, 끊김 5,184.5달러. 빠진 금액은 묶음마다 3∼5%
- 서서히 줄어든 가구는 아기용품을 넣든 빼든 상위 절반이 98.5% 같음 → 판단 기준대로 우선순위는 사실상 같다고 봄. 6절은 우선순위용 몫 하나로 씀
- 한 번에 줄어든 가구는 84.8%만 같음(65가구 가운데 상위 절반 33가구 중 5가구가 바뀜) → 기준(90%)에 못 미쳐, 6절에는 두 순위를 함께 넘김. 이 차이는 아기용품 몫(2,356.6달러)이 이 묶음의 우선순위용 몫에서 12%쯤을 차지하기 때문에 생김
- 끊김 가구는 단골 품목이 있는 18가구만 순위를 매길 수 있고, 나머지 11가구는 금액이 아닌 다른 기준으로 다가가야 함(6절)
- 단골 품목이 없는 가구는 서서히 14가구, 한 번에 2가구도 있음. 이 가구들도 되찾을 몫은 0으로 잡혀 순위의 맨 뒤에 놓임

02 노트북에서 가져온 착안점은 네 가지임.

- 효과가 있었던 B형·C형은 효과의 3분의 2 이상이 쿠폰을 받고 쓰지 않은 가구에서 나왔음(02 노트북 8.1절). 쿠폰 할인 금액보다 쿠폰이 든 발송물을 받은 것이 효과와 이어졌을 가능성(외부 연구도 같은 방향, 02 노트북 8.1절) → 쿠폰이 든 안내를 첫 수단으로 검토(쿠폰의 할인 폭은 데이터에 없어 정할 근거가 없음). 쿠폰 없이 안내만 보내도 같은지는 02의 캠페인이 모두 쿠폰을 담고 있어 알 수 없음
- 가구마다 이전 구매로 고른 쿠폰(A형)은 끝난 뒤 효과가 남지 않았음(02 노트북 7절) → 끊긴 품목 쿠폰 하나에 기대지 않고 안내와 함께 씀
- 줄어들던 가구의 오는 횟수를 늘린 것은 C형뿐(4주당 +1.34번, 약함, 02 노트북 9절). C형이 무엇이 달랐는지는 데이터에 없어 그대로 고를 수 없음 → 실험에서 확인할 가설
- B형은 줄어들던 가구가 올 때 더 담게 했음(02 노트북 9절) → 아직 오고 있는 가구가 끊긴 품목을 다시 이 마트에서 사게 하는 쪽에 씀. 다만 B형 효과는 받기 전에 적게·중간쯤 사던 가구(4주당 구매액 약 223달러 아래)에서만 나왔음(02 노트북 7.3절). 지속 감소 가구로는 줄기 전 4주당 구매액으로 약 70%가 이 범위. 223달러는 02의 캠페인 직전 12주 기준이라 줄기 전(1·2분기)과 시점이 달라 대략의 경계로만 씀

먼저 묶음마다 '먼저 다가갈 가구'를 몇 가구로 잡을지 정하려고 우선순위용 몫이 가구 사이에 얼마나 몰려 있는지 확인함. 예를 들어 상위 절반 가구가 몫의 80%를 차지하면, 절반에게만 먼저 다가가도 되찾을 몫 대부분을 대상으로 삼을 수 있음.

```python
# 묶음마다 우선순위용 몫이 상위 가구에 얼마나 몰렸나 (단골 품목이 있는 가구만)
def concentration(s):
    s = s.sort_values(ascending=False)
    total = s.sum()
    return pd.Series({f"상위 {q}% 가구의 몫 %": s.iloc[: max(1, round(len(s) * q / 100))].sum() / total * 100 for q in (20, 50)})

ranked = per_hh[~per_hh["단골 품목 없음"]]
conc = ranked.groupby("묶음")["우선순위용 몫"].apply(concentration).unstack().reindex(group_order)
conc.insert(0, "가구 (단골 품목 있음)", ranked.groupby("묶음").size().reindex(group_order))
conc
```

- 따라서 비용을 줄여야 할 때 상위 절반에게 먼저 다가가도 되찾을 몫 대부분을 대상으로 삼을 수 있음. 아래 시나리오의 '먼저 다가갈 가구'는 이 순서를 따름

  - 4주당 구매액: 그대로 견줄 때와, 다가가기 전 구매액을 함께 넣어 견줄 때(공변량 보정). 보정하면 가구마다 원래 많이 사던 몫을 걸러낼 수 있어 흩어짐이 줄어듦. 줄어드는 정도는 지난 자료로 잼(3·4분기 구매액으로 5·6분기 구매액을 설명할 때 남는 흩어짐). 다가가기 전 값만 씀
  - 끊긴 단골 품목을 하나라도 다시 산 가구 비율: 아무것도 하지 않아도 일부는 다시 사므로, 그 비율(기준 비율)을 지난 자료로 어림함. 1·2분기 단골 품목이 3·4분기에 끊겼던 가구 가운데 5·6분기에 그 품목을 다시 산 가구 비율. 관측 자료로 잡은 가정값이므로 이번 대상과 다를 수 있고, 26주 동안 본 값이므로 12주 실험의 기준 비율보다 높게 잡혔을 수 있음

  - 재구매 비율: 견줄 만한 지난 효과가 없어 가를 수 있는지는 판정하지 않고, 기준 비율에서 얼마나 올라야 가를 수 있는지만 적음

> **나중에 덧붙임 (02 노트북 7.3절 반영):** 02 노트북 7.3절에서 B형 효과가 받기 전 구매액에 따라 달랐음(적게·중간쯤 사던 가구에서만 늘었음). 그래서 실행할 때는 서서히·한 번에 줄어든 가구를 줄기 전 4주당 구매액 약 223달러를 기준으로 두 층으로 나누고, 층 안에서 무작위로 다가갈 쪽과 비교 쪽을 나눔(층화 무작위 배정). 두 층이 양쪽에 고르게 들어가 층 차이가 결과에 섞이지 않음. 기본 비교는 '쿠폰이 든 안내 대 아무것도 하지 않음'. 쿠폰 없이 안내만 보내는 갈래를 더하면 안내만으로도 되는지 알 수 있지만, 갈래가 늘수록 갈래마다 가구가 줄어 지금 가구 수로는 어려움. 그래서 이 갈래는 실험을 되풀이하거나 가구를 늘릴 때 더할 가설로 둠. 같은 이유로, 원인 분석이 방안에 쓸모 있었는지를 가장 직접 보여 줄 비교인 '끊긴 품목 쿠폰 + 안내(원인에 맞춘 제안)' 대 '모두에게 같은 쿠폰 묶음(02의 B형 방식)'도 첫 실험에는 넣지 않음. 비교 쪽까지 세 갈래로 나누면 서서히 줄어든 274가구도 갈래마다 약 90가구라 차이를 가리기 더 어려움. 첫 실험은 '원인에 맞춘 쿠폰 안내 대 아무것도 하지 않음'으로 하고, 맞춤 대 공통은 가구를 늘리거나 실험을 되풀이할 때 확인할 가설로 둠. 층마다 효과를 보는 것은 미리 정해 둔 부분 분석으로 하되, 층마다 가구가 적어 아래 최소 검출 효과(묶음 전체 기준)보다 더 큰 효과만 가를 수 있으므로 탐색적으로만 해석함. 223달러는 02의 캠페인 직전 12주 기준이라 줄기 전(1·2분기)과 시점이 다른 대략의 경계

```python
# 4주당 구매액·방문 수: 3·4분기(보정에 쓸 '다가가기 전' 값의 대용)와 5·6분기 (26주 = 4주 × 6.5)
def per4(quarters):
    q = xq[xq["분기"].isin(quarters) & xq["household_key"].isin(groups.index)]
    return pd.DataFrame({"구매액": q.groupby("household_key")["SALES_VALUE"].sum() / 6.5,
                         "방문 수": q.groupby("household_key")["BASKET_ID"].nunique() / 6.5}).reindex(groups.index).fillna(0)
p34, p56 = per4([3, 4]), per4([5, 6])

# 최소 검출 효과 (helpers에 검정력 계산 함수가 없어 statsmodels로 직접 계산)
from statsmodels.stats.power import TTestIndPower, NormalIndPower
rows = {}
for name in ["서서히 감소", "한 번에 감소"]:
    keys = groups.index[groups.eq(name)]
    d_min = TTestIndPower().solve_power(nobs1=len(keys) // 2, alpha=0.05, power=0.8, ratio=(len(keys) - len(keys) // 2) / (len(keys) // 2))
    for col in ["구매액", "방문 수"]:
        y, x = p56.loc[keys, col], p34.loc[keys, col]
        rho = np.corrcoef(x, y)[0, 1]
        mde = d_min * y.std()
        rows[(name, col)] = {"가구": len(keys), "평균": y.mean(), "표준편차": y.std(),
                             "최소 검출 효과": mde, "평균 대비 %": mde / y.mean() * 100,
                             "전후 상관": rho, "보정 뒤 평균 대비 %": mde * np.sqrt(1 - rho ** 2) / y.mean() * 100}
with pd.option_context("display.float_format", "{:,.2f}".format):
    display(pd.DataFrame(rows).T)
```

값의 단위는 4주당 달러와 4주당 방문 수. '전후 상관'은 3·4분기 값과 5·6분기 값의 상관.

- 그대로 견주면, 서서히 줄어든 274가구는 4주당 구매액이 평균(78.11달러)보다 38.7% 넘게 늘어야 가를 수 있음. 한 번에 줄어든 67가구는 91.1%
- 가구마다 앞뒤 구매액이 강하게 이어져 있으므로(상관 0.84∼0.90), 다가가기 전 값으로 보정하면 문턱이 절반쯤으로 내려감. 서서히 20.9%, 한 번에 45.2%
- 그래도 02 노트북의 B형 효과(+4.5∼6.4%)보다 서서히는 세 배 이상(20.9%), 한 번에는 일곱 배 이상(45.2%) 큼 → 판단 기준대로, 보정해도 이 묶음들의 가구 수만으로는 B형만 한 효과를 한 번의 실험으로 가를 수 없음
- B형 효과 +4.5∼6.4%는 받기 전에 적게·중간쯤 사던 가구에서 나온 효과가 섞인 평균(02 노트북 7.3절). 많이 사던 가구에서는 효과가 없었으므로, 그런 가구가 많은 묶음에서는 B형만 한 효과도 기대하기 어려울 수 있음

```python
# 재구매 비율의 기준 비율: 1·2분기 단골 품목이 3·4분기에 끊겼던 가구 가운데 5·6분기에 그 품목을 다시 산 가구 비율
bought = {qs: xq[xq["분기"].isin(qs) & xq["household_key"].isin(groups.index)]
             .groupby("household_key")["COMMODITY_DESC"].agg(lambda s: set(s.dropna())) for qs in [(3, 4), (5, 6)]}
reg2 = reg.copy()
reg2["3·4분기 끊김"] = [c not in bought[(3, 4)].get(k, set()) for k, c in zip(reg2["household_key"], reg2["COMMODITY_DESC"])]
reg2["5·6분기 다시 삼"] = [c in bought[(5, 6)].get(k, set()) for k, c in zip(reg2["household_key"], reg2["COMMODITY_DESC"])]
cut34 = reg2[reg2["3·4분기 끊김"]]
back = cut34.groupby("household_key")["5·6분기 다시 삼"].any()

rows = {}
for name in ["서서히 감소", "한 번에 감소"]:
    b = back[back.index.isin(groups.index[groups.eq(name)])]
    p0 = b.mean()
    n = int((per_hh.loc[per_hh["묶음"].eq(name), "끊긴 단골 품목 수"] > 0).sum())   # 실험 대상: 지금 끊긴 품목이 있는 가구
    h = NormalIndPower().solve_power(nobs1=n // 2, alpha=0.05, power=0.8, ratio=(n - n // 2) / (n // 2))
    p1 = np.sin(np.arcsin(np.sqrt(p0)) + h / 2) ** 2          # 코언의 h를 비율로 되돌림
    rows[name] = {"기준 비율을 잰 가구": len(b), "기준 비율 %": p0 * 100, "실험 대상 가구": n,
                  "가를 수 있는 비율 %": p1 * 100, "최소 검출 차이 %p": (p1 - p0) * 100,
                  "기준 비율 대비 %": (p1 - p0) / p0 * 100}
pd.DataFrame(rows).T
```

- 끊긴 단골 품목을 아무것도 하지 않아도 다시 산 가구가 많음. 3·4분기에 끊겼던 가구 가운데 5·6분기에 하나라도 다시 산 가구가 서서히 69.4%, 한 번에 78.6%. 그래서 이 지표는 원래 높은 데서 시작함
- 지금 끊긴 품목이 있는 가구(서서히 243, 한 번에 61)를 반씩 나누면, 다가간 쪽의 비율이 서서히 84.4%(+15.0%p), 한 번에 98.5%(+19.9%p)까지 올라야 가를 수 있음. 한 번에 줄어든 가구는 거의 모든 가구가 다시 사야 하는 수준이므로 현실적인 문턱이 아님
- 기준 비율 대비로는 서서히 21.6%, 한 번에 25.4%

#### 주 지표

| 묶음 | 구매액 (보정 뒤) | 재구매 비율 | 주 지표 (판단 기준대로 더 작은 쪽) |
|---|---|---|---|
| 서서히 감소 | 20.9% | 21.6% | 4주당 구매액 (다가가기 전 구매액으로 보정). 다만 두 지표의 문턱이 거의 같아 재구매 비율을 보조 지표로 함께 봄 |
| 한 번에 감소 | 45.2% | 25.4% | 끊긴 단골 품목을 다시 산 가구 비율. 다만 문턱(98.5%)이 거의 모든 가구라 이 지표로도 가르기 어려움 |

- 어느 지표를 고르든 한 번의 실험으로 작은 효과를 가르기는 어려움. 그래서 실행할 때는 같은 방식의 실험을 몇 차례 되풀이해 합치거나, 다른 매장·다른 시기의 지속 감소 가구를 더해 가구 수를 늘려야 함

```python
# 끊김 29가구: 피셔 정확 검정으로 본 문턱과 검정력 (다가간 쪽 14, 비교 쪽 15, 비교 쪽에서는 아무도 안 온다고 가정)
from scipy.stats import fisher_exact, binom
n_t, n_c = 14, 15
k_min = next(k for k in range(n_t + 1) if fisher_exact([[k, n_t - k], [0, n_c]])[1] < 0.05)
p80 = next(p / 100 for p in range(1, 101) if binom.sf(k_min - 1, n_t, p / 100) >= 0.8)
print(f"다가간 {n_t}가구 가운데 {k_min}가구 이상 다시 와야 비교 쪽(0가구)과 가를 수 있음 (피셔 정확 검정, 양측 5%)")
print(f"그 문턱을 80% 확률로 넘으려면 다가간 가구의 실제 재방문 비율이 {p80 * 100:.0f}% 이상이어야 함")
```

- 비교 쪽에서 아무도 다시 오지 않는다는, 비교 쪽에 가장 유리한 가정을 해도 다가간 14가구 가운데 4가구 이상이 다시 와야 가를 수 있음
- 그 문턱을 80% 확률로 넘으려면 다가간 가구의 실제 재방문 비율이 36% 이상이어야 함. 비교 쪽에서도 몇 가구가 저절로 돌아온다면 문턱은 더 높아짐
- 29가구로는 작은 재방문 효과를 안정적으로 검증하기 어려움. 이 묶음은 다가간 뒤 다시 온 가구를 세는 기록으로 봄
